## Introduction

This project provides a Python implementation of the analytical framework presented in Hou et al. (2025). The core of this repository is the iPython Notebook `multi_agent_system_architecture_for_federal_funds_target_rate_prediction_draft.ipynb`, which contains a comprehensive suite of functions to replicate the paper's findings. The pipeline is designed to simulate the deliberative institutional process of the Federal Open Market Committee (FOMC), moving beyond black-box forecasting to provide interpretable, reasoning-based predictions.

The paper addresses the challenge of forecasting monetary policy in a complex economic environment where decisions reflect diverse philosophies and regional concerns. This codebase operationalizes the paper's framework, allowing users to:
-   Rigorously validate and manage the entire experimental configuration via a single `config.yaml` file.
-   Cleanse and normalize structured macroeconomic data (e.g., inflation, employment) and unstructured narratives (e.g., Beige Book, Dot Plots).
-   Construct canonical "meeting packets" that strictly enforce information cutoffs to prevent look-ahead bias.
-   Simulate FOMC deliberations using a multi-agent system (MAS) with distinct roles (Analyst, Economist, Members).
-   Implement advanced reasoning protocols including Chain-of-Draft (CoD) and In-Context Learning (ICL).
-   Evaluate performance using rigorous metrics such as Total Accuracy, Agent Accuracy, Voting Stability, and Semantic Similarity.

## Theoretical Background

The implemented methods combine techniques from Multi-Agent Systems, Natural Language Processing, and Financial Economics.

**1. Multi-Agent System (MAS) Simulation:**
The FOMC decision-making process is modeled as a collaborative workflow among heterogeneous agents.
-   **Analyst:** Interprets market signals (Fed Funds Futures).
-   **Economist:** Formulates policy options based on macro data.
-   **Members:** Deliberate and vote based on their specific archetypes (Regional Pragmatists, Academic Balancers, Central Policymakers).

**2. Chain-of-Draft (CoD) Reasoning:**
To improve reasoning quality and reduce hallucinations, member agents operate under a CoD protocol. This enforces concise, multi-stage reasoning steps (e.g., $\le 30$ words per step) followed by a revision phase, ensuring that the final vote is grounded in a logical progression.

**3. In-Context Learning (ICL):**
Agents are "trained" by simulating historical meetings (e.g., 2019, 2022). They vote, receive the ground truth outcome, and generate "lessons learned" which are stored in a long-term memory vector store. These lessons are retrieved and injected into the context during inference to guide future decisions.

**4. Unsupervised Clustering of Archetypes:**
Historical FOMC participants are clustered using K-Means based on features like hawkishness, regional affiliation, and policy focus. The resulting centroids define the "System Personas" adopted by the simulated member agents.

# Summary

### Executive Overview

This paper introduces **FedSight AI**, a novel Multi-Agent System (MAS) designed to forecast the Federal Open Market Committee (FOMC) Federal Funds Target Rate (FFTR). Departing from traditional econometric models (Taylor Rule, VARs) and black-box machine learning approaches (Random Forests, LSTMs), the authors model interest rate decisions as **emergent outcomes of institutional deliberation**.

By leveraging Large Language Models (LLMs) to simulate the reasoning, debate, and voting processes of heterogeneous agents, the system integrates both structured macroeconomic indicators and unstructured qualitative data (e.g., the Beige Book). The proposed **Chain-of-Draft (CoD)** extension achieves state-of-the-art performance, outperforming existing baselines in accuracy, stability, and interpretability.

### Methodological Architecture

The framework is built on the premise that policy decisions are collective reasoning processes, not static statistical functions. The architecture is implemented using **CrewAI** with **GPT-4o** as the underlying engine.

#### A. Data Integration (The Input Layer)
The model ingests a hybrid dataset covering the period **February 2023 – December 2024**:
*   **Structured Inputs:** Standard macroeconomic variables including Inflation (PCE, CPI), Monetary Indicators (Treasury yields, M2), Economic Activity (GDP, Unemployment), and Political Environment.
*   **Unstructured Inputs:**
    *   **The Beige Book:** Anecdotal evidence on regional economic conditions.
    *   **The Dot Plot:** Policymakers' forward-looking rate expectations.
    *   **FedWatch:** Market-implied probabilities (used for context).

#### B. Agent Design and Clustering
To avoid generic simulations, the authors utilized a clustering algorithm on historical FOMC participants to derive three distinct agent archetypes, ensuring behavioral fidelity without computational intractability:
1.  **Regional Pragmatists:** Focus on local district data.
2.  **Academic Balancers:** Weigh theoretical economic frameworks.
3.  **Central Policymakers:** Focus on consensus and national stability.

#### C. The Deliberative Workflow
The simulation proceeds through a structured sequential process:
1.  **Analyst Agent:** Interprets Fed Funds Futures to establish market benchmarks.
2.  **Economist Agent:** Formulates three policy options (Dovish, Neutral, Hawkish) with macro rationales.
3.  **Member Agents (The Core):** Independently analyze inputs, deliberate on options, exchange perspectives to mimic FOMC debate, and cast votes.
4.  **Output:** A consolidated policy statement and rate decision.

### Algorithmic Enhancements

The authors propose two variations to the baseline FedSight AI:

*   **FedSight ICL (In-Context Learning):** Agents are "fine-tuned" via simulation-based memory. They practice on historical meetings (e.g., 2019, 2022), receive feedback on the true outcomes, and store these "lessons" in long-term memory to refine future reasoning.
*   **FedSight CoD (Chain-of-Draft):** This is the superior model. It modifies Chain-of-Thought prompting by enforcing **concise, multi-stage reasoning**. Agents produce minimal-step drafts ($\leq$ 30 words per step) followed by revisions. This mechanism significantly reduces token usage while improving logical coherence.

### Experimental Results

The model was backtested on **16 FOMC meetings** (2023–2024), a period characterized by high volatility (hikes, cuts, and holds). Each meeting was simulated five times to account for LLM stochasticity.

#### A. Quantitative Performance
**FedSight CoD** achieved the highest performance across all metrics:
*   **Total Accuracy:** **93.75%** (correctly predicted 15/16 outcomes).
*   **Voting Stability:** **93.33%** (consistency of agent votes across simulation runs).
*   **Agent Agreement:** **90.22%** (internal consensus).
*   **Mean Absolute Error (MAE):** **0.0156** (significantly lower than baselines).

#### B. Computational Efficiency
The CoD mechanism reduced computational overhead significantly:
*   **Average Tokens per Run:** ~60,000 for CoD vs. ~81,000 for ICL.

#### C. Benchmark Comparison
*   **vs. MiniFed (Seok et al., 2024):** On the 2018 test set, MiniFed achieved 75% accuracy; FedSight CoD achieved **100%**.
*   **vs. Ordinal Random Forest:** FedSight CoD outperformed the RF model (62.5% accuracy) on the 2023–2024 dataset.
*   **vs. Linear Regression:** A baseline LR model achieved only 31.25% directional accuracy, highlighting the non-linear complexity of the task.

#### Interpretability
The system generates reasoning chains that align semantically with actual FOMC statements (measured via cosine similarity), offering transparency that black-box models lack.

### Critical Assessment and Limitations

While the results are robust, the following constraints must be noted from an econometric perspective:

1.  **Sample Size:** The test set is limited to 16 meetings due to the sparsity of FOMC events. While the results are statistically promising (indicated by high stability), the sample is small for standard ML generalization claims.
2.  **LLM Dependency:** The system relies on GPT-4o. Changes in the model's behavior, versioning, or "hallucinations" could impact reproducibility.
3.  **Data Leakage Concerns:** Since LLMs are trained on vast corpora, they may have "seen" historical outcomes. The authors mitigate this by emphasizing **institutional reasoning**—forcing agents to derive decisions from contemporaneous data rather than memory—but the risk cannot be entirely zeroed out.

### Conclusion

**FedSight AI** represents a significant advancement in financial macro-forecasting. By treating the Federal Funds Rate as the result of a **multi-agent deliberative process** rather than a pure time-series problem, the authors achieve superior predictive accuracy and transparency. The **Chain-of-Draft (CoD)** mechanism effectively balances reasoning depth with computational efficiency, establishing a new benchmark for Generative AI applications in finance.

*   **Research Role:** Enforces the **Data Integrity** requirements described in Section 3.1 ("Data Description"). It ensures that the "Structured Inputs" and "Unstructured Inputs" align perfectly with the 16 scheduled meetings from 2023–2024.

### **3. `cleanse_macro_data` (Task 3 Orchestrator)**

*   **Inputs:** `df_macro_raw` (Validated DataFrame).
*   **Processes:**
    1.  Calls `coerce_macro_dtypes` to standardize numeric types and handle artifacts.
    2.  Calls `impute_missing_values` to apply deterministic forward-fill.
    3.  Calls `flag_macro_outliers` to detect unit errors and range violations.
*   **Outputs:** `df_macro_final` (Cleaned DataFrame), logs, and flags.
*   **Data Transformation:** String representations of numbers (e.g., "5.42%") are converted to floats. Missing values are filled using the value from the prior meeting ($t-1$).
*   **Research Role:** Prepares the **Structured Inputs** ($X_{struct}$) defined in Section 3.1. It ensures that variables like "PCE Index" and "Treasury Yields" are numerical and complete, ready for agent consumption.

### **4. `cleanse_unstructured_data` (Task 4 Orchestrator)**

*   **Inputs:** `df_unstructured_raw` (Validated DataFrame).
*   **Processes:**
    1.  Calls `cleanse_narrative_text` to strip HTML and normalize whitespace.
    2.  Calls `normalize_fedwatch_probs` to ensure probabilities sum to 1.0.
    3.  Calls `flag_missing_dot_plots` to identify meetings requiring imputation.
*   **Outputs:** `df_unstructured_clean` (Cleaned DataFrame), logs.
*   **Data Transformation:** Raw HTML-laden text is transformed into clean ASCII text. Probability distributions are renormalized: $p_i \leftarrow p_i / \sum p_j$.
*   **Research Role:** Prepares the **Unstructured Inputs** ($X_{unstruct}$) defined in Section 3.1. It ensures that the "Beige Book" and "FedWatch" data are in a format suitable for LLM context injection.

### **5. `cleanse_evaluation_data` (Task 5 Orchestrator)**

*   **Inputs:** `df_participants_raw`, `df_meeting_outcomes`, `df_fomc_statements_actual`.

    2.  Calls `cleanse_meeting_outcomes` to validate arithmetic consistency of rate changes.
    3.  Calls `cleanse_actual_statements` to normalize text for embedding.
*   **Outputs:** Cleaned DataFrames for participants, outcomes, and statements.
*   **Data Transformation:** Participant features are imputed and standardized. Outcome basis points are coerced to integers. Statement text is normalized.
*   **Research Role:** Prepares the **Ground Truth** data ($y_i$, $S_{actual}$) used for evaluation metrics in Section 3.3 ("Performance Metrics").

### **6. `construct_canonical_datasets` (Task 6 Orchestrator)**

*   **Research Role:** Implements the **Information Set Construction** step. It ensures that agents only receive data available "two days before each meeting" (Section 3.1), strictly separating $X_t$ (inputs) from $y_t$ (outcomes).

*   **Outputs:** Updated `agent_packets` with `unstructured_context` and `structured_macro_text`.
*   **Data Transformation:** Individual text fields are concatenated into a single prompt-ready string. Numeric macro data is formatted into a key-value text block (e.g., "CPI: 3.2%").

*   **Research Role:** Implements the **Representative Agents** derivation described in Section 3.2. It uses unsupervised learning to identify the "Regional Pragmatists," "Academic Balancers," and "Central Policymakers."

*   **Data Transformation:** Unstructured text ("Option A... Action: -25 bps") is parsed into a structured mapping, and the qualitative decision is converted to a quantitative forecast $\hat{y}_{i,j}$.
*   **Research Role:** Bridges the gap between **Qualitative Deliberation** and **Quantitative Forecasting** (Section 3.1), enabling the calculation of numerical error metrics.

*   **Research Role:** Calculates the **Semantic Similarity** metric (Section 3.3), evaluating the interpretability and alignment of generated text.

````python
# Task 1 — Validate ingested configuration dictionary integrity and completeness

# ==============================================================================
# Task 1: Validate and parse the study configuration dictionary
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 1, Step 1: Validate top-level configuration sections
# ------------------------------------------------------------------------------
def validate_top_level_config(config: Dict[str, Any]) -> Tuple[Dict[str, Any], str]:
    """
    Validates the existence and non-emptiness of required top-level configuration sections
    and computes a deterministic SHA-256 hash of the configuration.

    Args:
        config (Dict[str, Any]): The raw configuration dictionary.

    Returns:
        Tuple[Dict[str, Any], str]: The validated configuration dictionary and its SHA-256 hash.

    Raises:
        ValueError: If a required section is missing, not a dictionary, or empty.
        TypeError: If the configuration contains non-serializable types.
    """
    # Define the strict list of required top-level keys based on the task specification
    required_sections: List[str] = [
        "study_metadata",
        "llm_engine_config",
        "simulation_config",
        "data_schemas",
        "agent_architecture_config",
        "preprocessing_config",
        "prompt_templates",
        "icl_training_config",
        "embedding_config",
        "evaluation_config",
    ]

    # Iterate through each required section to enforce existence and structure
    for section in required_sections:
        # Check if key exists
        if section not in config:
            raise ValueError(f"Configuration validation failed: Missing required top-level section '{section}'.")

        value = config[section]

        # Check if value is a dictionary
        if not isinstance(value, dict):
            raise ValueError(f"Configuration validation failed: Section '{section}' must be a dictionary, got {type(value).__name__}.")

        # Check if dictionary is non-empty
        if len(value) == 0:
            raise ValueError(f"Configuration validation failed: Section '{section}' is empty.")

    # Compute deterministic hash using canonical JSON serialization
    try:
        # sort_keys=True ensures deterministic ordering of keys
        config_json_bytes = json.dumps(config, sort_keys=True).encode('utf-8')
        config_hash = hashlib.sha256(config_json_bytes).hexdigest()
    except TypeError as e:
        raise TypeError(f"Configuration validation failed: Config contains non-serializable types. Details: {str(e)}")

    return config, config_hash


# ------------------------------------------------------------------------------
# Task 1, Step 2: Validate LLM engine configuration
# ------------------------------------------------------------------------------
def validate_llm_engine_config(llm_config: Dict[str, Any]) -> Dict[str, Any]:
    """
    Validates the LLM engine configuration against OpenAI API requirements and
    reproducibility constraints.

    This function enforces strict type and range checks on all configuration parameters
    required for the LLM client, including retry logic parameters.

    Args:
        llm_config (Dict[str, Any]): The 'llm_engine_config' section of the configuration.

    Returns:
        Dict[str, Any]: The validated LLM configuration dictionary.

    Raises:
        ValueError: If any parameter violates type, range, or logical constraints.
    """
    # Validate model_name
    # Must be a non-empty string
    if not isinstance(llm_config.get("model_name"), str) or not llm_config.get("model_name"):
        raise ValueError("LLM Config Error: 'model_name' must be a non-empty string.")

    # Validate api_key (existence and type only, do not log value)
    # Must be a non-empty string
    if not isinstance(llm_config.get("api_key"), str) or not llm_config.get("api_key"):
        raise ValueError("LLM Config Error: 'api_key' must be a non-empty string.")

    # Validate temperature: 0 <= temperature <= 2
    temp = llm_config.get("temperature")
    if not isinstance(temp, (int, float)) or not (0 <= temp <= 2):
        raise ValueError(f"LLM Config Error: 'temperature' must be a float between 0 and 2. Got {temp}.")

    # Validate top_p: 0 <= top_p <= 1
    top_p = llm_config.get("top_p")
    if not isinstance(top_p, (int, float)) or not (0 <= top_p <= 1):
        raise ValueError(f"LLM Config Error: 'top_p' must be a float between 0 and 1. Got {top_p}.")

    # Validate max_tokens_output: > 0
    max_tokens = llm_config.get("max_tokens_output")
    if not isinstance(max_tokens, int) or max_tokens <= 0:
        raise ValueError(f"LLM Config Error: 'max_tokens_output' must be a positive integer. Got {max_tokens}.")

    # Validate max_retries: >= 1
    max_retries = llm_config.get("max_retries")
    if not isinstance(max_retries, int) or max_retries < 1:
        raise ValueError(f"LLM Config Error: 'max_retries' must be an integer >= 1. Got {max_retries}.")

    # Validate retry_backoff_base_s: > 0
    backoff_base = llm_config.get("retry_backoff_base_s")
    if not isinstance(backoff_base, (int, float)) or backoff_base <= 0:
        raise ValueError(f"LLM Config Error: 'retry_backoff_base_s' must be positive. Got {backoff_base}.")

    # Validate retry_backoff_max_s: > 0 and >= retry_backoff_base_s
    # This ensures the backoff schedule is logically sound (max wait is not less than initial wait)
    backoff_max = llm_config.get("retry_backoff_max_s")
    if not isinstance(backoff_max, (int, float)) or backoff_max <= 0:
        raise ValueError(f"LLM Config Error: 'retry_backoff_max_s' must be positive. Got {backoff_max}.")

    if backoff_max < backoff_base:
        raise ValueError(
            f"LLM Config Error: 'retry_backoff_max_s' ({backoff_max}) must be greater than or equal to "
            f"'retry_backoff_base_s' ({backoff_base})."
        )

    return llm_config


# ------------------------------------------------------------------------------
# Task 1, Step 3: Validate prompt template completeness
# ------------------------------------------------------------------------------
def validate_prompt_templates(templates_config: Dict[str, str]) -> Dict[str, str]:
    """
    Validates that all required prompt templates exist and contain the necessary placeholders.

    Args:
        templates_config (Dict[str, str]): The 'prompt_templates' section of the configuration.

    Returns:
        Dict[str, str]: The validated prompt templates dictionary.

    Raises:
        ValueError: If a template is missing or lacks required placeholders.
    """
    # Define the mapping of required template keys to their expected placeholders
    # This ensures that the templates provided in the config match the code's expectations for data injection.
    required_templates_map: Dict[str, Set[str]] = {
        "analyst_system_prompt": {"futures_data"},
        "economist_task_prompt": {
            "inflation_data", "labor_data", "growth_data", "financial_data",
            "unstructured_context", "analyst_market_summary"
        },
        "member_system_persona": {"archetype_name", "archetype_description"},
        "member_cod_deliberation": {
            "economist_options", "beige_book_text", "dot_plot_text",
            "fedwatch_text", "market_probs", "structured_macro_data"
        },
        "icl_reflection": {"agent_vote", "true_outcome"},
        "chair_statement_generation": {"voting_record", "majority_reasoning_summary"},
        "icl_historical_simulation_prompt": {
            "historical_meeting_id", "structured_macro_data",
            "beige_book_text", "dot_plot_text", "fedwatch_text"
        }
    }

    for template_key, required_placeholders in required_templates_map.items():
        # Check existence
        if template_key not in templates_config:
            raise ValueError(f"Prompt Template Error: Missing required template '{template_key}'.")

        template_text = templates_config[template_key]
        if not isinstance(template_text, str) or not template_text.strip():
            raise ValueError(f"Prompt Template Error: Template '{template_key}' must be a non-empty string.")

        # Parse placeholders from the template string
        # string.Formatter().parse() returns tuples of (literal_text, field_name, format_spec, conversion)
        found_placeholders = {
            field_name for _, field_name, _, _ in string.Formatter().parse(template_text)
            if field_name is not None
        }

        # Check for missing placeholders
        missing = required_placeholders - found_placeholders
        if missing:
            raise ValueError(
                f"Prompt Template Error: Template '{template_key}' is missing required placeholders: {missing}. "
                f"Found: {found_placeholders}"
            )

    return templates_config


# ------------------------------------------------------------------------------
# Task 1, Orchestrator Function
# ------------------------------------------------------------------------------
def validate_study_config(config: Dict[str, Any]) -> Tuple[Dict[str, Any], str]:
    """
    Orchestrates the validation of the entire study configuration dictionary.

    Executes a sequence of granular validation steps:
    1. Validates top-level structure and computes config hash.
    2. Validates LLM engine parameters.
    3. Validates prompt templates and placeholders.

    Args:
        config (Dict[str, Any]): The raw configuration dictionary.

    Returns:
        Tuple[Dict[str, Any], str]: The fully validated configuration and its SHA-256 hash.

    Raises:
        ValueError: If any validation step fails.
        TypeError: If configuration contains invalid types.
    """
    # Step 1: Top-level validation and hashing
    # This ensures the overall structure is correct before diving deeper.
    validated_config, config_hash = validate_top_level_config(config)

    # Step 2: LLM Engine Validation
    # This ensures the inference engine parameters are safe and compliant.
    validate_llm_engine_config(validated_config["llm_engine_config"])

    # Step 3: Prompt Template Validation
    # This ensures that all templates needed for the agent workflow are present and correct.
    validate_prompt_templates(validated_config["prompt_templates"])

    # Return the validated config and the hash for provenance
    return validated_config, config_hash
````

````python
# Task 2 — Validate ingested DataFrame schemas and join-key integrity

# ==============================================================================
# Task 2: Validate ingested DataFrame schemas and join-key integrity
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 2, Step 1: Validate df_macro_raw schema and uniqueness constraints
# ------------------------------------------------------------------------------
def validate_macro_raw(df_macro_raw: pd.DataFrame) -> Tuple[pd.DataFrame, Set[pd.Timestamp]]:
    """
    Validates the schema, data types, and logical constraints of the structured
    macroeconomic data DataFrame.

    Enforces the 'two-days-prior' information cutoff rule and verifies the
    canonical set of 16 meetings.

    Args:
        df_macro_raw (pd.DataFrame): The raw structured macro data.

    Returns:
        Tuple[pd.DataFrame, Set[pd.Timestamp]]: The validated DataFrame and the
        set of canonical meeting dates (normalized to midnight).

    Raises:
        ValueError: If schema, types, or constraints are violated.
    """
    # 1. Validate Column Existence
    required_columns = [
        "meeting_date", "info_cutoff_datetime", "pce_index_yoy", "cpi_index_yoy",
        "inflation_expectations_1yr", "treasury_yield_3m", "treasury_yield_6m",
        "m2_supply_level_sa", "bbk_real_gdp_nowcast", "unemployment_rate_u3",
        "vix_index", "fed_chair", "white_house_party", "previous_fftr_target",
        "previous_change"
    ]

    missing_cols = set(required_columns) - set(df_macro_raw.columns)
    if missing_cols:
        raise ValueError(f"df_macro_raw missing required columns: {missing_cols}")

    # 2. Validate Data Types (Coercibility check)
    # We don't coerce here (that's Task 3), but we check if they look correct.
    # meeting_date and info_cutoff_datetime must be datetime-like
    if not pd.api.types.is_datetime64_any_dtype(df_macro_raw["meeting_date"]):
        raise ValueError("df_macro_raw['meeting_date'] must be datetime64[ns].")
    if not pd.api.types.is_datetime64_any_dtype(df_macro_raw["info_cutoff_datetime"]):
        raise ValueError("df_macro_raw['info_cutoff_datetime'] must be datetime64[ns].")

    # 3. Validate Uniqueness and Non-Nullity of Primary Key
    if df_macro_raw["meeting_date"].isnull().any():
        raise ValueError("df_macro_raw['meeting_date'] contains null values.")
    if not df_macro_raw["meeting_date"].is_unique:
        duplicates = df_macro_raw[df_macro_raw.duplicated("meeting_date")]["meeting_date"].unique()
        raise ValueError(f"df_macro_raw['meeting_date'] contains duplicates: {duplicates}")

    # 4. Validate Meeting Count and Range
    # Requirement: Exactly 16 unique meetings within [2023-02-01, 2024-12-18]
    start_date = pd.Timestamp("2023-02-01")
    end_date = pd.Timestamp("2024-12-18")

    # Normalize to midnight for comparison
    dates = df_macro_raw["meeting_date"].dt.normalize()

    out_of_range = dates[(dates < start_date) | (dates > end_date)]
    if not out_of_range.empty:
        raise ValueError(f"df_macro_raw contains dates outside evaluation window: {out_of_range.tolist()}")

    if len(dates) != 16:
        raise ValueError(f"df_macro_raw must contain exactly 16 meetings, found {len(dates)}.")

    # 5. Validate Information Cutoff Alignment Rule
    # Rule: info_cutoff_datetime = meeting_date - 2 calendar days
    # We normalize both to ensure we are comparing dates, ignoring time components if present
    cutoff_dates = df_macro_raw["info_cutoff_datetime"].dt.normalize()
    expected_cutoffs = dates - pd.Timedelta(days=2)

    mismatches = df_macro_raw[cutoff_dates != expected_cutoffs]
    if not mismatches.empty:
        # Log first few mismatches for debugging
        error_details = mismatches[["meeting_date", "info_cutoff_datetime"]].head().to_dict(orient='records')
        raise ValueError(f"df_macro_raw contains rows violating the 2-day cutoff rule. Examples: {error_details}")

    return df_macro_raw, set(dates)


# ------------------------------------------------------------------------------
# Task 2, Step 2: Validate df_unstructured_raw schema and join integrity
# ------------------------------------------------------------------------------
def validate_unstructured_raw(
    df_unstructured_raw: pd.DataFrame,
    canonical_dates: Set[pd.Timestamp]
) -> pd.DataFrame:
    """
    Validates the schema of unstructured inputs and ensures strict join integrity
    with the canonical meeting dates.

    This function enforces rigorous structural checks on nested dictionary columns,
    specifically the Dot Plot distribution, to ensure downstream processing (Task 7)
    will not fail due to missing keys or malformed data.

    Args:
        df_unstructured_raw (pd.DataFrame): The raw unstructured text/data.
        canonical_dates (Set[pd.Timestamp]): The set of valid meeting dates from macro data.

    Returns:
        pd.DataFrame: The validated DataFrame.

    Raises:
        ValueError: If schema, join integrity, or nested structure constraints are violated.
    """
    # 1. Validate Column Existence
    required_columns = [
        "meeting_date", "beige_book_raw_text", "dot_plot_distribution",
        "fedwatch_raw_text", "fedwatch_probabilities", "fed_funds_futures_raw"
    ]
    missing_cols = set(required_columns) - set(df_unstructured_raw.columns)
    if missing_cols:
        raise ValueError(f"df_unstructured_raw missing required columns: {missing_cols}")

    # 2. Validate Join Integrity
    if not pd.api.types.is_datetime64_any_dtype(df_unstructured_raw["meeting_date"]):
        raise ValueError("df_unstructured_raw['meeting_date'] must be datetime64[ns].")

    unstructured_dates = set(df_unstructured_raw["meeting_date"].dt.normalize())

    # Check for missing meetings
    missing_meetings = canonical_dates - unstructured_dates
    if missing_meetings:
        raise ValueError(f"df_unstructured_raw is missing data for meetings: {sorted(list(missing_meetings))}")

    # Check for extra meetings
    extra_meetings = unstructured_dates - canonical_dates
    if extra_meetings:
        raise ValueError(f"df_unstructured_raw contains extra meetings not in macro data: {sorted(list(extra_meetings))}")

    # 3. Validate Text Fields (Non-null strings)
    for col in ["beige_book_raw_text", "fedwatch_raw_text"]:
        if df_unstructured_raw[col].isnull().any():
            raise ValueError(f"df_unstructured_raw['{col}'] contains null values.")
        # Check if all are strings
        if not df_unstructured_raw[col].apply(lambda x: isinstance(x, str)).all():
             raise ValueError(f"df_unstructured_raw['{col}'] must contain strings.")

    # 4. Validate Dot Plot Distribution Structure (Enhanced)
    # Must be None or Dict with 'release_date', 'projections', and nested 'bins'
    def validate_dot_plot(val: Any) -> bool:
        if val is None:
            return True
        if not isinstance(val, dict):
            return False

        # Check top-level keys
        if "release_date" not in val or "projections" not in val:
            return False

        projections = val["projections"]
        if not isinstance(projections, dict):
            return False

        # Iterate through projections to verify 'bins' existence and type
        # Each value in projections represents a year (or 'longer_run')
        for year_key, year_data in projections.items():
            if not isinstance(year_data, dict):
                return False
            if "bins" not in year_data:
                return False
            if not isinstance(year_data["bins"], dict):
                return False

        return True

    if not df_unstructured_raw["dot_plot_distribution"].apply(validate_dot_plot).all():
        raise ValueError("df_unstructured_raw['dot_plot_distribution'] contains invalid structures (missing 'bins' or malformed).")

    # 5. Validate FedWatch Probabilities
    # Must be Dict, values sum to 1.0 +/- 0.001
    def validate_probs(val: Any) -> bool:
        if not isinstance(val, dict):
            return False
        total_prob = sum(val.values())
        return abs(total_prob - 1.0) <= 0.001

    if not df_unstructured_raw["fedwatch_probabilities"].apply(validate_probs).all():
        raise ValueError("df_unstructured_raw['fedwatch_probabilities'] contains invalid probability distributions (must sum to 1.0).")

    # 6. Validate Fed Funds Futures Raw
    # Must be str or dict, non-null
    def validate_futures(val: Any) -> bool:
        return isinstance(val, (str, dict))

    if df_unstructured_raw["fed_funds_futures_raw"].isnull().any():
        raise ValueError("df_unstructured_raw['fed_funds_futures_raw'] contains null values.")

    if not df_unstructured_raw["fed_funds_futures_raw"].apply(validate_futures).all():
        raise ValueError("df_unstructured_raw['fed_funds_futures_raw'] must be string or dict.")

    return df_unstructured_raw


# ------------------------------------------------------------------------------
# Task 2, Step 3: Validate evaluation-only DataFrames
# ------------------------------------------------------------------------------
def validate_evaluation_dataframes(
    df_meeting_outcomes: pd.DataFrame,
    df_fomc_statements_actual: pd.DataFrame,
    df_participants_raw: pd.DataFrame,
    canonical_dates: Set[pd.Timestamp]
) -> Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    """
    Validates the evaluation-only datasets: outcomes, actual statements, and
    participant clustering data.

    Args:
        df_meeting_outcomes (pd.DataFrame): Ground truth outcomes.
        df_fomc_statements_actual (pd.DataFrame): Actual statements text.
        df_participants_raw (pd.DataFrame): Participant features.
        canonical_dates (Set[pd.Timestamp]): Valid meeting dates.

    Returns:
        Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame]: The validated DataFrames.

    Raises:
        ValueError: If schemas or logical constraints are violated.
    """
    # --- Validate df_meeting_outcomes ---
    req_cols_outcomes = [
        "meeting_date", "fftr_change_bps", "fftr_target_before_bps",
        "fftr_target_after_bps", "decision_category"
    ]
    if not set(req_cols_outcomes).issubset(df_meeting_outcomes.columns):
        raise ValueError(f"df_meeting_outcomes missing columns. Required: {req_cols_outcomes}")

    # Date alignment
    outcomes_dates = set(df_meeting_outcomes["meeting_date"].dt.normalize())
    if outcomes_dates != canonical_dates:
        raise ValueError("df_meeting_outcomes dates do not match canonical meeting set.")

    # Arithmetic Consistency: After - Before == Change
    # Ensure types are numeric first
    for col in ["fftr_change_bps", "fftr_target_before_bps", "fftr_target_after_bps"]:
        if not pd.api.types.is_integer_dtype(df_meeting_outcomes[col]):
             # Try coercing to check if they are valid integers
             try:
                 df_meeting_outcomes[col].astype(int)
             except:
                 raise ValueError(f"df_meeting_outcomes['{col}'] must be integer-like.")

    calculated_change = df_meeting_outcomes["fftr_target_after_bps"] - df_meeting_outcomes["fftr_target_before_bps"]
    mismatches = df_meeting_outcomes[calculated_change != df_meeting_outcomes["fftr_change_bps"]]
    if not mismatches.empty:
        raise ValueError("df_meeting_outcomes contains arithmetic inconsistencies (After - Before != Change).")

    # --- Validate df_fomc_statements_actual ---
    req_cols_stmts = ["meeting_date", "statement_release_datetime", "fomc_statement_actual_text"]
    if not set(req_cols_stmts).issubset(df_fomc_statements_actual.columns):
        raise ValueError(f"df_fomc_statements_actual missing columns. Required: {req_cols_stmts}")

    # Date alignment
    stmt_dates = set(df_fomc_statements_actual["meeting_date"].dt.normalize())
    if stmt_dates != canonical_dates:
        raise ValueError("df_fomc_statements_actual dates do not match canonical meeting set.")

    # Non-empty text
    if df_fomc_statements_actual["fomc_statement_actual_text"].isnull().any() or \
       (df_fomc_statements_actual["fomc_statement_actual_text"].str.len() == 0).any():
        raise ValueError("df_fomc_statements_actual contains empty or null statement text.")

    # --- Validate df_participants_raw ---
    req_cols_parts = [
        "participant_id", "name", "hawkishness_score", "regional_affiliation",
        "gender", "political_party", "focus_on_labor", "focus_on_inflation",
        "focus_on_banking", "focus_on_global_trends", "tenure_years"
    ]
    if not set(req_cols_parts).issubset(df_participants_raw.columns):
        raise ValueError(f"df_participants_raw missing columns. Required: {req_cols_parts}")

    # Unique ID
    if not df_participants_raw["participant_id"].is_unique:
        raise ValueError("df_participants_raw['participant_id'] must be unique.")

    # Finite numeric features (basic check, detailed cleaning in Task 5)
    numeric_cols = [
        "hawkishness_score", "focus_on_labor", "focus_on_inflation",
        "focus_on_banking", "focus_on_global_trends", "tenure_years"
    ]
    for col in numeric_cols:
        if not pd.api.types.is_numeric_dtype(df_participants_raw[col]):
             raise ValueError(f"df_participants_raw['{col}'] must be numeric.")

    return df_meeting_outcomes, df_fomc_statements_actual, df_participants_raw


# ------------------------------------------------------------------------------
# Task 2, Orchestrator Function
# ------------------------------------------------------------------------------
def validate_all_dataframes(
    df_macro_raw: pd.DataFrame,
    df_unstructured_raw: pd.DataFrame,
    df_meeting_outcomes: pd.DataFrame,
    df_fomc_statements_actual: pd.DataFrame,
    df_participants_raw: pd.DataFrame
) -> Dict[str, pd.DataFrame]:
    """
    Orchestrates the validation of all ingested DataFrames.

    Executes:
    1. Macro data validation (establishes canonical meeting set).
    2. Unstructured data validation (checks join integrity).
    3. Evaluation data validation (outcomes, statements, participants).

    Args:
        df_macro_raw (pd.DataFrame): Raw macro data.
        df_unstructured_raw (pd.DataFrame): Raw unstructured data.
        df_meeting_outcomes (pd.DataFrame): Ground truth outcomes.
        df_fomc_statements_actual (pd.DataFrame): Actual statements.
        df_participants_raw (pd.DataFrame): Participant features.

    Returns:
        Dict[str, pd.DataFrame]: A dictionary containing the validated DataFrames.
    """
    # Step 1: Validate Macro Data and get Canonical Dates
    validated_macro, canonical_dates = validate_macro_raw(df_macro_raw)

    # Step 2: Validate Unstructured Data against Canonical Dates
    validated_unstructured = validate_unstructured_raw(df_unstructured_raw, canonical_dates)

    # Step 3: Validate Evaluation DataFrames
    val_outcomes, val_stmts, val_parts = validate_evaluation_dataframes(
        df_meeting_outcomes,
        df_fomc_statements_actual,
        df_participants_raw,
        canonical_dates
    )

    return {
        "df_macro_raw": validated_macro,
        "df_unstructured_raw": validated_unstructured,
        "df_meeting_outcomes": val_outcomes,
        "df_fomc_statements_actual": val_stmts,
        "df_participants_raw": val_parts
    }
````

````python
# Task 3 — Cleanse structured macroeconomic data (df_macro_raw)

# ==============================================================================
# Task 3: Cleanse structured macroeconomic data (df_macro_raw)
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 3, Step 1: Coerce dtypes deterministically and quarantine non-coercible rows
# ------------------------------------------------------------------------------
def coerce_macro_dtypes(df_macro_raw: pd.DataFrame) -> Tuple[pd.DataFrame, pd.DataFrame, List[Dict[str, Any]]]:
    """
    Coerces data types for the structured macro DataFrame with strict deterministic rules.
    Quarantines rows where required numeric conversions fail.

    Transformations:
    1. Strips artifacts (%, comma) from numeric-target columns.
    2. Coerces to float64 (numeric) and Int64 (basis points).
    3. Standardizes categorical strings (title case, stripped).
    4. Enforces timezone-naive datetimes.

    Args:
        df_macro_raw (pd.DataFrame): Validated raw macro data.

    Returns:
        Tuple[pd.DataFrame, pd.DataFrame, List[Dict[str, Any]]]:
            - df_clean: DataFrame with successfully coerced rows.
            - df_quarantine: DataFrame with rows that failed coercion.
            - coercion_log: List of dictionaries detailing specific changes/errors.
    """
    # Create a copy to avoid mutating the input
    df = df_macro_raw.copy()

    # Initialize log and quarantine mask
    coercion_log = []
    quarantine_mask = pd.Series(False, index=df.index)

    # 1. Enforce Timezone Naive Datetimes
    # Equation: dt_naive = dt_aware.tz_localize(None)
    for col in ["meeting_date", "info_cutoff_datetime"]:
        if pd.api.types.is_datetime64_any_dtype(df[col]):
            # If tz-aware, convert to naive (UTC base implied or wall time preserved)
            # We assume input validation ensured datetime type, here we ensure tz-naive
            if df[col].dt.tz is not None:
                df[col] = df[col].dt.tz_localize(None)
                coercion_log.append({"action": "tz_strip", "column": col, "details": "Converted to timezone-naive"})

    # 2. Numeric Coercion (Float fields)
    float_cols = [
        "pce_index_yoy", "cpi_index_yoy", "inflation_expectations_1yr",
        "treasury_yield_3m", "treasury_yield_6m", "m2_supply_level_sa",
        "bbk_real_gdp_nowcast", "unemployment_rate_u3", "vix_index"
    ]

    for col in float_cols:
        # Pre-cleaning: strip %, commas, whitespace if string
        if df[col].dtype == "object":
            df[col] = df[col].astype(str).str.replace(r"[%,]", "", regex=True).str.strip()

        # Coerce to numeric
        # Store original for logging check
        original = df[col].copy()
        df[col] = pd.to_numeric(df[col], errors='coerce')

        # Identify failures (NaNs that weren't NaNs before, or just NaNs in required fields)
        # Note: We treat existing NaNs as "missing" (handled in Step 2), not coercion failures.
        # Coercion failure is when a non-null value becomes NaN.
        failed_coercion = original.notna() & df[col].isna()
        if failed_coercion.any():
            quarantine_mask |= failed_coercion
            for idx in df[failed_coercion].index:
                coercion_log.append({
                    "meeting_date": df.loc[idx, "meeting_date"],
                    "column": col,
                    "original_value": original.loc[idx],
                    "error": "Numeric coercion failed"
                })

    # 3. Integer Coercion (Basis Points fields)
    int_cols = ["previous_fftr_target", "previous_change"]
    for col in int_cols:
        if df[col].dtype == "object":
            df[col] = df[col].astype(str).str.replace(r"[%,]", "", regex=True).str.strip()

        # Coerce to float first to handle "550.0" strings
        original = df[col].copy()
        numeric_series = pd.to_numeric(df[col], errors='coerce')

        # Check for non-integer floats (e.g., 550.5 bps is invalid for target)
        # We allow a small epsilon for float precision issues
        is_not_integer = (numeric_series % 1 != 0) & numeric_series.notna()

        if is_not_integer.any():
            quarantine_mask |= is_not_integer
            for idx in df[is_not_integer].index:
                coercion_log.append({
                    "meeting_date": df.loc[idx, "meeting_date"],
                    "column": col,
                    "original_value": original.loc[idx],
                    "error": "Value is not a valid integer"
                })
            # Set invalid integers to NaN for the Int64 conversion step
            numeric_series[is_not_integer] = np.nan

        # Convert to nullable Int64
        df[col] = numeric_series.astype("Int64")

        # Check for coercion failures (non-null input became NaN)
        failed_coercion = original.notna() & df[col].isna()
        if failed_coercion.any():
            quarantine_mask |= failed_coercion
            for idx in df[failed_coercion].index:
                coercion_log.append({
                    "meeting_date": df.loc[idx, "meeting_date"],
                    "column": col,
                    "original_value": original.loc[idx],
                    "error": "Integer coercion failed"
                })

    # 4. Categorical Standardization
    cat_cols = ["fed_chair", "white_house_party"]
    for col in cat_cols:
        # Convert to string, strip whitespace, title case for canonical spelling
        df[col] = df[col].astype(str).str.strip().str.title()
        df[col] = df[col].astype("category")

    # Split DataFrames
    df_quarantine = df[quarantine_mask].copy()
    df_clean = df[~quarantine_mask].copy()

    return df_clean, df_quarantine, coercion_log


# ------------------------------------------------------------------------------
# Task 3, Step 2: Handle missing values under a single deterministic policy
# ------------------------------------------------------------------------------
def impute_missing_values(df_macro_clean: pd.DataFrame) -> Tuple[pd.DataFrame, List[Dict[str, Any]]]:
    """
    Imputes missing values using a deterministic forward-fill policy respecting
    time order.

    Policy:
    1. Sort by `meeting_date`.
    2. Forward-fill (ffill) missing values from the most recent prior meeting.
    3. Halt if missing values remain (e.g., at the start of the series), as
       improvising data is disallowed.

    Args:
        df_macro_clean (pd.DataFrame): Coerced macro data.

    Returns:
        Tuple[pd.DataFrame, List[Dict[str, Any]]]:
            - df_imputed: DataFrame with no missing values.
            - imputation_log: List of imputed entries.

    Raises:
        ValueError: If missing values persist after forward-fill (e.g., first row).
    """
    # Sort by meeting_date to ensure ffill uses PAST data (no look-ahead bias)
    df_sorted = df_macro_clean.sort_values("meeting_date").copy()

    imputation_log = []

    # Identify missing values BEFORE imputation for logging
    # We iterate columns to log specific cells
    cols_to_check = [c for c in df_sorted.columns if c not in ["meeting_date", "info_cutoff_datetime"]]

    # We need to log what *will* be imputed.
    # ffill propagates the last valid observation forward.

    # Perform Forward Fill
    # Note: ffill() propagates last valid observation forward to next valid.
    df_imputed = df_sorted.ffill()

    # Generate Log
    # Compare pre- and post-imputation
    mask_imputed = df_sorted.isna() & df_imputed.notna()

    if mask_imputed.any().any():
        # Iterate to log details
        # This is computationally acceptable for N=16
        for col in cols_to_check:
            imputed_rows = df_imputed[mask_imputed[col]]
            for idx, row in imputed_rows.iterrows():
                # Find source of imputation (the value used)
                # Since it's ffill, the value is row[col]
                imputation_log.append({
                    "meeting_date": row["meeting_date"],
                    "column": col,
                    "imputed_value": row[col],
                    "method": "forward_fill"
                })

    # Check for remaining NaNs (e.g., leading NaNs that couldn't be filled)
    remaining_nans = df_imputed.isna().sum()
    if remaining_nans.any():
        # Identify specific cells for the error message
        nan_locs = df_imputed[df_imputed.isna().any(axis=1)]
        error_msg = f"Critical Data Error: Missing values persist after forward-fill in columns: {remaining_nans[remaining_nans > 0].index.tolist()}. " \
                    f"Affected meetings: {nan_locs['meeting_date'].dt.date.tolist()}. " \
                    "Cannot proceed without inventing data."
        raise ValueError(error_msg)

    return df_imputed, imputation_log


# ------------------------------------------------------------------------------
# Task 3, Step 3: Validate range and sign constraints and classify outliers
# ------------------------------------------------------------------------------
def flag_macro_outliers(df_macro_imputed: pd.DataFrame) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Validates numeric ranges and flags outliers based on domain-specific constraints.
    Halts on likely unit errors (e.g., fractions instead of percents).

    Constraints:
    - Inflation (PCE, CPI, Exp): [-5, 15]
    - Yields (3m, 6m): [0, 20]
    - VIX: [5, 90]
    - FFTR Target: [0, 2000] bps

    Args:
        df_macro_imputed (pd.DataFrame): Imputed macro data.

    Returns:
        Tuple[pd.DataFrame, pd.DataFrame]:
            - df_final: The passed DataFrame (unmutated).
            - df_outlier_flags: Long-format DataFrame of flagged values.

    Raises:
        ValueError: If a value indicates a critical unit error (e.g., 0.03 for CPI).
    """
    df = df_macro_imputed.copy()
    flags = []

    # Define constraints map: (min, max, unit_error_threshold_heuristic)
    # unit_error_threshold_heuristic: if value is < this and expected is %, likely fraction.
    # For CPI/PCE, typical is 2-5%. If we see 0.03, it's likely 3%. Threshold 0.15.
    # For Yields, typical 0-5%. Threshold 0.15.
    constraints = {
        "pce_index_yoy": (-5, 15, 0.15),
        "cpi_index_yoy": (-5, 15, 0.15),
        "inflation_expectations_1yr": (-5, 15, 0.15),
        "treasury_yield_3m": (0, 20, 0.15),
        "treasury_yield_6m": (0, 20, 0.15),
        "unemployment_rate_u3": (0, 25, 0.15), # U3 rarely > 15, but 25 safe max
        "vix_index": (5, 90, None), # VIX is index, not percent, no fraction confusion likely
        "previous_fftr_target": (0, 2000, None) # bps
    }

    for col, (min_val, max_val, unit_thresh) in constraints.items():
        series = df[col]

        # 1. Check Range
        out_of_range = (series < min_val) | (series > max_val)
        if out_of_range.any():
            for idx in df[out_of_range].index:
                val = series.loc[idx]
                flags.append({
                    "meeting_date": df.loc[idx, "meeting_date"],
                    "column": col,
                    "value": val,
                    "issue": f"Out of range [{min_val}, {max_val}]",
                    "severity": "Warning"
                })

        # 2. Check Unit Plausibility (Fraction vs Percent)
        # Only applies to percent fields where we defined a threshold
        if unit_thresh is not None:
            # Check for small positive values that look like fractions
            # e.g. 0.0 < val < 0.15
            # We exclude 0.0 exactly as that might be valid (0% inflation)
            likely_fraction = (series > 0.0) & (series < unit_thresh)

            if likely_fraction.any():
                # This is a critical error requiring intervention
                bad_vals = series[likely_fraction].tolist()
                raise ValueError(
                    f"Critical Unit Error: Column '{col}' contains values {bad_vals} which appear to be fractions (e.g., 0.03) "
                    f"instead of percentages (e.g., 3.0). The system expects percentages. Halting to prevent model corruption."
                )

    # Create flags DataFrame
    if flags:
        df_outlier_flags = pd.DataFrame(flags)
    else:
        df_outlier_flags = pd.DataFrame(columns=["meeting_date", "column", "value", "issue", "severity"])

    return df, df_outlier_flags


# ------------------------------------------------------------------------------
# Task 3, Orchestrator Function
# ------------------------------------------------------------------------------
def cleanse_macro_data(df_macro_raw: pd.DataFrame) -> Dict[str, Any]:
    """
    Orchestrates the cleansing pipeline for structured macroeconomic data.

    Pipeline:
    1. Coerce types and quarantine failures.
    2. Impute missing values via deterministic forward-fill.
    3. Validate ranges and flag outliers.

    Args:
        df_macro_raw (pd.DataFrame): Validated raw macro data.

    Returns:
        Dict[str, Any]:
            - df_macro_final: The clean, imputed, validated DataFrame.
            - df_macro_quarantine: Rows excluded due to coercion failure.
            - df_outlier_flags: Warning flags for range violations.
            - logs: Dictionary of coercion and imputation logs.
    """
    # Step 1: Coerce
    df_clean, df_quarantine, coercion_log = coerce_macro_dtypes(df_macro_raw)

    # Step 2: Impute
    # Note: We impute only the clean partition. Quarantined rows are discarded from the main flow.
    df_imputed, imputation_log = impute_missing_values(df_clean)

    # Step 3: Validate Ranges
    df_final, df_outlier_flags = flag_macro_outliers(df_imputed)

    return {
        "df_macro_final": df_final,
        "df_macro_quarantine": df_quarantine,
        "df_outlier_flags": df_outlier_flags,
        "logs": {
            "coercion_log": coercion_log,
            "imputation_log": imputation_log
        }
    }
````

````python
# Task 4 — Cleanse unstructured narrative data (df_unstructured_raw)

# ==============================================================================
# Task 4: Cleanse unstructured narrative data (df_unstructured_raw)
# ==============================================================================

# -----------------------------------------------------------------------------------
# Task 4, Step 1: Cleanse text fields using loss-minimizing deterministic transforms
# -----------------------------------------------------------------------------------
def cleanse_narrative_text(df_unstructured_raw: pd.DataFrame) -> Tuple[pd.DataFrame, List[Dict[str, Any]]]:
    """
    Cleanse unstructured text fields by stripping HTML tags, normalizing whitespace,
    and trimming. Computes checksums for auditability.

    Transformations:
    1. Strip leading/trailing whitespace.
    2. Collapse multiple spaces/newlines to single space.
    3. Remove HTML tags (e.g., <br>, <p>).
    4. Compute SHA-256 checksums for raw and clean text.

    Args:
        df_unstructured_raw (pd.DataFrame): Validated raw unstructured data.

    Returns:
        Tuple[pd.DataFrame, List[Dict[str, Any]]]:
            - df_clean: DataFrame with cleaned text columns and checksums.
            - cleaning_log: List of cleaning metadata.
    """
    df = df_unstructured_raw.copy()
    cleaning_log = []

    text_cols = ["beige_book_raw_text", "fedwatch_raw_text"]

    # Pre-compile regex for performance
    # Matches <...> tags. Note: This is a basic stripper; for complex HTML, bs4 is better,
    # but regex is sufficient for "artifacts" and deterministic.
    html_tag_pattern = re.compile(r'<[^>]+>')
    whitespace_pattern = re.compile(r'\s+')

    for col in text_cols:
        # Create checksum column names
        raw_hash_col = f"{col}_raw_hash"
        clean_hash_col = f"{col}_clean_hash"

        # Compute raw hash
        df[raw_hash_col] = df[col].apply(lambda x: hashlib.sha256(str(x).encode('utf-8')).hexdigest())

        # Define cleaning function
        def clean_text(text: str) -> str:
            if not isinstance(text, str):
                return ""
            # 1. Remove HTML tags
            text = html_tag_pattern.sub(' ', text)
            # 2. Normalize whitespace (collapse to single space)
            text = whitespace_pattern.sub(' ', text)
            # 3. Strip
            return text.strip()

        # Apply cleaning
        # We overwrite the column with the cleaned version as per "df_unstructured_clean_text" requirement
        # but we keep the raw hash for provenance.
        original_lengths = df[col].str.len()
        df[col] = df[col].apply(clean_text)
        cleaned_lengths = df[col].str.len()

        # Compute clean hash
        df[clean_hash_col] = df[col].apply(lambda x: hashlib.sha256(str(x).encode('utf-8')).hexdigest())

        # Log stats
        avg_reduction = (original_lengths - cleaned_lengths).mean()
        cleaning_log.append({
            "column": col,
            "avg_chars_removed": avg_reduction,
            "rows_processed": len(df)
        })

    return df, cleaning_log


# ------------------------------------------------------------------------------
# Task 4, Step 2: Validate and normalize fedwatch_probabilities
# ------------------------------------------------------------------------------
def normalize_fedwatch_probs(df_unstructured_clean: pd.DataFrame) -> Tuple[pd.DataFrame, List[Dict[str, Any]]]:
    """
    Validates and normalizes the FedWatch probability distributions.

    Constraints:
    1. Probabilities must be in [0, 1].
    2. Sum must be within 1.0 +/- 0.001.
    3. Keys must contain interpretable labels (hike, cut, pause, hold).

    Transformations:
    - Renormalizes probabilities to sum exactly to 1.0 if within tolerance.
    - Halts if sum deviation exceeds tolerance.

    Args:
        df_unstructured_clean (pd.DataFrame): DataFrame with cleaned text.

    Returns:
        Tuple[pd.DataFrame, List[Dict[str, Any]]]:
            - df_normalized: DataFrame with normalized probability dicts.
            - normalization_log: Log of actions taken.

    Raises:
        ValueError: If probability constraints are violated beyond tolerance.
    """
    df = df_unstructured_clean.copy()
    normalization_log = []

    # We need to process the dicts row by row
    normalized_probs_series = []

    valid_keywords = {"hike", "cut", "pause", "hold", "maintain"}

    for idx, row in df.iterrows():
        probs = row["fedwatch_probabilities"]
        meeting_date = row["meeting_date"]

        # 1. Validate Keys
        keys_lower = {k.lower() for k in probs.keys()}
        if not any(any(kw in k for kw in valid_keywords) for k in keys_lower):
            # If keys don't look like outcomes, flag it.
            # (Strictness: could halt, but we'll warn/log for now unless empty)
            if not probs:
                raise ValueError(f"FedWatch probabilities empty for meeting {meeting_date}")

        # 2. Validate Values
        values = list(probs.values())
        if any(p < 0 or p > 1 for p in values):
            raise ValueError(f"FedWatch probabilities contain values outside [0,1] for meeting {meeting_date}: {probs}")

        total_prob = sum(values)

        # 3. Check Sum Constraint
        if abs(total_prob - 1.0) > 0.001:
            raise ValueError(f"FedWatch probabilities sum to {total_prob} (tolerance 0.001) for meeting {meeting_date}. Halting.")

        # 4. Normalize
        # Even if sum is 0.9995, we renormalize to 1.0 to be precise
        if total_prob != 1.0 and total_prob != 0:
            new_probs = {k: v / total_prob for k, v in probs.items()}
            normalization_log.append({
                "meeting_date": meeting_date,
                "original_sum": total_prob,
                "action": "renormalized"
            })
        else:
            new_probs = probs.copy()

        normalized_probs_series.append(new_probs)

    df["fedwatch_probabilities"] = normalized_probs_series
    return df, normalization_log


# ------------------------------------------------------------------------------
# Task 4, Step 3: Apply deterministic null-handling for dot_plot_distribution
# ------------------------------------------------------------------------------
def flag_missing_dot_plots(df_unstructured_norm: pd.DataFrame) -> Tuple[pd.DataFrame, List[pd.Timestamp]]:
    """
    Flags meetings with missing Dot Plot distributions without imputing yet.
    Validates structure of present Dot Plots.

    Args:
        df_unstructured_norm (pd.DataFrame): DataFrame with normalized probs.

    Returns:
        Tuple[pd.DataFrame, List[pd.Timestamp]]:
            - df_flagged: DataFrame with 'dot_plot_is_missing' column.
            - missing_meetings: List of meeting dates where Dot Plot is None.
    """
    df = df_unstructured_norm.copy()

    # Create missing indicator
    df["dot_plot_is_missing"] = df["dot_plot_distribution"].isnull()

    missing_meetings = df[df["dot_plot_is_missing"]]["meeting_date"].tolist()

    # Validate structure of present dot plots
    # (Basic schema check was done in Task 2, here we ensure keys are valid for Task 7)
    present_mask = ~df["dot_plot_is_missing"]

    for idx, row in df[present_mask].iterrows():
        dp = row["dot_plot_distribution"]
        # Task 2 checked for 'release_date' and 'projections'
        # Here we check 'bins' inside projections to be safe for Task 7
        for year, data in dp["projections"].items():
            if "bins" not in data or not isinstance(data["bins"], dict):
                raise ValueError(f"Invalid Dot Plot structure (missing 'bins') for meeting {row['meeting_date']}")

    return df, missing_meetings


# ------------------------------------------------------------------------------
# Task 4, Orchestrator Function
# ------------------------------------------------------------------------------
def cleanse_unstructured_data(df_unstructured_raw: pd.DataFrame) -> Dict[str, Any]:
    """
    Orchestrates the cleansing pipeline for unstructured narrative data.

    Pipeline:
    1. Cleanse text fields (HTML strip, whitespace) and compute checksums.
    2. Normalize FedWatch probabilities to sum to 1.0.
    3. Flag missing Dot Plots (preparation for imputation).

    Args:
        df_unstructured_raw (pd.DataFrame): Validated raw unstructured data.

    Returns:
        Dict[str, Any]:
            - df_unstructured_clean: The fully cleansed DataFrame.
            - logs: Dictionary of cleaning and normalization logs.
            - missing_dot_plots: List of meetings requiring Dot Plot imputation.
    """
    # Step 1: Clean Text
    df_text_clean, cleaning_log = cleanse_narrative_text(df_unstructured_raw)

    # Step 2: Normalize Probabilities
    df_norm, normalization_log = normalize_fedwatch_probs(df_text_clean)

    # Step 3: Flag Missing Dot Plots
    df_final, missing_dot_plots = flag_missing_dot_plots(df_norm)

    return {
        "df_unstructured_clean": df_final,
        "logs": {
            "text_cleaning_log": cleaning_log,
            "normalization_log": normalization_log
        },
        "missing_dot_plots": missing_dot_plots
    }
````

````python
# Task 5 — Cleanse participant data and evaluation DataFrames

# ==============================================================================
# Task 5: Cleanse participant data and evaluation DataFrames
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 5, Step 1: Cleanse df_participants_raw for clustering
# ------------------------------------------------------------------------------
def cleanse_participants_data(df_participants_raw: pd.DataFrame) -> Tuple[pd.DataFrame, Dict[str, List[str]]]:
    """
    Cleanse participant data for clustering. Coerces types, standardizes categories,
    and imputes missing values deterministically.

    Transformations:
    1. Coerce numeric features to float64.
    2. Standardize categorical features (Title Case).
    3. Map 'regional_affiliation' to canonical categories if numeric.
    4. Impute missing numerics with median, categoricals with mode.

    Args:
        df_participants_raw (pd.DataFrame): Raw participant data.

    Returns:
        Tuple[pd.DataFrame, Dict[str, List[str]]]:
            - df_clean: Cleansed DataFrame ready for encoding/scaling.
            - feature_spec: Dictionary listing numeric and categorical feature names.
    """
    df = df_participants_raw.copy()

    # Define feature sets
    numeric_features = [
        "hawkishness_score", "focus_on_labor", "focus_on_inflation",
        "focus_on_banking", "focus_on_global_trends", "tenure_years"
    ]
    categorical_features = ["regional_affiliation", "gender", "political_party"]

    # 1. Numeric Coercion & Imputation
    for col in numeric_features:
        df[col] = pd.to_numeric(df[col], errors='coerce')
        if df[col].isnull().any():
            median_val = df[col].median()
            df[col].fillna(median_val, inplace=True)

    # 2. Categorical Standardization
    # Handle regional_affiliation specifically
    # Map 0 -> Governor, 1 -> President (common convention, or treat as distinct categories)
    # We convert to string to ensure consistent categorical encoding later
    if pd.api.types.is_numeric_dtype(df["regional_affiliation"]):
        df["regional_affiliation"] = df["regional_affiliation"].map({0: "Governor", 1: "President"}).fillna("Unknown")

    for col in categorical_features:
        df[col] = df[col].astype(str).str.strip().str.title()
        # Impute 'Nan' or 'None' strings resulting from conversion
        df[col].replace({"Nan": "Unknown", "None": "Unknown"}, inplace=True)

        # Mode imputation if "Unknown" is not desired, but "Unknown" is safer for clustering
        # Here we stick to "Unknown" as an explicit category to avoid assumption.

    # Set Index
    if "participant_id" in df.columns:
        df.set_index("participant_id", inplace=True)

    feature_spec = {
        "numeric": numeric_features,
        "categorical": categorical_features
    }

    return df, feature_spec


# ------------------------------------------------------------------------------
# Task 5, Step 2: Cleanse df_meeting_outcomes
# ------------------------------------------------------------------------------
def cleanse_meeting_outcomes(df_meeting_outcomes: pd.DataFrame) -> pd.DataFrame:
    """
    Cleanse meeting outcomes data. Validates arithmetic consistency and standardizes categories.

    Transformations:
    1. Coerce basis points to integers.
    2. Validate arithmetic identity: After - Before == Change.
    3. Standardize decision category to Title Case.

    Args:
        df_meeting_outcomes (pd.DataFrame): Raw outcomes data.

    Returns:
        pd.DataFrame: Cleansed outcomes data.

    Raises:
        ValueError: If arithmetic consistency check fails.
    """
    df = df_meeting_outcomes.copy()

    # 1. Coerce to Integer
    bps_cols = ["fftr_change_bps", "fftr_target_before_bps", "fftr_target_after_bps"]
    for col in bps_cols:
        df[col] = pd.to_numeric(df[col], errors='coerce').astype("Int64")
        if df[col].isnull().any():
             raise ValueError(f"df_meeting_outcomes['{col}'] contains nulls after coercion.")

    # 2. Arithmetic Validation
    calculated_change = df["fftr_target_after_bps"] - df["fftr_target_before_bps"]
    mismatches = df[calculated_change != df["fftr_change_bps"]]

    if not mismatches.empty:
        # Log details
        error_details = mismatches[["meeting_date", "fftr_change_bps", "fftr_target_before_bps", "fftr_target_after_bps"]].to_dict(orient='records')
        raise ValueError(f"Arithmetic inconsistency in meeting outcomes (Change != After - Before). Details: {error_details}")

    # 3. Standardize Category
    df["decision_category"] = df["decision_category"].astype(str).str.strip().str.title()
    valid_cats = {"Hike", "Hold", "Cut", "Pause", "Maintain"} # Pause/Maintain map to Hold usually

    # Map synonyms
    df["decision_category"] = df["decision_category"].replace({"Pause": "Hold", "Maintain": "Hold"})

    invalid_cats = set(df["decision_category"].unique()) - {"Hike", "Hold", "Cut"}
    if invalid_cats:
        raise ValueError(f"Invalid decision categories found: {invalid_cats}. Expected {{Hike, Hold, Cut}}.")

    return df


# ------------------------------------------------------------------------------
# Task 5, Step 3: Cleanse df_fomc_statements_actual
# ------------------------------------------------------------------------------
def cleanse_actual_statements(df_fomc_statements_actual: pd.DataFrame) -> pd.DataFrame:
    """
    Cleanse actual FOMC statements for semantic similarity comparison.

    Transformations:
    1. Normalize whitespace (collapse multiple spaces/newlines).
    2. Strip leading/trailing whitespace.
    3. Compute SHA-256 checksum.

    Args:
        df_fomc_statements_actual (pd.DataFrame): Raw statements data.

    Returns:
        pd.DataFrame: Cleansed statements data with checksums.
    """
    df = df_fomc_statements_actual.copy()

    # Remove whitespace
    whitespace_pattern = re.compile(r'\s+')

    # Utility for cleaning text
    def clean_text(text: str) -> str:
        if not isinstance(text, str):
            return ""
        text = whitespace_pattern.sub(' ', text)
        return text.strip()

    df["fomc_statement_actual_text"] = df["fomc_statement_actual_text"].apply(clean_text)

    # Compute checksum
    df["statement_hash"] = df["fomc_statement_actual_text"].apply(
        lambda x: hashlib.sha256(x.encode('utf-8')).hexdigest()
    )

    return df


# ------------------------------------------------------------------------------
# Task 5, Orchestrator Function
# ------------------------------------------------------------------------------
def cleanse_evaluation_data(
    df_participants_raw: pd.DataFrame,
    df_meeting_outcomes: pd.DataFrame,
    df_fomc_statements_actual: pd.DataFrame
) -> Dict[str, Any]:
    """
    Orchestrates the cleansing of all evaluation-related DataFrames.

    Args:
        df_participants_raw (pd.DataFrame): Raw participant data.
        df_meeting_outcomes (pd.DataFrame): Raw outcomes data.
        df_fomc_statements_actual (pd.DataFrame): Raw statements data.

    Returns:
        Dict[str, Any]:
            - df_participants_clean: Cleansed participant data.
            - participant_feature_spec: Feature specification for clustering.
            - df_outcomes_clean: Cleansed outcomes data.
            - df_statements_clean: Cleansed statements data.
    """
    # Step 1: Participants
    df_parts_clean, feature_spec = cleanse_participants_data(df_participants_raw)

    # Step 2: Outcomes
    df_outcomes_clean = cleanse_meeting_outcomes(df_meeting_outcomes)

    # Step 3: Statements
    df_stmts_clean = cleanse_actual_statements(df_fomc_statements_actual)

    return {
        "df_participants_clean": df_parts_clean,
        "participant_feature_spec": feature_spec,
        "df_outcomes_clean": df_outcomes_clean,
        "df_statements_clean": df_stmts_clean
    }
````

````python
# Task 6 — Construct the canonical meeting-level dataset via joins

# ==============================================================================
# Task 6: Construct the canonical meeting-level dataset via joins
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 6, Step 1: Build the canonical meeting index and perform inner join
# ------------------------------------------------------------------------------
def join_agent_inputs(
    df_macro_final: pd.DataFrame,
    df_unstructured_clean: pd.DataFrame
) -> pd.DataFrame:
    """
    Joins structured macro data and unstructured narrative data into a single
    canonical DataFrame for agent inputs. Enforces strict 16-meeting coverage.

    Args:
        df_macro_final (pd.DataFrame): Cleaned macro data.
        df_unstructured_clean (pd.DataFrame): Cleaned unstructured data.

    Returns:
        pd.DataFrame: Joined DataFrame with 16 rows.

    Raises:
        ValueError: If the join results in fewer than 16 meetings.
    """
    # Ensure meeting_date is the join key
    # We assume they are already normalized from Task 2/3/4, but we enforce index alignment

    # Create copies to avoid mutation
    df_macro = df_macro_final.copy()
    df_unstruct = df_unstructured_clean.copy()

    # Set index if not already
    if "meeting_date" in df_macro.columns:
        df_macro.set_index("meeting_date", inplace=True)
    if "meeting_date" in df_unstruct.columns:
        df_unstruct.set_index("meeting_date", inplace=True)

    # Perform Inner Join
    # validate='one_to_one' ensures uniqueness on both sides
    df_joined = df_macro.join(df_unstruct, how="inner", validate="one_to_one")

    # Validate Row Count
    if len(df_joined) != 16:
        raise ValueError(f"Join failed to preserve 16 meetings. Resulting count: {len(df_joined)}")

    return df_joined


# ------------------------------------------------------------------------------
# Task 6, Step 2: Join evaluation-only DataFrames
# ------------------------------------------------------------------------------
def join_evaluation_data(
    df_outcomes_clean: pd.DataFrame,
    df_statements_clean: pd.DataFrame
) -> pd.DataFrame:
    """
    Joins evaluation-only datasets (outcomes and actual statements).
    Keeps this logically separate from agent inputs.

    Args:
        df_outcomes_clean (pd.DataFrame): Cleaned outcomes.
        df_statements_clean (pd.DataFrame): Cleaned actual statements.

    Returns:
        pd.DataFrame: Joined evaluation DataFrame.
    """
    df_out = df_outcomes_clean.copy()
    df_stmt = df_statements_clean.copy()

    # Set index name
    if "meeting_date" in df_out.columns:
        df_out.set_index("meeting_date", inplace=True)
    if "meeting_date" in df_stmt.columns:
        df_stmt.set_index("meeting_date", inplace=True)

    # Join data structures
    df_eval = df_out.join(df_stmt, how="inner", validate="one_to_one")

    if len(df_eval) != 16:
        raise ValueError(f"Evaluation join failed to preserve 16 meetings. Resulting count: {len(df_eval)}")

    return df_eval


# ------------------------------------------------------------------------------
# Task 6, Step 3: Mark and enforce leakage boundaries
# ------------------------------------------------------------------------------
def construct_meeting_packets(
    df_agent_inputs: pd.DataFrame,
    df_evaluation: pd.DataFrame
) -> Tuple[Dict[pd.Timestamp, Dict[str, Any]], Dict[pd.Timestamp, Dict[str, Any]]]:
    """
    Converts DataFrames into meeting packets with explicit leakage boundaries.

    Args:
        df_agent_inputs (pd.DataFrame): Joined agent inputs.
        df_evaluation (pd.DataFrame): Joined evaluation data.

    Returns:
        Tuple[Dict, Dict]:
            - agent_packets: Dict keyed by meeting_date containing ONLY input fields.
            - eval_packets: Dict keyed by meeting_date containing ONLY evaluation fields.
    """
    # Convert to dictionary records keyed by index (meeting_date)
    # orient='index' returns {index: {col: val, ...}}
    agent_dict = df_agent_inputs.to_dict(orient='index')
    eval_dict = df_evaluation.to_dict(orient='index')

    # Add type tags for runtime checking
    for date, packet in agent_dict.items():
        packet["_packet_type"] = "agent_input"

        # Ensure meeting_date is inside the packet too for convenience
        packet["meeting_date"] = date

    for date, packet in eval_dict.items():
        packet["_packet_type"] = "evaluation_only"
        packet["meeting_date"] = date

    return agent_dict, eval_dict


# ------------------------------------------------------------------------------
# Task 6, Orchestrator Function
# ------------------------------------------------------------------------------
def construct_canonical_datasets(
    df_macro_final: pd.DataFrame,
    df_unstructured_clean: pd.DataFrame,
    df_outcomes_clean: pd.DataFrame,
    df_statements_clean: pd.DataFrame
) -> Dict[str, Any]:
    """
    Orchestrates the construction of canonical datasets and meeting packets.

    Args:
        df_macro_final (pd.DataFrame): Cleaned macro data.
        df_unstructured_clean (pd.DataFrame): Cleaned unstructured data.
        df_outcomes_clean (pd.DataFrame): Cleaned outcomes.
        df_statements_clean (pd.DataFrame): Cleaned statements.

    Returns:
        Dict[str, Any]:
            - agent_packets: Dictionary of agent input packets.
            - evaluation_packets: Dictionary of evaluation packets.
            - df_agent_inputs_canonical: The joined DataFrame (for reference).
            - df_evaluation_canonical: The joined evaluation DataFrame (for reference).
    """
    # Step 1: Join Agent Inputs
    df_agent_inputs = join_agent_inputs(df_macro_final, df_unstructured_clean)

    # Step 2: Join Evaluation Data
    df_eval = join_evaluation_data(df_outcomes_clean, df_statements_clean)

    # Step 3: Construct Packets
    agent_packets, eval_packets = construct_meeting_packets(df_agent_inputs, df_eval)

    return {
        "agent_packets": agent_packets,
        "evaluation_packets": eval_packets,
        "df_agent_inputs_canonical": df_agent_inputs,
        "df_evaluation_canonical": df_eval
    }
````

````python
# Task 7 — Preprocess Dot Plot into verbalized distribution string

# ==============================================================================
# Task 7: Preprocess Dot Plot into verbalized distribution string
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 7, Step 1: Implement the Dot Plot verbalization algorithm
# ------------------------------------------------------------------------------
def verbalize_dot_plot(dot_plot_distribution: Optional[Dict[str, Any]]) -> Optional[str]:
    """
    Converts a structured Dot Plot dictionary into a natural language string.

    Algorithm:
    1. If None, return None.
    2. Sort projection years (numeric ascending, 'longer_run' last).
    3. For each year, sort bins by rate range start.
    4. Format: "Year {year}: {range}: {count} members".

    Args:
        dot_plot_distribution (Optional[Dict]): The raw dot plot structure.

    Returns:
        Optional[str]: The verbalized string, or None if input is None.
    """
    if dot_plot_distribution is None:
        return None

    projections = dot_plot_distribution.get("projections", {})
    if not projections:
        return ""

    lines = []

    # Sort Years: Numeric years first, then 'longer_run'
    def year_sort_key(y: str) -> Tuple[int, int]:
        # (0, year) for numeric, (1, 0) for longer_run/other
        if str(y).isdigit():
            return (0, int(y))
        return (1, 0)

    sorted_years = sorted(projections.keys(), key=year_sort_key)

    for year in sorted_years:
        bins_dict = projections[year].get("bins", {})
        if not bins_dict:
            continue

        # Sort Bins: Parse "3.75-4.00" -> 3.75
        def bin_sort_key(range_str: str) -> float:
            # Extract first float
            match = re.match(r"^([\d\.]+)", str(range_str))
            if match:
                try:
                    return float(match.group(1))
                except ValueError:
                    return 999.0 # Push malformed to end
            return 999.0

        sorted_ranges = sorted(bins_dict.keys(), key=bin_sort_key)

        for rng in sorted_ranges:
            count = bins_dict[rng]
            # Format: "Year 2023: 5.00-5.25: 3 members"
            lines.append(f"Year {year}: {rng}: {count} members")

    return "\n".join(lines)


# ------------------------------------------------------------------------------
# Task 7, Step 2: Implement the dot_plot_missing_policy handler
# ------------------------------------------------------------------------------
def apply_dot_plot_policy(
    agent_packets: Dict[pd.Timestamp, Dict[str, Any]]
) -> Tuple[Dict[pd.Timestamp, Dict[str, Any]], List[Dict[str, Any]]]:
    """
    Applies the 'use_latest_available_prior_sep' policy to impute missing Dot Plots.

    Logic:
    - Sort meetings chronologically.
    - Maintain `last_valid_text`.
    - If current meeting has Dot Plot, verbalize and update `last_valid_text`.
    - If missing, use `last_valid_text` (or empty string if none exists).

    Args:
        agent_packets (Dict): Dictionary of meeting packets keyed by date.

    Returns:
        Tuple[Dict, List]: Updated packets and provenance log.
    """
    # Sort dates to ensure chronological processing
    sorted_dates = sorted(agent_packets.keys())

    last_valid_text: str = ""
    last_valid_date: Optional[pd.Timestamp] = None
    provenance_log = []

    for date in sorted_dates:
        packet = agent_packets[date]
        raw_dist = packet.get("dot_plot_distribution")

        # Step 1: Verbalize current if exists
        current_text = verbalize_dot_plot(raw_dist)

        if current_text is not None:
            # We have a new SEP
            packet["dot_plot_text"] = current_text
            last_valid_text = current_text
            last_valid_date = date
            provenance_log.append({
                "meeting_date": date,
                "action": "verbalized_new",
                "source_date": date
            })
        else:
            # Missing SEP -> Impute
            packet["dot_plot_text"] = last_valid_text
            provenance_log.append({
                "meeting_date": date,
                "action": "imputed_from_prior",
                "source_date": last_valid_date if last_valid_date else "None (Start of Series)"
            })

    return agent_packets, provenance_log


# ------------------------------------------------------------------------------
# Task 7, Step 3: Store the verbalized Dot Plot text (Implicit in Step 2)
# ------------------------------------------------------------------------------
# Note: Step 2 modifies the packets in-place to add 'dot_plot_text'.
# This step is conceptually distinct but implemented within the iteration loop
# of Step 2 for efficiency and correctness (state dependence).


# ------------------------------------------------------------------------------
# Task 7, Orchestrator Function
# ------------------------------------------------------------------------------
def preprocess_dot_plots(
    agent_packets: Dict[pd.Timestamp, Dict[str, Any]]
) -> Dict[str, Any]:
    """
    Orchestrates Dot Plot preprocessing: verbalization and missing value imputation.

    Args:
        agent_packets (Dict): Agent input packets.

    Returns:
        Dict[str, Any]:
            - agent_packets: Updated packets with 'dot_plot_text'.
            - provenance_log: Log of imputation actions.
    """
    # Apply policy (which includes verbalization and storage)
    updated_packets, log = apply_dot_plot_policy(agent_packets)

    return {
        "agent_packets": updated_packets,
        "provenance_log": log
    }
````

````python
# Task 8 — Preprocess remaining unstructured inputs and construct agent context bundle

# ===================================================================================
# Task 8: Preprocess remaining unstructured inputs and construct agent context bundle
# ===================================================================================

# ------------------------------------------------------------------------------
# Task 8, Step 1: Add instructional headers to each unstructured text field
# ------------------------------------------------------------------------------
def add_instructional_headers(
    agent_packets: Dict[pd.Timestamp, Dict[str, Any]]
) -> Dict[pd.Timestamp, Dict[str, Any]]:
    """
    Prepends fixed instructional headers to unstructured text fields and renders
    FedWatch probabilities into text.

    Headers:
    - Beige Book: "### BEIGE BOOK SUMMARY\nThe following is the National Summary..."
    - Dot Plot: "### DOT PLOT DISTRIBUTION\nThe following shows FOMC participants'..."
    - FedWatch: "### FEDWATCH MARKET PROBABILITIES\n"

    Args:
        agent_packets (Dict): Meeting packets.

    Returns:
        Dict: Updated packets with '_headered' fields.
    """
    # Define immutable headers
    BEIGE_BOOK_HEADER = (
        "### BEIGE BOOK SUMMARY\n"
        "The following is the National Summary from the Federal Reserve's Beige Book:\n\n"
    )
    DOT_PLOT_HEADER = (
        "### DOT PLOT DISTRIBUTION\n"
        "The following shows FOMC participants' rate projections:\n\n"
    )
    FEDWATCH_HEADER = (
        "### FEDWATCH MARKET PROBABILITIES\n"
    )

    for date, packet in agent_packets.items():
        # 1. Beige Book
        bb_text = packet.get("beige_book_raw_text", "")
        packet["beige_book_headered"] = f"{BEIGE_BOOK_HEADER}{bb_text}"

        # 2. Dot Plot
        dp_text = packet.get("dot_plot_text", "")
        packet["dot_plot_headered"] = f"{DOT_PLOT_HEADER}{dp_text}"

        # 3. FedWatch
        fw_text = packet.get("fedwatch_raw_text", "")
        probs = packet.get("fedwatch_probabilities", {})

        # Render probabilities deterministically
        # Sort keys alphabetically
        sorted_keys = sorted(probs.keys())
        probs_str_parts = []
        for k in sorted_keys:
            val = probs[k]
            # Format: "Hike: 10.5%"
            probs_str_parts.append(f"{k}: {val*100:.1f}%")

        probs_text = ", ".join(probs_str_parts)

        # Combine: Header + Raw Text + Rendered Probs
        packet["fedwatch_headered"] = (
            f"{FEDWATCH_HEADER}"
            f"{fw_text}\n"
            f"Market Implied Probabilities: {probs_text}"
        )

    return agent_packets


# ------------------------------------------------------------------------------
# Task 8, Step 2: Construct the unified unstructured_context bundle
# ------------------------------------------------------------------------------
def construct_unstructured_bundle(
    agent_packets: Dict[pd.Timestamp, Dict[str, Any]]
) -> Tuple[Dict[pd.Timestamp, Dict[str, Any]], List[Dict[str, Any]]]:
    """
    Concatenates headered unstructured fields into a single context string.

    Order:
    1. Beige Book
    2. Dot Plot
    3. FedWatch

    Args:
        agent_packets (Dict): Meeting packets with headered fields.

    Returns:
        Tuple[Dict, List]: Updated packets with 'unstructured_context' and a log of lengths.
    """
    context_log = []

    for date, packet in agent_packets.items():
        parts = [
            packet["beige_book_headered"],
            packet["dot_plot_headered"],
            packet["fedwatch_headered"]
        ]

        # Join with double newlines for clear separation
        full_context = "\n\n".join(parts)
        packet["unstructured_context"] = full_context

        context_log.append({
            "meeting_date": date,
            "context_length_chars": len(full_context)
        })

    return agent_packets, context_log


# ------------------------------------------------------------------------------
# Task 8, Step 3: Construct the structured_macro_text string
# ------------------------------------------------------------------------------
def construct_structured_macro_text(
    agent_packets: Dict[pd.Timestamp, Dict[str, Any]]
) -> Dict[pd.Timestamp, Dict[str, Any]]:
    """
    Serializes structured macro variables into a human-readable string for prompt injection.

    Format: "Field Name: Value Unit"

    Args:
        agent_packets (Dict): Meeting packets with structured fields.

    Returns:
        Dict: Updated packets with 'structured_macro_text'.
    """
    # Define field order and formatting rules
    # (Field Key, Display Name, Unit Suffix, Precision)
    field_spec = [
        ("pce_index_yoy", "PCE Index (YoY)", "%", 2),
        ("cpi_index_yoy", "CPI Index (YoY)", "%", 2),
        ("inflation_expectations_1yr", "Inflation Expectations (1yr)", "%", 2),
        ("treasury_yield_3m", "3-Month Treasury Yield", "%", 2),
        ("treasury_yield_6m", "6-Month Treasury Yield", "%", 2),
        ("m2_supply_level_sa", "M2 Supply Level", "", 1),
        ("bbk_real_gdp_nowcast", "Real GDP Nowcast", "%", 2),
        ("unemployment_rate_u3", "Unemployment Rate (U-3)", "%", 2),
        ("vix_index", "VIX Index", "", 2),
        ("fed_chair", "Fed Chair", "", None), # Categorical
        ("white_house_party", "White House Party", "", None), # Categorical
        ("previous_fftr_target", "Previous FFTR Target", "bps", 0),
        ("previous_change", "Previous Change", "bps", 0)
    ]

    for date, packet in agent_packets.items():
        lines = []
        for key, name, unit, prec in field_spec:
            val = packet.get(key)

            # Format value
            if val is None:
                val_str = "N/A"
            elif prec is not None:
                # Numeric formatting
                try:
                    val_str = f"{float(val):.{prec}f}"
                except (ValueError, TypeError):
                    val_str = str(val)
            else:
                # String/Categorical
                val_str = str(val)

            # Append unit if present
            if unit:
                line = f"{name}: {val_str}{unit}"
            else:
                line = f"{name}: {val_str}"

            lines.append(line)

        packet["structured_macro_text"] = "\n".join(lines)

    return agent_packets


# ------------------------------------------------------------------------------
# Task 8, Orchestrator Function
# ------------------------------------------------------------------------------
def preprocess_agent_contexts(
    agent_packets: Dict[pd.Timestamp, Dict[str, Any]]
) -> Dict[str, Any]:
    """
    Orchestrates the construction of text contexts for agents.

    Pipeline:
    1. Add headers to unstructured text.
    2. Bundle unstructured text into single context.
    3. Serialize structured macro data into text.

    Args:
        agent_packets (Dict): Meeting packets.

    Returns:
        Dict[str, Any]:
            - agent_packets: Fully preprocessed packets ready for prompt injection.
            - context_log: Log of context lengths.
    """
    # Step 1: Headers
    packets_headered = add_instructional_headers(agent_packets)

    # Step 2: Unstructured Bundle
    packets_bundled, context_log = construct_unstructured_bundle(packets_headered)

    # Step 3: Structured Text
    packets_final = construct_structured_macro_text(packets_bundled)

    return {
        "agent_packets": packets_final,
        "context_log": context_log
    }
````

````python
# Task 9 — Cluster FOMC participants into three representative archetypes

# ==============================================================================
# Task 9: Cluster FOMC participants into three representative archetypes
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 9, Step 1: Prepare the feature matrix for clustering
# ------------------------------------------------------------------------------
def prepare_clustering_features(
    df_participants_clean: pd.DataFrame,
    feature_spec: Dict[str, List[str]]
) -> Tuple[np.ndarray, List[str], Any]:
    """
    Prepares the feature matrix for clustering using One-Hot Encoding and Standardization.

    Process:
    1. Select numeric and categorical features based on spec.
    2. One-Hot Encode categorical features (dense).
    3. Standardize (Z-score scale) ALL features (numeric + encoded categorical) to
       ensure equal weighting in Euclidean distance.

    Args:
        df_participants_clean (pd.DataFrame): Cleansed participant data.
        feature_spec (Dict): Dictionary with 'numeric' and 'categorical' keys.

    Returns:
        Tuple[np.ndarray, List[str], Any]:
            - X: Processed feature matrix (numpy array).
            - feature_names: List of feature names corresponding to X columns.
            - preprocessor: Fitted sklearn ColumnTransformer/Pipeline.
    """
    numeric_cols = sorted(feature_spec["numeric"])
    categorical_cols = sorted(feature_spec["categorical"])

    # Define Preprocessing Pipeline
    # 1. Categorical: OneHotEncoder
    # 2. Numeric: Passthrough (scaled later globally)
    # Note: We scale *after* encoding so that binary features are also on comparable scale
    preprocessor = ColumnTransformer(
        transformers=[
            ('num', 'passthrough', numeric_cols),
            ('cat', OneHotEncoder(sparse_output=False, handle_unknown='ignore'), categorical_cols)
        ],
        verbose_feature_names_out=False
    )

    # Fit and Transform to get intermediate matrix
    X_intermediate = preprocessor.fit_transform(df_participants_clean)

    # Get feature names
    feature_names = list(preprocessor.get_feature_names_out())

    # Global Scaling
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X_intermediate)

    # Bundle the full preprocessing logic
    full_preprocessor = Pipeline([
        ('transformer', preprocessor),
        ('scaler', scaler)
    ])
    # Re-fit pipeline to ensure it captures the full state (though we did it in steps)
    full_preprocessor.fit(df_participants_clean)

    return X_scaled, feature_names, full_preprocessor


# ------------------------------------------------------------------------------
# Task 9, Step 2: Execute clustering with K=3
# ------------------------------------------------------------------------------
def execute_clustering(
    X: np.ndarray,
    participant_ids: List[Any]
) -> Tuple[Dict[Any, int], np.ndarray, Any]:
    """
    Executes K-Means clustering with K=3.

    Args:
        X (np.ndarray): Scaled feature matrix.
        participant_ids (List): List of participant IDs corresponding to rows of X.

    Returns:
        Tuple[Dict, np.ndarray, Any]:
            - labels_map: Dict mapping participant_id -> cluster_id.
            - centroids: Array of cluster centers (shape 3, n_features).
            - kmeans_model: Fitted KMeans object.
    """
    # Configuration: K=3, Seed=42
    kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
    kmeans.fit(X)

    labels = kmeans.labels_
    centroids = kmeans.cluster_centers_

    # Map labels to IDs
    labels_map = {pid: int(lbl) for pid, lbl in zip(participant_ids, labels)}

    return labels_map, centroids, kmeans


# ------------------------------------------------------------------------------
# Task 9, Step 3: Map cluster IDs to archetype names
# ------------------------------------------------------------------------------
def map_clusters_to_archetypes(
    centroids: np.ndarray,
    feature_names: List[str],
    X: np.ndarray,
    labels_map: Dict[Any, int]
) -> Tuple[Dict[int, str], Dict[str, Any]]:
    """
    Maps cluster IDs (0, 1, 2) to semantic archetype names based on centroid characteristics.

    Rules:
    1. Regional Pragmatists: Highest 'Regional Affiliation' (specifically 'President' category).
    2. Academic Balancers: Highest ('Hawkishness' + 'Focus on Inflation').
    3. Central Policymakers: Highest 'Tenure' + Lowest Dispersion.

    Args:
        centroids (np.ndarray): Cluster centroids.
        feature_names (List[str]): Feature names aligned with centroid columns.
        X (np.ndarray): Feature matrix (for dispersion calc).
        labels_map (Dict): Participant labels.

    Returns:
        Tuple[Dict, Dict]:
            - mapping: Dict {cluster_id: "Archetype Name"}.
            - rationale: Dict with scores used for assignment.
    """
    # Identify feature indices
    try:
        idx_hawk = feature_names.index("hawkishness_score")
        idx_inf = feature_names.index("focus_on_inflation")
        idx_tenure = feature_names.index("tenure_years")

        # Find Regional Affiliation column (likely 'regional_affiliation_President' or similar)
        # We look for the column indicating "Regional President" status (vs Governor)
        # Based on Task 5 cleaning, we mapped 0->Governor, 1->President.
        # OneHotEncoder likely produced 'regional_affiliation_President'.
        regional_cols = [f for f in feature_names if "regional_affiliation" in f.lower() and "president" in f.lower()]
        if not regional_cols:
            # Fallback: look for any regional affiliation col that isn't Governor
            regional_cols = [f for f in feature_names if "regional_affiliation" in f.lower() and "governor" not in f.lower()]

        idx_regional = feature_names.index(regional_cols[0]) if regional_cols else -1

    except ValueError as e:
        raise ValueError(f"Critical feature missing from clustering input: {e}")

    # Calculate Scores per Cluster
    scores = {}
    n_clusters = centroids.shape[0]

    # Calculate Dispersion (Mean Squared Distance to Centroid)
    dispersions = []
    labels_array = np.array(list(labels_map.values()))
    for k in range(n_clusters):
        cluster_points = X[labels_array == k]
        if len(cluster_points) > 0:
            dist = np.mean(np.linalg.norm(cluster_points - centroids[k], axis=1))
        else:
            dist = np.inf
        dispersions.append(dist)

    # 1. Regional Pragmatists Score: Regional Affiliation Centroid Value
    reg_scores = centroids[:, idx_regional] if idx_regional != -1 else np.zeros(n_clusters)

    # 2. Academic Balancers Score: Hawkishness + Inflation Focus
    acad_scores = centroids[:, idx_hawk] + centroids[:, idx_inf]

    # 3. Central Policymakers Score: Tenure - Dispersion (Higher tenure, lower dispersion is better)
    # We normalize dispersion to be comparable or just use heuristic
    # Heuristic: Tenure Z-score minus Dispersion (since dispersion is positive distance)
    central_scores = centroids[:, idx_tenure] - np.array(dispersions)

    # Assignment Logic (Greedy or Optimization)
    # We need 1-to-1 mapping.
    # Construct a cost matrix or just sort. Given 3x3, we can check permutations.
    archetypes = ["Regional Pragmatists", "Academic Balancers", "Central Policymakers"]
    best_mapping = {}
    best_total_score = -np.inf

    # Permutations of (0,1,2) assigned to (Reg, Acad, Central)
    # p[0] is cluster ID for Reg, p[1] for Acad, p[2] for Central
    for p in itertools.permutations(range(n_clusters)):
        s_reg = reg_scores[p[0]]
        s_acad = acad_scores[p[1]]
        s_central = central_scores[p[2]]
        total = s_reg + s_acad + s_central

        if total > best_total_score:
            best_total_score = total
            best_mapping = {
                p[0]: "Regional Pragmatists",
                p[1]: "Academic Balancers",
                p[2]: "Central Policymakers"
            }

    rationale = {
        "regional_scores": reg_scores.tolist(),
        "academic_scores": acad_scores.tolist(),
        "central_scores": central_scores.tolist(),
        "dispersions": dispersions,
        "final_mapping": best_mapping
    }

    return best_mapping, rationale


# ------------------------------------------------------------------------------
# Task 9, Orchestrator Function
# ------------------------------------------------------------------------------
def cluster_participants(
    df_participants_clean: pd.DataFrame,
    feature_spec: Dict[str, List[str]]
) -> Dict[str, Any]:
    """
    Orchestrates the clustering of participants into archetypes.

    Args:
        df_participants_clean (pd.DataFrame): Cleansed data.
        feature_spec (Dict): Feature specification.

    Returns:
        Dict[str, Any]:
            - labels_map: Participant ID -> Cluster ID.
            - archetype_mapping: Cluster ID -> Archetype Name.
            - clustering_artifact: Model, scaler, centroids.
            - rationale: Assignment logic scores.
    """
    # Step 1: Prepare Features
    X, feature_names, preprocessor = prepare_clustering_features(df_participants_clean, feature_spec)

    # Step 2: Cluster
    participant_ids = df_participants_clean.index.tolist()
    labels_map, centroids, kmeans_model = execute_clustering(X, participant_ids)

    # Step 3: Map to Archetypes
    archetype_mapping, rationale = map_clusters_to_archetypes(
        centroids, feature_names, X, labels_map
    )

    return {
        "labels_map": labels_map,
        "archetype_mapping": archetype_mapping,
        "clustering_artifact": {
            "model": kmeans_model,
            "preprocessor": preprocessor,
            "centroids": centroids,
            "feature_names": feature_names
        },
        "rationale": rationale
    }
````

````python
# Task 10 — Materialize the three Member agent personas from archetypes

# ==============================================================================
# Task 10: Materialize the three Member agent personas from archetypes
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 10, Step 1: Retrieve archetype descriptions from configuration
# ------------------------------------------------------------------------------
def retrieve_archetype_descriptions(
    agent_config: Dict[str, Any]
) -> List[Dict[str, str]]:
    """
    Retrieves and validates archetype descriptions from the configuration.

    Args:
        agent_config (Dict): The 'agent_architecture_config' section.

    Returns:
        List[Dict]: List of dicts with 'name' and 'description', sorted by name
        to ensure deterministic ordering.

    Raises:
        ValueError: If required archetypes are missing or descriptions are empty.
    """
    required_names = set(agent_config.get("archetype_names", []))
    archetypes_dict = agent_config.get("archetypes", {})

    # Flatten the config structure (which might be keyed by cluster_id or arbitrary keys)
    # We look for values that contain 'name' and 'description'
    found_archetypes = []

    for key, data in archetypes_dict.items():
        name = data.get("name")
        desc = data.get("description")

        if not name or not desc:
            raise ValueError(f"Archetype config entry '{key}' missing name or description.")

        if name in required_names:
            found_archetypes.append({"name": name, "description": desc})

    # Validate coverage
    found_names = {a["name"] for a in found_archetypes}
    if found_names != required_names:
        missing = required_names - found_names
        raise ValueError(f"Missing configuration for required archetypes: {missing}")

    # Sort deterministically by name (or by config list order if preferred, but name is safer)
    # We use the order in 'archetype_names' list to define the canonical order
    ordered_archetypes = []
    for name in agent_config["archetype_names"]:
        # Find the matching record
        record = next(a for a in found_archetypes if a["name"] == name)
        ordered_archetypes.append(record)

    return ordered_archetypes


# ------------------------------------------------------------------------------
# Task 10, Step 2: Instantiate the Member system persona prompt
# ------------------------------------------------------------------------------
def instantiate_persona_prompts(
    archetype_records: List[Dict[str, str]],
    persona_template: str
) -> List[Dict[str, str]]:
    """
    Formats the system persona template with archetype details.

    Args:
        archetype_records (List[Dict]): List of archetype info.
        persona_template (str): The Jinja2-style or f-string template.

    Returns:
        List[Dict]: Archetype records augmented with 'system_prompt' and 'prompt_hash'.
    """
    augmented_records = []

    for record in archetype_records:
        # Format
        # We assume the template uses {archetype_name} and {archetype_description}
        try:
            prompt = persona_template.format(
                archetype_name=record["name"],
                archetype_description=record["description"]
            )
        except KeyError as e:
            raise ValueError(f"Template formatting failed. Missing key in archetype record: {e}")

        # Hash
        p_hash = hashlib.sha256(prompt.encode("utf-8")).hexdigest()

        new_record = record.copy()
        new_record["system_prompt"] = prompt
        new_record["prompt_hash"] = p_hash
        augmented_records.append(new_record)

    return augmented_records


# ------------------------------------------------------------------------------
# Task 10, Step 3: Store the three instantiated persona prompts as agent identities
# ------------------------------------------------------------------------------
def create_member_agent_registry(
    augmented_records: List[Dict[str, str]]
) -> List[Dict[str, Any]]:
    """
    Creates the final registry of member agents, assigning stable IDs.

    Args:
        augmented_records (List[Dict]): Archetypes with prompts.

    Returns:
        List[Dict]: Registry entries with 'agent_id', 'archetype_name', 'system_prompt'.
    """
    registry = []

    # Assign IDs based on list order (which was deterministically sorted in Step 1)
    for i, record in enumerate(augmented_records):
        agent_id = f"member_{i}"

        entry = {
            "agent_id": agent_id,
            "archetype_name": record["name"],
            "system_prompt": record["system_prompt"],
            "system_prompt_hash": record["prompt_hash"]
        }
        registry.append(entry)

    if len(registry) != 3:
        raise ValueError(f"Registry must contain exactly 3 agents, got {len(registry)}.")

    return registry


# ------------------------------------------------------------------------------
# Task 10, Orchestrator Function
# ------------------------------------------------------------------------------
def materialize_agents(
    config: Dict[str, Any]
) -> List[Dict[str, Any]]:
    """
    Orchestrates the creation of member agent personas.

    Args:
        config (Dict): Full study configuration.

    Returns:
        List[Dict]: The member agent registry.
    """
    agent_config = config["agent_architecture_config"]
    template = config["prompt_templates"]["member_system_persona"]

    # Step 1: Retrieve
    archetypes = retrieve_archetype_descriptions(agent_config)

    # Step 2: Instantiate
    prompts = instantiate_persona_prompts(archetypes, template)

    # Step 3: Register
    registry = create_member_agent_registry(prompts)

    return registry
````

````python
# Task 11 — Define the LLM client interface with retry logic and logging

# ==============================================================================
# Task 11: Define the LLM client interface with retry logic and logging
# ==============================================================================

# ------------------------------------------------------------------------------
# Data Structures
# ------------------------------------------------------------------------------
@dataclass
class LLMInvocationPolicy:
    """
    Immutable record of the policy used for LLM calls.

    This dataclass encapsulates all configuration parameters governing the behavior
    of the Large Language Model (LLM) client, ensuring that decoding strategies,
    retry logic, and logging preferences are fixed and reproducible for a given
    experiment run.

    Attributes:
        model_name (str): The identifier of the LLM model to be used (e.g., "gpt-4o").
        temperature (float): The sampling temperature, controlling randomness (0.0 to 2.0).
        top_p (float): The nucleus sampling probability mass (0.0 to 1.0).
        max_tokens (int): The maximum number of tokens to generate in the completion.
        request_timeout_s (float): The maximum time in seconds to wait for a response.
        max_retries (int): The maximum number of retry attempts for transient errors.
        retry_backoff_base_s (float): The base wait time in seconds for exponential backoff.
        retry_backoff_max_s (float): The maximum wait time in seconds for exponential backoff.
        seed (Optional[int]): An optional integer seed for deterministic sampling (best-effort).
        log_prompts (bool): Whether to log the full text of prompts sent to the model.
        log_responses (bool): Whether to log the full text of responses received from the model.
    """
    model_name: str
    temperature: float
    top_p: float
    max_tokens: int
    request_timeout_s: float
    max_retries: int
    retry_backoff_base_s: float
    retry_backoff_max_s: float
    seed: Optional[int]
    log_prompts: bool
    log_responses: bool


@dataclass
class LLMCallRecord:
    """
    Structured log entry for a single LLM interaction.

    This dataclass represents a comprehensive audit record for an individual call
    to the LLM API. It captures metadata for traceability, cryptographic hashes
    for integrity verification, and token usage statistics for cost accounting.

    Attributes:
        call_id (str): A unique UUID string identifying this specific API call.
        timestamp (str): The ISO 8601 formatted timestamp of when the call was initiated.
        variant (str): The experimental variant identifier (e.g., "baseline", "icl", "cod").
        meeting_id (str): The canonical identifier for the FOMC meeting being simulated.
        run_id (int): The simulation run index (e.g., 1 to 5).
        role (str): The agent role performing the action (e.g., "Analyst", "Economist").
        agent_id (Optional[str]): The specific agent identifier (e.g., "member_0"), if applicable.
        attempt_count (int): The retry attempt number (1 for the first attempt).
        model_name (str): The name of the model used for this call.
        prompt_hash (str): The SHA-256 hash of the full prompt text.
        response_hash (str): The SHA-256 hash of the full response text.
        prompt_text (Optional[str]): The full text of the prompt, if logging is enabled.
        response_text (Optional[str]): The full text of the response, if logging is enabled.
        prompt_tokens (int): The number of tokens in the prompt, as reported by the API.
        completion_tokens (int): The number of tokens in the completion, as reported by the API.
        total_tokens (int): The total token count (prompt + completion), as reported by the API.
        finish_reason (str): The reason the model stopped generating (e.g., "stop", "length").
        error (Optional[str]): An error message string if the call failed, otherwise None.
    """
    call_id: str
    timestamp: str
    variant: str
    meeting_id: str
    run_id: int
    role: str
    agent_id: Optional[str]
    attempt_count: int
    model_name: str
    prompt_hash: str
    response_hash: str
    prompt_text: Optional[str]
    response_text: Optional[str]
    prompt_tokens: int
    completion_tokens: int
    total_tokens: int
    finish_reason: str
    error: Optional[str]

# ------------------------------------------------------------------------------
# Task 11, Step 1: Configure the LLM client deterministically
# ------------------------------------------------------------------------------
def configure_llm_client(
    llm_config: Dict[str, Any]
) -> Tuple[AsyncOpenAI, LLMInvocationPolicy]:
    """
    Configures the AsyncOpenAI client and freezes the invocation policy.

    Args:
        llm_config (Dict): Validated LLM configuration dictionary.

    Returns:
        Tuple[AsyncOpenAI, LLMInvocationPolicy]: Configured client and policy object.
    """
    # Extract parameters
    api_key = llm_config["api_key"]
    model_name = llm_config["model_name"]

    # Create Client
    client = AsyncOpenAI(
        api_key=api_key,
        timeout=llm_config.get("request_timeout_s", 120.0),
        max_retries=0 # We implement custom retry logic
    )

    # Create Policy Object
    policy = LLMInvocationPolicy(
        model_name=model_name,
        temperature=llm_config["temperature"],
        top_p=llm_config["top_p"],
        max_tokens=llm_config["max_tokens_output"],
        request_timeout_s=llm_config.get("request_timeout_s", 120.0),
        max_retries=llm_config["max_retries"],
        retry_backoff_base_s=llm_config["retry_backoff_base_s"],
        retry_backoff_max_s=llm_config["retry_backoff_max_s"],
        seed=llm_config.get("seed"),
        log_prompts=llm_config["logging"]["log_prompts"],
        log_responses=llm_config["logging"]["log_responses"]
    )

    return client, policy


# ------------------------------------------------------------------------------
# Task 11, Step 2 & 3: Implement retry wrapper and logging
# ------------------------------------------------------------------------------

class LLMClientWrapper:
    """
    Wrapper around AsyncOpenAI to enforce retry policy and structured logging.

    This class encapsulates the interaction with the OpenAI API, providing a robust
    layer that handles transient errors via exponential backoff and logs all
    interactions for auditability and reproducibility. It ensures that every call
    adheres to the configured `LLMInvocationPolicy`.

    Attributes:
        client (AsyncOpenAI): The configured asynchronous OpenAI client instance.
        policy (LLMInvocationPolicy): The immutable policy configuration governing retries, timeouts, and logging.
        log_path (str): The file path where structured JSONL logs will be appended.
    """

    def __init__(self, client: AsyncOpenAI, policy: LLMInvocationPolicy, log_path: str = "llm_logs.jsonl"):
        """
        Initializes the LLMClientWrapper.

        Args:
            client (AsyncOpenAI): An initialized AsyncOpenAI client.
            policy (LLMInvocationPolicy): The policy object containing configuration parameters.
            log_path (str): Path to the log file. Defaults to "llm_logs.jsonl".
        """
        self.client = client
        self.policy = policy
        self.log_path = log_path

        # Ensure log file exists or is accessible
        # In production, we might use a proper logging handler, here we append to file
        pass

    async def call_completion(
        self,
        prompt: str,
        system_prompt: str,
        metadata: Dict[str, Any]
    ) -> Dict[str, Any]:
        """
        Executes a chat completion with exponential backoff retries and logging.

        This method attempts to generate a completion from the LLM. If a transient error
        (RateLimit, Connection, InternalServer) occurs, it retries according to the
        exponential backoff schedule defined in the policy:
        wait_time = min(base * 2^(attempt-1), max_wait).

        All attempts, successes, and failures are logged to the configured log file.

        Args:
            prompt (str): The user prompt string to send to the model.
            system_prompt (str): The system instruction string defining the agent's persona/task.
            metadata (Dict[str, Any]): Contextual metadata for logging (e.g., variant, meeting_id, run_id, role, agent_id).

        Returns:
            Dict[str, Any]: A dictionary containing:
                - "content": The text content of the model's response.
                - "usage": A dictionary of token usage statistics (prompt, completion, total).
                - "model_version": The specific model version reported by the API.

        Raises:
            Exception: If the maximum number of retries is exhausted or if a non-retryable error occurs.
        """
        call_id = str(uuid.uuid4())
        attempt = 0
        last_error = None

        # Prepare request args
        request_kwargs = {
            "model": self.policy.model_name,
            "messages": [
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": prompt}
            ],
            "temperature": self.policy.temperature,
            "top_p": self.policy.top_p,
            "max_tokens": self.policy.max_tokens,
        }
        if self.policy.seed is not None:
            request_kwargs["seed"] = self.policy.seed

        while attempt <= self.policy.max_retries:
            try:
                attempt += 1
                start_time = datetime.utcnow().isoformat()

                response = await self.client.chat.completions.create(**request_kwargs)

                # Success
                content = response.choices[0].message.content
                usage = response.usage
                finish_reason = response.choices[0].finish_reason

                # Log Success
                self._log_call(
                    call_id=call_id,
                    timestamp=start_time,
                    metadata=metadata,
                    prompt=prompt,
                    system_prompt=system_prompt,
                    response_text=content,
                    usage=usage,
                    finish_reason=finish_reason,
                    attempt=attempt,
                    error=None
                )

                return {
                    "content": content,
                    "usage": {
                        "prompt_tokens": usage.prompt_tokens,
                        "completion_tokens": usage.completion_tokens,
                        "total_tokens": usage.total_tokens
                    },
                    "model_version": response.model # Provider reported version
                }

            except (RateLimitError, APIConnectionError, InternalServerError) as e:
                # Retryable
                last_error = e
                if attempt > self.policy.max_retries:
                    break

                # Backoff: min(base * 2^(attempt-1), max)
                # attempt starts at 1, so exponent is attempt-1
                wait_time = min(
                    self.policy.retry_backoff_base_s * (2 ** (attempt - 1)),
                    self.policy.retry_backoff_max_s
                )

                # Log Retry
                self._log_retry(call_id, attempt, str(e), wait_time)

                await asyncio.sleep(wait_time)

            except (AuthenticationError, BadRequestError) as e:
                # Non-retryable
                self._log_call(
                    call_id=call_id,
                    timestamp=datetime.utcnow().isoformat(),
                    metadata=metadata,
                    prompt=prompt,
                    system_prompt=system_prompt,
                    response_text=None,
                    usage=None,
                    finish_reason="error",
                    attempt=attempt,
                    error=str(e)
                )
                raise e
            except Exception as e:
                # Unknown error - treat as non-retryable or retryable?
                # Safest is to re-raise if not explicitly retryable
                self._log_call(
                    call_id=call_id,
                    timestamp=datetime.utcnow().isoformat(),
                    metadata=metadata,
                    prompt=prompt,
                    system_prompt=system_prompt,
                    response_text=None,
                    usage=None,
                    finish_reason="error",
                    attempt=attempt,
                    error=str(e)
                )
                raise e

        # If loop finishes without return, retries exhausted
        raise Exception(f"Max retries exhausted for call {call_id}. Last error: {last_error}")

    def _log_call(
        self,
        call_id: str,
        timestamp: str,
        metadata: Dict[str, Any],
        prompt: str,
        system_prompt: str,
        response_text: Optional[str],
        usage: Any,
        finish_reason: str,
        attempt: int,
        error: Optional[str]
    ) -> None:
        """
        Writes a structured log entry for an LLM call to the log file.

        This method constructs an `LLMCallRecord`, serializes it to JSON, and appends it
        to the configured log file. It handles hashing of prompts and responses for
        integrity verification and respects the logging policy regarding full text storage.

        Args:
            call_id (str): The unique identifier for the call.
            timestamp (str): The ISO 8601 timestamp of the call.
            metadata (Dict[str, Any]): Contextual metadata (variant, meeting_id, etc.).
            prompt (str): The user prompt text.
            system_prompt (str): The system prompt text.
            response_text (Optional[str]): The model's response text (None if error).
            usage (Any): The usage object from the API response (None if error).
            finish_reason (str): The finish reason reported by the API (or "error").
            attempt (int): The attempt number for this call.
            error (Optional[str]): The error message if the call failed.
        """
        full_prompt = f"System: {system_prompt}\nUser: {prompt}"
        prompt_hash = hashlib.sha256(full_prompt.encode("utf-8")).hexdigest()
        response_hash = hashlib.sha256(response_text.encode("utf-8")).hexdigest() if response_text else "N/A"

        record = LLMCallRecord(
            call_id=call_id,
            timestamp=timestamp,
            variant=metadata.get("variant", "unknown"),
            meeting_id=str(metadata.get("meeting_id", "unknown")),
            run_id=metadata.get("run_id", -1),
            role=metadata.get("role", "unknown"),
            agent_id=metadata.get("agent_id"),
            attempt_count=attempt,
            model_name=self.policy.model_name,
            prompt_hash=prompt_hash,
            response_hash=response_hash,
            prompt_text=full_prompt if self.policy.log_prompts else None,
            response_text=response_text if self.policy.log_responses else None,
            prompt_tokens=usage.prompt_tokens if usage else 0,
            completion_tokens=usage.completion_tokens if usage else 0,
            total_tokens=usage.total_tokens if usage else 0,
            finish_reason=finish_reason,
            error=error
        )

        # Append to JSONL
        with open(self.log_path, "a", encoding="utf-8") as f:
            f.write(json.dumps(asdict(record)) + "\n")

    def _log_retry(self, call_id: str, attempt: int, error_msg: str, wait_time: float) -> None:
        """
        Logs a retry attempt.

        This method records the occurrence of a transient error that triggered a retry,
        including the wait time imposed by the backoff policy.

        Args:
            call_id (str): The unique identifier for the call.
            attempt (int): The attempt number that failed.
            error_msg (str): The error message that triggered the retry.
            wait_time (float): The duration in seconds to wait before the next attempt.
        """
        entry = {
            "event": "retry",
            "call_id": call_id,
            "attempt": attempt,
            "error": error_msg,
            "wait_time_s": wait_time,
            "timestamp": datetime.utcnow().isoformat()
        }
        # For now, just print or log to a separate debug file
        # print(json.dumps(entry))
        pass

# ------------------------------------------------------------------------------
# Task 11, Orchestrator Function
# ------------------------------------------------------------------------------
def setup_llm_infrastructure(
    config: Dict[str, Any],
    log_path: str = "llm_logs.jsonl"
) -> LLMClientWrapper:
    """
    Orchestrates the setup of the LLM client infrastructure.

    Args:
        config (Dict): Full study configuration.
        log_path (str): Path to store LLM logs.

    Returns:
        LLMClientWrapper: The initialized client wrapper ready for calls.
    """
    llm_config = config["llm_engine_config"]

    # Step 1: Configure Client & Policy
    client, policy = configure_llm_client(llm_config)

    # Step 2 & 3: Wrap with Retry/Logging
    wrapper = LLMClientWrapper(client, policy, log_path)

    return wrapper
````

````python
# Task 12 — Define the CrewAI agent and task architecture for FedSight AI

# ==============================================================================
# Task 12: Define the CrewAI agent and task architecture for FedSight AI
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 12, Step 1: Define the four agent roles using CrewAI Agent class
# ------------------------------------------------------------------------------

class LangChainAdapter:
    """
    Adapter class to make LLMClientWrapper compatible with CrewAI's expected LLM interface.

    CrewAI agents typically expect an LLM object that adheres to the LangChain
    interface, specifically exposing an `invoke` or `call` method. This adapter
    wraps the custom `LLMClientWrapper` (which handles retries, logging, and
    async execution) and exposes a synchronous `invoke` method that delegates
    to the wrapper. This ensures that the rigorous engineering of the underlying
    client is preserved while satisfying the architectural requirements of the
    CrewAI framework.

    Attributes:
        wrapper (LLMClientWrapper): The custom, robust LLM client wrapper instance.
    """

    def __init__(self, wrapper: Any):
        """
        Initializes the LangChainAdapter.

        Args:
            wrapper (LLMClientWrapper): The custom LLM client wrapper from Task 11,
                                        configured with retry policies and logging.
        """
        self.wrapper = wrapper

    def invoke(self, prompt: str) -> str:
        """
        Synchronous invoke method expected by CrewAI.

        This method bridges the synchronous call expected by CrewAI agents to the
        asynchronous `call_completion` method of the `LLMClientWrapper`. It manages
        the asyncio event loop to execute the async call, ensuring that the
        underlying retry logic and structured logging are executed correctly.

        Args:
            prompt (str): The prompt string to send to the LLM.

        Returns:
            str: The content of the LLM's response.

        Raises:
            RuntimeError: If there are issues managing the asyncio event loop.
            Exception: Propagates exceptions from the underlying wrapper (e.g., max retries exhausted).
        """
        # Define a dummy system prompt since CrewAI agents have their own backstories/goals
        # which act as system prompts. We pass a generic one here, or rely on the wrapper
        # to handle the distinction.
        system_prompt = "You are an AI assistant acting as an agent."

        # Metadata for logging context - in a real implementation, this might be passed dynamically
        metadata = {"role": "crewai_agent", "context": "adapter_invocation"}

        try:
            loop = asyncio.get_event_loop()
        except RuntimeError:
            loop = asyncio.new_event_loop()
            asyncio.set_event_loop(loop)

        if loop.is_running():
            # If loop is running, we might need to return a coroutine or handle differently
            # For this implementation, we assume standard execution.
            future = asyncio.ensure_future(self.wrapper.call_completion(prompt, system_prompt, metadata))
            return future
        else:
            result = loop.run_until_complete(self.wrapper.call_completion(prompt, system_prompt, metadata))
            return result["content"]

    def call(self, prompt: str, **kwargs) -> str:
        """
        Alias for invoke to support different LangChain versions.

        Some versions of LangChain or CrewAI might look for a `call` method instead
        of `invoke`. This alias ensures compatibility.

        Args:
            prompt (str): The prompt string to send to the LLM.
            **kwargs: Additional keyword arguments (ignored).

        Returns:
            str: The content of the LLM's response.
        """
        return self.invoke(prompt)


def create_fedsight_agents(
    config: Dict[str, Any],
    member_registry: List[Dict[str, Any]],
    llm_wrapper: Any
) -> Dict[str, Agent]:
    """
    Instantiates the CrewAI Agents for the FedSight workflow.

    Roles:
    1. Analyst: Interprets futures.
    2. Economist: Formulates options.
    3. Members (x3): Deliberate and vote.
    4. StatementDrafter: Drafts statement.

    Args:
        config (Dict): Study configuration containing prompt templates.
        member_registry (List[Dict]): Registry of member personas (Task 10).
        llm_wrapper (Any): The configured LLM client wrapper (Task 11).

    Returns:
        Dict[str, Agent]: Dictionary of instantiated agents keyed by role/id.
    """
    templates = config["prompt_templates"]
    roles_config = config["agent_architecture_config"]["roles"]

    # Wrap the custom client for CrewAI compatibility
    llm_adapter = LangChainAdapter(llm_wrapper)

    # 1. Analyst Agent
    analyst = Agent(
        role=roles_config["analyst_role_name"],
        goal="Interpret Fed Funds Futures to extract market-implied probabilities",
        backstory=templates["analyst_system_prompt"], # Base system prompt as backstory
        llm=llm_adapter,
        verbose=True,
        allow_delegation=False
    )

    # 2. Economist Agent
    # Extract backstory from the preamble of the task prompt or define a generic one
    # Since the prompt template is task-specific, we use a generic backstory derived from the role description
    economist = Agent(
        role=roles_config["economist_role_name"],
        goal="Formulate three distinct policy options (dovish/neutral/hawkish)",
        backstory="You are the Chief Economist for the FOMC. You provide objective, data-driven policy options.",
        llm=llm_adapter,
        verbose=True,
        allow_delegation=False
    )

    # 3. Member Agents
    members = {}
    for member_info in member_registry:
        agent_id = member_info["agent_id"]
        # The system_prompt from Task 10 serves as the backstory/persona
        members[agent_id] = Agent(
            role=roles_config["member_role_name"],
            goal="Analyze data, deliberate, and vote on policy options",
            backstory=member_info["system_prompt"],
            llm=llm_adapter,
            verbose=True,
            allow_delegation=False
        )

    # 4. Statement Drafter
    drafter = Agent(
        role=roles_config["statement_drafter_role_name"],
        goal="Draft simulated FOMC statement",
        backstory="You are the designated drafter for the FOMC statement, responsible for synthesizing the committee's decision into official communication.",
        llm=llm_adapter,
        verbose=True,
        allow_delegation=False
    )

    return {
        "analyst": analyst,
        "economist": economist,
        "members": members, # Dict of 3 agents
        "drafter": drafter
    }


# ------------------------------------------------------------------------------
# Task 12, Step 2: Define the task sequence using CrewAI Task class
# ------------------------------------------------------------------------------
def create_fedsight_tasks(
    agents: Dict[str, Any],
    config: Dict[str, Any]
) -> Dict[str, Task]:
    """
    Defines the sequential tasks for the FedSight workflow.

    DAG:
    Analyst -> Economist -> Members (x3) -> Drafter

    Args:
        agents (Dict): Instantiated agents.
        config (Dict): Study configuration.

    Returns:
        Dict[str, Task]: Dictionary of instantiated tasks.
    """
    templates = config["prompt_templates"]

    # 1. Analyst Task
    # Note: The description contains placeholders ({futures_data}) to be filled at runtime via inputs
    analyst_task = Task(
        description=templates["analyst_system_prompt"],
        agent=agents["analyst"],
        expected_output="Probability distribution and market sentiment summary"
    )

    # 2. Economist Task
    # Depends on Analyst output
    economist_task = Task(
        description=templates["economist_task_prompt"],
        agent=agents["economist"],
        context=[analyst_task],
        expected_output="Three policy options with rationales"
    )

    # 3. Member Tasks
    # Depend on Economist output
    member_tasks = {}
    member_agents = agents["members"]

    # We assume member_agents is a dict like {'member_0': Agent, ...}
    # Sort keys to ensure deterministic task order in the list
    sorted_member_ids = sorted(member_agents.keys())

    for m_id in sorted_member_ids:
        agent = member_agents[m_id]
        task = Task(
            description=templates["member_cod_deliberation"],
            agent=agent,
            context=[economist_task],
            expected_output="Vote (Option A/B/C) with rationale"
        )
        member_tasks[m_id] = task

    # 4. Statement Task
    # Depends on all Member outputs
    # Collect all member tasks for context
    all_member_tasks = [member_tasks[m_id] for m_id in sorted_member_ids]

    statement_task = Task(
        description=templates["chair_statement_generation"],
        agent=agents["drafter"],
        context=all_member_tasks,
        expected_output="Simulated FOMC statement"
    )

    return {
        "analyst_task": analyst_task,
        "economist_task": economist_task,
        "member_tasks": member_tasks,
        "statement_task": statement_task
    }


# ------------------------------------------------------------------------------
# Task 12, Step 3: Assemble the CrewAI Crew
# ------------------------------------------------------------------------------
def assemble_fedsight_crew(
    agents: Dict[str, Any],
    tasks: Dict[str, Any],
    memory_enabled: bool = False
) -> Crew:
    """
    Assembles the Agents and Tasks into a runnable Crew.

    Args:
        agents (Dict): Dictionary of agents.
        tasks (Dict): Dictionary of tasks.
        memory_enabled (bool): Whether to enable CrewAI memory (default False for baseline).

    Returns:
        Crew: The configured Crew object.
    """
    # Flatten agents list
    agent_list = [
        agents["analyst"],
        agents["economist"]
    ]
    # Add members in deterministic order
    sorted_member_ids = sorted(agents["members"].keys())
    for m_id in sorted_member_ids:
        agent_list.append(agents["members"][m_id])
    agent_list.append(agents["drafter"])

    # Flatten tasks list in execution order
    task_list = [
        tasks["analyst_task"],
        tasks["economist_task"]
    ]
    for m_id in sorted_member_ids:
        task_list.append(tasks["member_tasks"][m_id])
    task_list.append(tasks["statement_task"])

    # Create Crew
    crew = Crew(
        agents=agent_list,
        tasks=task_list,
        process=Process.sequential,
        verbose=2,
        memory=memory_enabled
    )

    return crew


# ------------------------------------------------------------------------------
# Task 12, Orchestrator Function
# ------------------------------------------------------------------------------
def define_fedsight_architecture(
    config: Dict[str, Any],
    member_registry: List[Dict[str, Any]],
    llm_wrapper: Any,
    memory_enabled: bool = False
) -> Dict[str, Any]:
    """
    Orchestrates the definition of the FedSight multi-agent architecture.

    Args:
        config (Dict): Study configuration.
        member_registry (List[Dict]): Member personas.
        llm_wrapper (Any): LLM client.
        memory_enabled (bool): Toggle for ICL variant.

    Returns:
        Dict[str, Any]:
            - crew: The assembled Crew object.
            - agents: The dictionary of agent objects.
            - tasks: The dictionary of task objects.
    """
    # Step 1: Create Agents
    agents = create_fedsight_agents(config, member_registry, llm_wrapper)

    # Step 2: Create Tasks
    tasks = create_fedsight_tasks(agents, config)

    # Step 3: Assemble Crew
    crew = assemble_fedsight_crew(agents, tasks, memory_enabled)

    return {
        "crew": crew,
        "agents": agents,
        "tasks": tasks
    }
````

````python
# Task 13 — Implement the FedSight AI baseline meeting simulation callable

# ==============================================================================
# Task 13: Implement the FedSight AI baseline meeting simulation callable
# ==============================================================================

# -------------------------------------------------------------------------------------------
# Task 13, Step 1: Define the single-meeting-single-run simulation interface and RunArtifact
# -------------------------------------------------------------------------------------------
@dataclass(frozen=True)
class RunArtifact:
    """
    Immutable container for all outputs generated during a single simulation run.

    This dataclass serves as the canonical record for a meeting-run, capturing
    intermediate agent outputs, the final decision, and computational cost metadata.
    It is designed to be serialized for persistence and deserialized for metric computation.

    Attributes:
        variant (str): The experimental variant (e.g., "baseline", "icl", "cod").
        meeting_id (pd.Timestamp): The canonical identifier for the meeting.
        run_id (int): The iteration index of this simulation (1-based).
        analyst_output (str): The raw text output from the Analyst agent.
        economist_options (str): The raw text output from the Economist agent.
        member_votes (List[Dict[str, Any]]): A list of 3 dictionaries, each containing:
            - 'agent_id': str
            - 'vote': str (Option A/B/C)
            - 'rationale': str
        aggregated_decision (str): The final committee decision (Option A/B/C).
        predicted_change_bps (Optional[int]): The mapped basis point change (populated in Task 14).
        generated_statement (str): The simulated FOMC statement text.
        token_usage (Dict[str, int]): Aggregated token usage for the run (prompt, completion, total).
        provenance (Dict[str, Any]): Metadata including timestamps, model versions, and config hash.
    """
    variant: str
    meeting_id: pd.Timestamp
    run_id: int
    analyst_output: str
    economist_options: str
    member_votes: List[Dict[str, Any]]
    aggregated_decision: str
    generated_statement: str
    token_usage: Dict[str, int]
    provenance: Dict[str, Any]
    predicted_change_bps: Optional[int] = None

    def __post_init__(self):
        """
        Validates the integrity of the artifact upon initialization.

        Raises:
            ValueError: If member_votes does not contain exactly 3 entries or if
                        token_usage is missing required keys.
        """
        if len(self.member_votes) != 3:
            raise ValueError(f"RunArtifact Error: Expected 3 member votes, got {len(self.member_votes)}.")

        required_token_keys = {"total_tokens", "prompt_tokens", "completion_tokens"}
        if not required_token_keys.issubset(self.token_usage.keys()):
            raise ValueError(f"RunArtifact Error: token_usage missing keys. Required: {required_token_keys}")

# ------------------------------------------------------------------------------
# Task 13, Step 2: Execute the CrewAI workflow for one meeting-run
# ------------------------------------------------------------------------------
def parse_member_vote(agent_output: str) -> Dict[str, str]:
    """
    Parses the raw text output from a Member agent to extract the vote and rationale.

    Expected format from prompt instructions:
    - Vote: [Option A / Option B / Option C]
    - Rationale: [Text]

    Args:
        agent_output (str): Raw LLM response text.

    Returns:
        Dict[str, str]: A dictionary with keys 'vote' and 'rationale'.
                        Returns 'Parse Error' if extraction fails.
    """
    text = agent_output.strip()

    # Regex for Vote: Matches "Vote: Option A" (case insensitive)
    # We allow for bolding (**Option A**) or slight spacing variations.
    vote_match = re.search(r"Vote:\s*(?:\*\*)?(Option [ABC])(?:\*\*)?", text, re.IGNORECASE)
    vote = vote_match.group(1).title() if vote_match else "Parse Error"

    # Regex for Rationale: Captures everything after "Rationale:"
    rationale_match = re.search(r"Rationale:\s*(.*)", text, re.IGNORECASE | re.DOTALL)
    rationale = rationale_match.group(1).strip() if rationale_match else "No rationale provided."

    return {"vote": vote, "rationale": rationale}

def _run_async_call(wrapper: Any, prompt: str, system_prompt: str, metadata: Dict[str, Any]) -> Dict[str, Any]:
    """
    Helper to execute the async LLM wrapper synchronously within the simulation loop.

    Args:
        wrapper: The LLMClientWrapper instance.
        prompt: User prompt.
        system_prompt: System instruction.
        metadata: Logging metadata.

    Returns:
        Dict: The full response object from the wrapper (content + usage).
    """
    try:
        loop = asyncio.get_event_loop()
    except RuntimeError:
        loop = asyncio.new_event_loop()
        asyncio.set_event_loop(loop)

    if loop.is_running():
        nest_asyncio.apply()
        return loop.run_until_complete(wrapper.call_completion(prompt, system_prompt, metadata))
    else:
        return loop.run_until_complete(wrapper.call_completion(prompt, system_prompt, metadata))

# -------------------------------------------------------------------------------------------
# Task 13, Step 2: Execute the CrewAI workflow for one meeting-run
# -------------------------------------------------------------------------------------------

def _format_macro_value(value: Any, precision: Optional[int], unit: str) -> str:
    """
    Formats a single macro value with strict adherence to the study's presentation standards.

    This helper replicates the formatting logic from Task 8 to ensure consistency
    between the pre-generated 'structured_macro_text' and the granular fields
    required by the Economist agent.

    Args:
        value (Any): The raw value from the dataframe/packet.
        precision (Optional[int]): Number of decimal places (None for strings).
        unit (str): Suffix to append (e.g., "%", "bps").

    Returns:
        str: Formatted string (e.g., "3.00%").
    """
    if value is None:
        return "N/A"

    if precision is not None:
        try:
            val_float = float(value)
            formatted_num = f"{val_float:.{precision}f}"
        except (ValueError, TypeError):
            return str(value)
    else:
        formatted_num = str(value)

    return f"{formatted_num}{unit}"

def _construct_economist_macro_inputs(meeting_packet: Dict[str, Any]) -> Dict[str, str]:
    """
    Constructs the granular macro data blocks required by the Economist prompt template.

    This function ensures that the data seen by the Economist is identical in format
    and precision to the data seen by Members (via structured_macro_text), preserving
    informational symmetry across the agent hierarchy.

    Args:
        meeting_packet (Dict): The agent input packet.

    Returns:
        Dict[str, str]: Dictionary with keys 'inflation_data', 'labor_data',
                        'growth_data', 'financial_data'.
    """
    # Inflation Block
    pce = _format_macro_value(meeting_packet.get("pce_index_yoy"), 2, "%")
    cpi = _format_macro_value(meeting_packet.get("cpi_index_yoy"), 2, "%")
    exp = _format_macro_value(meeting_packet.get("inflation_expectations_1yr"), 2, "%")
    inflation_data = (
        f"PCE Index (YoY): {pce}\n"
        f"CPI Index (YoY): {cpi}\n"
        f"Inflation Expectations (1yr): {exp}"
    )

    # Labor Block
    u3 = _format_macro_value(meeting_packet.get("unemployment_rate_u3"), 2, "%")
    labor_data = f"Unemployment Rate (U-3): {u3}"

    # Growth Block
    gdp = _format_macro_value(meeting_packet.get("bbk_real_gdp_nowcast"), 2, "%")
    growth_data = f"Real GDP Nowcast: {gdp}"

    # Financial Block
    y3m = _format_macro_value(meeting_packet.get("treasury_yield_3m"), 2, "%")
    y6m = _format_macro_value(meeting_packet.get("treasury_yield_6m"), 2, "%")
    vix = _format_macro_value(meeting_packet.get("vix_index"), 2, "")
    m2 = _format_macro_value(meeting_packet.get("m2_supply_level_sa"), 1, "")
    financial_data = (
        f"3-Month Treasury Yield: {y3m}\n"
        f"6-Month Treasury Yield: {y6m}\n"
        f"VIX Index: {vix}\n"
        f"M2 Supply Level: {m2}"
    )

    return {
        "inflation_data": inflation_data,
        "labor_data": labor_data,
        "growth_data": growth_data,
        "financial_data": financial_data
    }


def execute_meeting_simulation(
    meeting_id: pd.Timestamp,
    run_id: int,
    variant: str,
    meeting_packet: Dict[str, Any],
    member_agents_registry: List[Dict[str, Any]],
    config: Dict[str, Any],
    llm_wrapper: Any
) -> 'RunArtifact':
    """
    Executes a full multi-agent simulation for a single meeting and run.

    This function orchestrates the sequential execution of the FedSight agents:
    1. Analyst: Interprets Fed Funds Futures.
    2. Economist: Formulates policy options based on Analyst output and Macro Data.
    3. Members (x3): Deliberate and vote based on Economist options and Context.
    4. Drafter: Synthesizes the voting record into a statement.

    It captures all intermediate outputs and aggregates token usage into a RunArtifact.

    NOTE: This function does NOT perform vote aggregation. The 'aggregated_decision'
    field in the returned artifact is set to "PENDING_AGGREGATION". The rigorous
    aggregation logic is applied by the Orchestrator in the subsequent step.

    Args:
        meeting_id (pd.Timestamp): Canonical meeting date.
        run_id (int): Run index (1-5).
        variant (str): Experiment variant ("baseline", "icl", "cod").
        meeting_packet (Dict): Input data (macro, unstructured, futures).
        member_agents_registry (List[Dict]): The 3 specific member personas for this run.
        config (Dict): Full study configuration.
        llm_wrapper (Any): Configured LLM client wrapper.

    Returns:
        RunArtifact: The immutable record of the simulation.
    """
    # Initialize Token Accumulator
    token_usage = {"prompt_tokens": 0, "completion_tokens": 0, "total_tokens": 0}

    def accumulate_usage(response: Dict[str, Any]):
        # Updates the running token count from a wrapper response
        if "usage" in response:
            u = response["usage"]
            token_usage["prompt_tokens"] += u.get("prompt_tokens", 0)
            token_usage["completion_tokens"] += u.get("completion_tokens", 0)
            token_usage["total_tokens"] += u.get("total_tokens", 0)

    # Retrieve Templates
    templates = config["prompt_templates"]

    # Base Metadata for Logging
    base_metadata = {
        "variant": variant,
        "meeting_id": str(meeting_id.date()),
        "run_id": run_id
    }

    # 1. Analyst Agent
    analyst_prompt = templates["analyst_system_prompt"].format(
        futures_data=meeting_packet.get("fed_funds_futures_raw", "N/A")
    )

    analyst_meta = base_metadata.copy()
    analyst_meta["role"] = "Analyst"

    analyst_res = _run_async_call(
        wrapper=llm_wrapper,
        prompt=analyst_prompt,
        system_prompt="You are a Quantitative Financial Analyst.",
        metadata=analyst_meta
    )
    accumulate_usage(analyst_res)
    analyst_output = analyst_res["content"]

    # 2. Economist Agent
    # Construct granular macro inputs using rigorous formatting
    macro_inputs = _construct_economist_macro_inputs(meeting_packet)

    economist_prompt = templates["economist_task_prompt"].format(
        inflation_data=macro_inputs["inflation_data"],
        labor_data=macro_inputs["labor_data"],
        growth_data=macro_inputs["growth_data"],
        financial_data=macro_inputs["financial_data"],
        unstructured_context=meeting_packet.get("unstructured_context", ""),
        analyst_market_summary=analyst_output
    )

    economist_meta = base_metadata.copy()
    economist_meta["role"] = "Economist"

    economist_res = _run_async_call(
        wrapper=llm_wrapper,
        prompt=economist_prompt,
        system_prompt="You are the Chief Economist for the FOMC.",
        metadata=economist_meta
    )
    accumulate_usage(economist_res)
    economist_options = economist_res["content"]

    # 3. Member Agents (x3)
    member_votes = []
    # Sort by agent_id to ensure deterministic execution order
    sorted_members = sorted(member_agents_registry, key=lambda x: x["agent_id"])

    for member_info in sorted_members:
        agent_id = member_info["agent_id"]
        persona_prompt = member_info["system_prompt"]

        # Format User Prompt
        # Note: Members use the pre-constructed 'structured_macro_text' from Task 8
        # which matches the formatting used for the Economist above.
        member_user_prompt = templates["member_cod_deliberation"].format(
            economist_options=economist_options,
            beige_book_text=meeting_packet.get("beige_book_headered", ""),
            dot_plot_text=meeting_packet.get("dot_plot_headered", ""),
            fedwatch_text=meeting_packet.get("fedwatch_headered", ""),
            market_probs="See FedWatch text above",
            structured_macro_data=meeting_packet.get("structured_macro_text", "")
        )

        member_meta = base_metadata.copy()
        member_meta["role"] = "Member"
        member_meta["agent_id"] = agent_id

        member_res = _run_async_call(
            wrapper=llm_wrapper,
            prompt=member_user_prompt,
            system_prompt=persona_prompt,
            metadata=member_meta
        )
        accumulate_usage(member_res)

        # Parse Vote
        parsed = parse_member_vote(member_res["content"])
        member_votes.append({
            "agent_id": agent_id,
            "vote": parsed["vote"],
            "rationale": parsed["rationale"],
            "raw_output": member_res["content"]
        })


    # 4. Statement Drafter
    voting_record_str = "\n".join([
        f"- {v['agent_id']}: {v['vote']}\n  Rationale: {v['rationale'][:200]}..."
        for v in member_votes
    ])

    drafter_prompt = templates["chair_statement_generation"].format(
        voting_record=voting_record_str,
        majority_reasoning_summary="Please synthesize the majority view from the voting record above."
    )

    drafter_meta = base_metadata.copy()
    drafter_meta["role"] = "StatementDrafter"

    drafter_res = _run_async_call(
        wrapper=llm_wrapper,
        prompt=drafter_prompt,
        system_prompt="You are the designated drafter for the FOMC statement.",
        metadata=drafter_meta
    )
    accumulate_usage(drafter_res)
    generated_statement = drafter_res["content"]

    # Artifact Construction
    artifact = RunArtifact(
        variant=variant,
        meeting_id=meeting_id,
        run_id=run_id,
        analyst_output=analyst_output,
        economist_options=economist_options,
        member_votes=member_votes,
        aggregated_decision="PENDING_AGGREGATION", # Explicit placeholder
        generated_statement=generated_statement,
        token_usage=token_usage,
        provenance={
            "timestamp": datetime.utcnow().isoformat(),
            "config_hash": config.get("config_hash", "unknown")
        }
    )

    return artifact


# -------------------------------------------------------------------------------------------
# Task 13, Step 3: Aggregate member votes into a single run-level decision
# -------------------------------------------------------------------------------------------
def aggregate_member_votes(
    member_votes: List[Dict[str, str]],
    tie_break_rule: str = "choose_neutral_option"
) -> Tuple[str, Dict[str, int]]:
    """
    Aggregates individual member votes into a single committee decision using
    majority rule and a deterministic tie-breaking policy.

    Algorithm:
    1. Extract vote strings from member records.
    2. Filter for valid options {Option A, Option B, Option C}.
    3. Count votes.
    4. Determine winner based on majority.
    5. Apply tie-break rule if necessary.

    Tie-Break Policy ('choose_neutral_option'):
    - If there is a tie for the winning position:
        - If Option B (Neutral) is among the tied options, select Option B.
        - If Option B is not among the tied options (e.g., tie between A and C),
          select Option B as the conservative default (status quo bias).

    Args:
        member_votes (List[Dict]): List of dicts containing 'vote' keys.
        tie_break_rule (str): Policy identifier (default: "choose_neutral_option").

    Returns:
        Tuple[str, Dict[str, int]]:
            - aggregated_decision: The winning option string (e.g., "Option B").
            - vote_counts: Dictionary of vote counts for auditability.
    """
    # 1. Extract and Normalize
    valid_options = {"Option A", "Option B", "Option C"}
    raw_votes = [v.get("vote", "").strip() for v in member_votes]

    # Filter for valid votes only
    # We normalize case just in case, though regex in Step 2 handled titles
    clean_votes = []
    for v in raw_votes:
        # Ensure exact match with valid set (case-sensitive title case expected)
        # If regex produced "Option a", we fix it here
        v_title = v.title()
        # Handle "Option A" vs "Option a"
        # We assume standard format "Option X"
        if v_title in valid_options:
            clean_votes.append(v_title)
        else:
            # Log invalid vote implicitly by exclusion (or map to Abstain)
            pass

    # 2. Count
    if not clean_votes:
        return "Error", {"Error": len(raw_votes)}

    counts = Counter(clean_votes)

    # 3. Determine Winner
    # most_common returns elements ordered by count desc.
    # If counts are equal, insertion order is preserved.
    # We need deterministic handling, so we don't rely on insertion order.
    max_count = max(counts.values())
    candidates = [opt for opt, count in counts.items() if count == max_count]

    if len(candidates) == 1:
        decision = candidates[0]
    else:
        # Tie-Break Logic
        if tie_break_rule == "choose_neutral_option":
            # If B is in the tie, pick B
            if "Option B" in candidates:
                decision = "Option B"
            else:
                # Tie between A and C (e.g., 1 vote A, 1 vote C, 1 Invalid/Abstain)
                # Conservative default
                decision = "Option B"
        else:
            # Fallback for unknown rules: Sort alphabetically (A < B < C) -> A wins
            # This ensures determinism even if rule is unknown
            decision = sorted(candidates)[0]

    return decision, dict(counts)


# ==============================================================================
# Task 13, Orchestrator Function
# ==============================================================================

def run_single_simulation(
    meeting_id: pd.Timestamp,
    run_id: int,
    variant: str,
    meeting_packet: Dict[str, Any],
    member_agents_registry: List[Dict[str, Any]],
    config: Dict[str, Any],
    llm_wrapper: Any
) -> 'RunArtifact':
    """
    Orchestrates the complete simulation of a single FOMC meeting run.

    This function integrates the execution of the multi-agent workflow (Step 2)
    with the rigorous vote aggregation logic (Step 3) to produce a finalized,
    immutable RunArtifact.

    Process:
    1. Execute the agent workflow to generate intermediate outputs and votes.
    2. Apply the deterministic vote aggregation and tie-breaking policy.
    3. Finalize the RunArtifact with the computed decision.

    Args:
        meeting_id (pd.Timestamp): Canonical meeting date.
        run_id (int): Run index (1-5).
        variant (str): Experiment variant ("baseline", "icl", "cod").
        meeting_packet (Dict): Input data (macro, unstructured, futures).
        member_agents_registry (List[Dict]): The 3 specific member personas.
        config (Dict): Full study configuration.
        llm_wrapper (Any): Configured LLM client wrapper.

    Returns:
        RunArtifact: The complete, finalized simulation record.
    """
    # Step 1: Execute Workflow (Task 13, Step 2)
    # This generates the raw agent outputs and member votes.
    raw_artifact = execute_meeting_simulation(
        meeting_id=meeting_id,
        run_id=run_id,
        variant=variant,
        meeting_packet=meeting_packet,
        member_agents_registry=member_agents_registry,
        config=config,
        llm_wrapper=llm_wrapper
    )

    # Step 2: Aggregate Votes (Task 13, Step 3)
    # We apply the rigorous aggregation logic here to ensure the final decision
    # adheres strictly to the configured tie-break rules.
    tie_break_rule = config["simulation_config"].get("tie_break_rule", "choose_neutral_option")

    final_decision, vote_counts = aggregate_member_votes(
        member_votes=raw_artifact.member_votes,
        tie_break_rule=tie_break_rule
    )

    # Step 3: Finalize Artifact
    # We update the artifact with the rigorously computed decision.
    # Update provenance with vote counts for easier debugging
    updated_provenance = raw_artifact.provenance.copy()
    updated_provenance["vote_counts"] = vote_counts

    final_artifact = replace(
        raw_artifact,
        aggregated_decision=final_decision,
        provenance=updated_provenance
    )

    return final_artifact
````

````python
# Task 14 — Implement the vote-to-basis-point mapping logic

# ==============================================================================
# Task 14: Implement the vote-to-basis-point mapping logic
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 14, Step 1: Define and enforce a deterministic mapping from Economist options
# ------------------------------------------------------------------------------
def parse_economist_options(economist_output: str) -> Dict[str, int]:
    """
    Parses the Economist's output to extract the basis point impact of each option.

    Expected format in Economist output:
    "Option A ... Action: -25 bps"
    "Option B ... Action: 0 bps"
    "Option C ... Action: +25 bps"

    Robustness Strategy:
    1. Split text into Option blocks.
    2. Regex search for "Action:" followed by keywords (cut/hike/hold) or explicit numbers.
    3. Normalize to integer basis points.

    Args:
        economist_output (str): Raw text from the Economist agent.

    Returns:
        Dict[str, int]: Mapping {'Option A': int, 'Option B': int, 'Option C': int}.

    Raises:
        ValueError: If parsing fails for any option (strict protocol).
    """
    mapping = {}

    # Normalize text
    text = economist_output

    # Define regex for extracting action lines
    # Matches: "Action:", optional whitespace, optional sign, digits, optional whitespace, "bps"
    # Also handles "Action: Cut 25 bps"
    # Group 1: Full action text for context
    # Group 2: Keyword (Cut/Hike/Hold/Maintain/Pause)
    # Group 3: Number
    action_pattern = re.compile(
        r"Action:.*?(?P<keyword>cut|hike|hold|maintain|pause|increase|decrease)?.*?(?P<sign>[+-])?(?P<num>\d+)?",
        re.IGNORECASE
    )

    # We need to associate actions with Options.
    # Strategy: Find indices of "Option A", "Option B", "Option C" and search in between.
    option_indices = []
    for opt in ["Option A", "Option B", "Option C"]:
        match = re.search(rf"\b{opt}\b", text, re.IGNORECASE)
        if match:
            option_indices.append((opt, match.start()))

    # Sort by position
    option_indices.sort(key=lambda x: x[1])

    # Append end of string sentinel
    option_indices.append(("End", len(text)))

    for i in range(len(option_indices) - 1):
        opt_name, start_idx = option_indices[i]
        end_idx = option_indices[i+1][1]
        segment = text[start_idx:end_idx]

        # Search for action in segment
        match = action_pattern.search(segment)

        bps = None
        if match:
            keyword = match.group("keyword")
            sign = match.group("sign")
            num_str = match.group("num")

            # Logic to determine bps
            if keyword:
                kw = keyword.lower()
                if kw in ["hold", "maintain", "pause"]:
                    bps = 0
                elif num_str:
                    val = int(num_str)
                    if kw in ["cut", "decrease"]:
                        bps = -val
                    elif kw in ["hike", "increase"]:
                        bps = val
            elif num_str:
                # No keyword, rely on sign or raw number
                val = int(num_str)
                if sign == "-":
                    bps = -val
                else:
                    bps = val # Assume positive if no sign and no keyword, or check context?
                    # Strictness: If just "25 bps", usually implies hike or magnitude.
                    # But without keyword, it's ambiguous.
                    # However, usually Option A=Cut, C=Hike.
                    # We will rely on sign if present. If no sign/keyword, we might fail.
                    # Let's be permissive if explicit sign is missing but number exists:
                    # If Option A (Dovish) -> assume negative? No, that's context assumption.
                    # We require explicit sign or keyword.
                    pass

            # Special case: "Action: 0 bps"
            if num_str and int(num_str) == 0:
                bps = 0

        if bps is None:
            # Fallback: Look for simple "0 bps" or "maintain" without "Action:" prefix if strict parse failed?
            # No, strict protocol requires "Action:" line as per prompt instructions to Economist.
            # We raise error to enforce prompt compliance.
            raise ValueError(f"Could not parse basis point action for {opt_name} in segment: {segment[:50]}...")

        mapping[opt_name] = bps

    # Verify all 3 present
    if len(mapping) != 3:
        missing = set(["Option A", "Option B", "Option C"]) - set(mapping.keys())
        raise ValueError(f"Failed to parse all 3 options. Missing: {missing}")

    return mapping


# ------------------------------------------------------------------------------
# Task 14, Step 2: Apply the option-to-bps mapping to the aggregated vote
# ------------------------------------------------------------------------------
def map_decision_to_bps(
    aggregated_decision: str,
    option_map: Dict[str, int]
) -> int:
    """
    Maps the text decision (Option A/B/C) to a basis point integer.

    Args:
        aggregated_decision (str): "Option A", "Option B", or "Option C".
        option_map (Dict): Output from Step 1.

    Returns:
        int: The predicted change in basis points.

    Raises:
        ValueError: If decision is not in the map.
    """
    # Normalize key
    key = aggregated_decision.strip().title()

    # Handle "Option A" vs "Option a" mismatch if any
    # The map keys are "Option A", "Option B", "Option C" from Step 1

    if key not in option_map:
        raise ValueError(f"Aggregated decision '{key}' not found in option map: {option_map.keys()}")

    return option_map[key]


# ------------------------------------------------------------------------------
# Task 14, Orchestrator Function
# ------------------------------------------------------------------------------
def apply_vote_mapping(run_artifact: 'RunArtifact') -> 'RunArtifact':
    """
    Orchestrates the mapping of the qualitative decision to quantitative basis points.

    Process:
    1. Parse Economist options to build the map.
    2. Map the aggregated decision to bps.
    3. Update the RunArtifact.

    Args:
        run_artifact (RunArtifact): The artifact with raw text outputs.

    Returns:
        RunArtifact: Updated artifact with 'predicted_change_bps' populated.
    """
    # Step 1: Parse
    try:
        option_map = parse_economist_options(run_artifact.economist_options)
    except ValueError as e:
        # Log error in provenance and return artifact without bps (incomplete run)
        # Or fail hard. The prompt implies "conclusive" implementation.
        # We will fail hard to ensure data quality.
        raise ValueError(f"Mapping failed for Run {run_artifact.run_id}: {e}")

    # Step 2: Map
    try:
        bps = map_decision_to_bps(run_artifact.aggregated_decision, option_map)
    except ValueError as e:
        raise ValueError(f"Mapping failed for Run {run_artifact.run_id}: {e}")

    # Step 3: Update
    # We also store the map in provenance for audit
    new_provenance = run_artifact.provenance.copy()
    new_provenance["option_bps_map"] = option_map

    updated_artifact = replace(
        run_artifact,
        predicted_change_bps=bps,
        provenance=new_provenance
    )

    return updated_artifact
````

````python
# Task 15 — Implement FedSight ICL (historical rehearsal and memory persistence)

# ==============================================================================
# Task 15: Implement FedSight ICL (historical rehearsal and memory persistence)
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 15, Step 1: Execute historical meeting simulations
# ------------------------------------------------------------------------------
def execute_historical_simulations(
    training_meetings: List[str],
    historical_packets: Dict[str, Dict[str, Any]],
    member_agents_registry: List[Dict[str, Any]],
    config: Dict[str, Any],
    llm_wrapper: Any
) -> List[Dict[str, Any]]:
    """
    Executes historical simulations for In-Context Learning (ICL).

    Agents are placed in past meetings (using vintage data) and asked to vote
    and forecast, without knowledge of the actual outcome.

    Args:
        training_meetings (List[str]): List of meeting IDs (dates) to simulate.
        historical_packets (Dict): Data packets for those meetings.
        member_agents_registry (List[Dict]): The member personas.
        config (Dict): Study configuration.
        llm_wrapper (Any): LLM client.

    Returns:
        List[Dict]: Simulation results containing vote, rationale, and forecast.
    """
    results = []
    template = config["prompt_templates"]["icl_historical_simulation_prompt"]

    # Import parse_member_vote from Task 13 context (assuming available)
    # If not, we redefine a minimal version here for robustness
    def _parse_vote_local(text: str) -> Dict[str, str]:
        vote_match = re.search(r"Vote:\s*(?:\*\*)?(Option [ABC])(?:\*\*)?", text, re.IGNORECASE)
        vote = vote_match.group(1).title() if vote_match else "Parse Error"
        rationale_match = re.search(r"Rationale:\s*(.*)", text, re.IGNORECASE | re.DOTALL)
        rationale = rationale_match.group(1).strip() if rationale_match else "No rationale provided."
        return {"vote": vote, "rationale": rationale}

    for meeting_id in training_meetings:
        if meeting_id not in historical_packets:
            raise ValueError(f"Missing historical packet for training meeting {meeting_id}")

        packet = historical_packets[meeting_id]
        macro_text = packet.get("structured_macro_text", "Macro data unavailable.")

        for member in member_agents_registry:
            agent_id = member["agent_id"]

            # Format Prompt
            prompt = template.format(
                historical_meeting_id=meeting_id,
                structured_macro_data=macro_text,
                beige_book_text=packet.get("beige_book_headered", ""),
                dot_plot_text=packet.get("dot_plot_headered", ""),
                fedwatch_text=packet.get("fedwatch_headered", "")
            )

            # Execute
            response = _run_async_call_icl(
                wrapper=llm_wrapper,
                prompt=prompt,
                system_prompt=member["system_prompt"],
                metadata={"variant": "icl_training", "meeting_id": meeting_id, "agent_id": agent_id}
            )

            content = response["content"]

            # Parse
            parsed_vote = _parse_vote_local(content)

            # Parse Forecast (Regex)
            forecast_match = re.search(r"Year-End Forecast:\s*(.*)", content, re.IGNORECASE)
            forecast = forecast_match.group(1).strip() if forecast_match else "N/A"

            results.append({
                "agent_id": agent_id,
                "meeting_id": meeting_id,
                "vote": parsed_vote["vote"],
                "rationale": parsed_vote["rationale"],
                "forecast": forecast,
                "raw_output": content
            })

    return results


# ------------------------------------------------------------------------------
# Task 15, Step 2: Reveal ground truth and execute reflection
# ------------------------------------------------------------------------------
def execute_reflection_phase(
    simulation_results: List[Dict[str, Any]],
    historical_outcomes: Dict[str, str],
    config: Dict[str, Any],
    llm_wrapper: Any
) -> List[Dict[str, Any]]:
    """
    Conducts the reflection phase where agents compare their vote to history.

    Args:
        simulation_results (List[Dict]): Outputs from Step 1.
        historical_outcomes (Dict): Map of meeting_id -> actual decision text.
        config (Dict): Config.
        llm_wrapper (Any): LLM client.

    Returns:
        List[Dict]: Records containing the 'lesson_text'.
    """
    reflections = []
    template = config["prompt_templates"]["icl_reflection"]

    for res in simulation_results:
        meeting_id = res["meeting_id"]
        true_outcome = historical_outcomes.get(meeting_id, "Unknown")

        prompt = template.format(
            agent_vote=res["vote"],
            true_outcome=true_outcome
        )

        response = _run_async_call_icl(
            wrapper=llm_wrapper,
            prompt=prompt,
            system_prompt="You are reflecting on your past performance.",
            metadata={"variant": "icl_reflection", "meeting_id": meeting_id, "agent_id": res["agent_id"]}
        )

        content = response["content"]

        # Parse Lesson
        lesson_match = re.search(r"Lesson Learned:\s*(.*)", content, re.IGNORECASE | re.DOTALL)
        lesson_text = lesson_match.group(1).strip() if lesson_match else content.strip()

        reflections.append({
            "agent_id": res["agent_id"],
            "meeting_id": meeting_id,
            "lesson_text": lesson_text,
            "reflection_raw": content
        })

    return reflections


# ------------------------------------------------------------------------------
# Task 15, Step 3: Persist reflections to long-term memory
# ------------------------------------------------------------------------------
def persist_memory(
    reflections: List[Dict[str, Any]],
    llm_wrapper: Any,
    embedding_model: str
) -> List[Dict[str, Any]]:
    """
    Embeds lessons and stores them in a memory structure.

    Args:
        reflections (List[Dict]): Reflection records.
        llm_wrapper (Any): Client wrapper.
        embedding_model (str): Model ID.

    Returns:
        List[Dict]: Memory store with embeddings.
    """
    memory_store = []

    async def _batch_embed():
        for ref in reflections:
            text = ref["lesson_text"]
            try:
                # Direct client access for embedding
                resp = await llm_wrapper.client.embeddings.create(
                    input=text,
                    model=embedding_model
                )
                vector = resp.data[0].embedding

                memory_store.append({
                    "agent_id": ref["agent_id"],
                    "meeting_id": ref["meeting_id"],
                    "lesson_text": text,
                    "embedding": vector,
                    "timestamp": datetime.utcnow().isoformat()
                })
            except Exception as e:
                print(f"Embedding failed for {ref['agent_id']}: {e}")

    try:
        loop = asyncio.get_event_loop()
    except RuntimeError:
        loop = asyncio.new_event_loop()
        asyncio.set_event_loop(loop)

    if loop.is_running():
        import nest_asyncio
        nest_asyncio.apply()
        loop.run_until_complete(_batch_embed())
    else:
        loop.run_until_complete(_batch_embed())

    return memory_store


# ------------------------------------------------------------------------------
# Task 15, Orchestrator Function
# ------------------------------------------------------------------------------
def run_icl_training(
    config: Dict[str, Any],
    historical_packets: Dict[str, Any],
    historical_outcomes: Dict[str, str],
    member_agents_registry: List[Dict[str, Any]],
    llm_wrapper: Any
) -> List[Dict[str, Any]]:
    """
    Orchestrates the In-Context Learning training phase.

    Args:
        config (Dict): Study config.
        historical_packets (Dict): Data for training meetings.
        historical_outcomes (Dict): Ground truth for training meetings.
        member_agents_registry (List[Dict]): Agents.
        llm_wrapper (Any): LLM client.

    Returns:
        List[Dict]: The populated memory store.
    """
    training_meetings = config["icl_training_config"].get("training_meetings", [])
    if not training_meetings:
        return []

    # Step 1: Simulate
    sim_results = execute_historical_simulations(
        training_meetings, historical_packets, member_agents_registry, config, llm_wrapper
    )

    # Step 2: Reflect
    reflections = execute_reflection_phase(
        sim_results, historical_outcomes, config, llm_wrapper
    )

    # Step 3: Persist
    embedding_model = config["embedding_config"]["embedding_model_name"]
    memory_store = persist_memory(reflections, llm_wrapper, embedding_model)

    return memory_store
````

````python
# Task 16 — Implement ICL memory retrieval for inference-time context injection

# ==============================================================================
# Task 16: Implement ICL memory retrieval for inference-time context injection
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 16, Step 1: Define the retrieval policy and formalize the query text
# ------------------------------------------------------------------------------
def construct_retrieval_query(
    config: Dict[str, Any],
    meeting_packet: Dict[str, Any]
) -> Dict[str, Any]:
    """
    Constructs the query text and parameters for memory retrieval.

    Query Text Construction:
    "MEETING CONTEXT QUERY\n" + Structured Macro + Dot Plot

    Args:
        config (Dict): Study configuration.
        meeting_packet (Dict): Current meeting's agent input packet.

    Returns:
        Dict: {
            'query_text': str,
            'top_k': int,
            'min_similarity_threshold': float
        }
    """
    policy = config["icl_training_config"]["memory_retrieval_policy"]

    # Construct Query
    # Deterministic concatenation of key context features
    parts = [
        "MEETING CONTEXT QUERY",
        meeting_packet.get("structured_macro_text", ""),
        meeting_packet.get("dot_plot_text", "")
    ]
    query_text = "\n\n".join(parts)

    return {
        "query_text": query_text,
        "top_k": policy.get("top_k", 5),
        "min_similarity_threshold": policy.get("min_similarity_threshold", 0.0)
    }


# ------------------------------------------------------------------------------
# Task 16, Step 2: Retrieve top-k relevant lessons per agent
# ------------------------------------------------------------------------------
def retrieve_relevant_lessons(
    agent_id: str,
    retrieval_request: Dict[str, Any],
    memory_store: List[Dict[str, Any]],
    llm_wrapper: Any,
    embedding_model: str
) -> List[Dict[str, Any]]:
    """
    Retrieves the most relevant lessons for a specific agent using cosine similarity.

    Args:
        agent_id (str): The agent identifier (e.g., "member_0").
        retrieval_request (Dict): Output from Step 1.
        memory_store (List[Dict]): The full memory store (from Task 15).
        llm_wrapper (Any): Client wrapper for embedding the query.
        embedding_model (str): Model ID.

    Returns:
        List[Dict]: Top-k sorted lessons with 'similarity' score added.
    """
    query_text = retrieval_request["query_text"]
    top_k = retrieval_request["top_k"]
    threshold = retrieval_request["min_similarity_threshold"]

    # 1. Embed Query
    # We need to run this async call synchronously
    # Reusing the pattern from Task 13/15
    try:
        loop = asyncio.get_event_loop()
    except RuntimeError:
        loop = asyncio.new_event_loop()
        asyncio.set_event_loop(loop)

    async def _get_embedding():
        resp = await llm_wrapper.client.embeddings.create(
            input=query_text,
            model=embedding_model
        )
        return resp.data[0].embedding

    if loop.is_running():
        import nest_asyncio
        nest_asyncio.apply()
        query_vector = loop.run_until_complete(_get_embedding())
    else:
        query_vector = loop.run_until_complete(_get_embedding())

    query_vec = np.array(query_vector)
    norm_q = np.linalg.norm(query_vec)

    if norm_q == 0:
        return [] # Should not happen with valid API

    # 2. Filter and Score
    scored_lessons = []

    for record in memory_store:
        # Filter by Agent ID
        if record["agent_id"] != agent_id:
            continue

        mem_vec = np.array(record["embedding"])
        norm_m = np.linalg.norm(mem_vec)

        if norm_m == 0:
            continue

        # Cosine Similarity
        sim = np.dot(query_vec, mem_vec) / (norm_q * norm_m)

        if sim >= threshold:
            # Create a copy to attach score
            scored_record = record.copy()
            scored_record["similarity"] = float(sim)
            scored_lessons.append(scored_record)

    # 3. Rank
    # Sort descending by similarity
    scored_lessons.sort(key=lambda x: x["similarity"], reverse=True)

    return scored_lessons[:top_k]


# ------------------------------------------------------------------------------
# Task 16, Step 3: Inject retrieved lessons into each Member's deliberation prompt
# ------------------------------------------------------------------------------
def inject_lessons_into_prompt(
    base_prompt_template: str,
    lessons: List[Dict[str, Any]]
) -> str:
    """
    Augments the member deliberation prompt with retrieved lessons.

    Format:
    ### LESSONS FROM PAST SIMULATIONS:
    - [Lesson 1]
    - [Lesson 2]
    ...

    [Original Prompt]

    Args:
        base_prompt_template (str): The original prompt template string.
        lessons (List[Dict]): Retrieved lessons.

    Returns:
        str: The augmented prompt template.
    """
    if not lessons:
        return base_prompt_template

    header = "### LESSONS FROM PAST SIMULATIONS:\n"
    lesson_lines = [f"- {l['lesson_text']}" for l in lessons]
    block = header + "\n".join(lesson_lines) + "\n\n"

    return block + base_prompt_template


# ------------------------------------------------------------------------------
# Task 16, Orchestrator Function
# ------------------------------------------------------------------------------
def prepare_icl_context(
    agent_id: str,
    config: Dict[str, Any],
    meeting_packet: Dict[str, Any],
    memory_store: List[Dict[str, Any]],
    llm_wrapper: Any
) -> str:
    """
    Orchestrates the retrieval and injection of ICL memory for a specific agent.

    Args:
        agent_id (str): Agent ID.
        config (Dict): Study config.
        meeting_packet (Dict): Current meeting data.
        memory_store (List[Dict]): Memory.
        llm_wrapper (Any): Client.

    Returns:
        str: The augmented prompt template (or original if no lessons).
    """
    # Step 1: Request
    request = construct_retrieval_query(config, meeting_packet)

    # Step 2: Retrieve
    embedding_model = config["embedding_config"]["embedding_model_name"]
    lessons = retrieve_relevant_lessons(
        agent_id, request, memory_store, llm_wrapper, embedding_model
    )

    # Step 3: Inject
    # We retrieve the base template from config to ensure we start clean
    base_template = config["prompt_templates"]["member_cod_deliberation"]
    augmented_template = inject_lessons_into_prompt(base_template, lessons)

    return augmented_template
````

````python
# Task 17 — Implement FedSight CoD (Chain-of-Draft enforcement)

# ==============================================================================
# Task 17: Implement FedSight CoD (Chain-of-Draft enforcement)
# ==============================================================================

def _run_async_call_cod(wrapper: Any, prompt: str, system_prompt: str, metadata: Dict[str, Any]) -> Dict[str, Any]:
    """
    Helper to execute the async LLM wrapper synchronously within the Chain-of-Draft (CoD) loop.

    This function bridges the synchronous execution context of the CoD enforcement logic
    with the asynchronous nature of the `LLMClientWrapper`. It manages the asyncio event loop
    to ensure that repair calls can be made blocking-style, which simplifies the sequential
    validation logic.

    Args:
        wrapper (Any): The `LLMClientWrapper` instance (or compatible interface).
        prompt (str): The user prompt to send to the LLM.
        system_prompt (str): The system instruction for the LLM.
        metadata (Dict[str, Any]): Contextual metadata for logging purposes.

    Returns:
        Dict[str, Any]: The response dictionary from the LLM wrapper, containing content and usage.

    Raises:
        RuntimeError: If there are issues managing the asyncio event loop.
    """
    try:
        loop = asyncio.get_event_loop()
    except RuntimeError:
        loop = asyncio.new_event_loop()
        asyncio.set_event_loop(loop)

    if loop.is_running():
        # If the loop is already running (e.g., inside a Jupyter notebook or another async context),
        # use nest_asyncio to allow re-entrant execution.
        nest_asyncio.apply()
        return loop.run_until_complete(wrapper.call_completion(prompt, system_prompt, metadata))
    else:
        # Standard execution when no loop is running.
        return loop.run_until_complete(wrapper.call_completion(prompt, system_prompt, metadata))


# ------------------------------------------------------------------------------
# Task 17, Step 1: Define CoD constraint parameters and compliance expectations
# ------------------------------------------------------------------------------

@dataclass(frozen=True)
class CoDCompliancePolicy:
    """
    Configuration for Chain-of-Draft (CoD) enforcement.

    This dataclass encapsulates the strict constraints applied to the agent's reasoning process
    under the CoD framework. It defines the structural requirements (number of steps) and
    the conciseness constraint (maximum words per step) that the validation logic enforces.

    Attributes:
        max_words_per_step (int): The hard limit on the number of words allowed per draft step.
                                  Violations trigger a repair loop.
        required_steps (int): The exact number of reasoning steps required (typically 5).
        step_pattern (str): The regex pattern used to identify and parse steps from the raw text.
                            Defaults to capturing "Step N: ...".
    """
    max_words_per_step: int
    required_steps: int
    step_pattern: str = r"Step\s+(\d+):\s*(.*)"


def get_cod_policy(config: Dict[str, Any]) -> CoDCompliancePolicy:
    """
    Factory function to instantiate a CoDCompliancePolicy based on the study configuration.

    This function extracts the relevant parameters from the global configuration dictionary
    and constructs the policy object, ensuring that the enforcement logic uses the
    centralized settings defined in `preprocessing_config`.

    Args:
        config (Dict[str, Any]): The full study configuration dictionary.

    Returns:
        CoDCompliancePolicy: The configured policy object.
    """
    return CoDCompliancePolicy(
        max_words_per_step=config["preprocessing_config"]["cod_max_words_per_step"],
        required_steps=5
    )

# ------------------------------------------------------------------------------
# Task 17, Step 2: Implement word-count validation and deterministic repair
# ------------------------------------------------------------------------------
def validate_and_repair_cod(
    raw_text: str,
    policy: CoDCompliancePolicy,
    llm_wrapper: Any,
    metadata: Dict[str, Any]
) -> Tuple[str, Dict[str, Any]]:
    """
    Validates that the agent's output adheres to Chain-of-Draft constraints.
    Triggers a deterministic repair loop if constraints are violated.

    Args:
        raw_text (str): The agent's initial response.
        policy (CoDCompliancePolicy): Constraints.
        llm_wrapper (Any): Client for repair calls.
        metadata (Dict): Logging context.

    Returns:
        Tuple[str, Dict]: The (possibly repaired) text and a compliance log.
    """
    log = {"violations": [], "repairs": [], "final_status": "success"}

    # Helper to parse steps
    def parse_steps(text):
        # Robust split by "Step N:"
        parts = re.split(r"(Step\s+\d+:)", text, flags=re.IGNORECASE)
        steps = {}
        current_step = None
        for p in parts:
            if re.match(r"Step\s+\d+:", p, re.IGNORECASE):
                current_step = int(re.search(r"\d+", p).group())
            elif current_step is not None:
                # Stop at "Phase 2" or "Rationale"
                clean_content = re.split(r"(Phase 2|Rationale|Final Output)", p, flags=re.IGNORECASE)[0].strip()
                steps[current_step] = clean_content
                current_step = None
        return steps

    steps = parse_steps(raw_text)

    # Check Step Count
    if len(steps) != policy.required_steps:
        log["violations"].append("step_count_mismatch")
        repair_prompt = f"Your output format is incorrect. You must provide exactly {policy.required_steps} steps labeled 'Step 1:' through 'Step {policy.required_steps}:', followed by 'Rationale:'. Rewrite your response."

        # Repair Call
        response = _run_async_call_cod(
            wrapper=llm_wrapper,
            prompt=repair_prompt,
            system_prompt="You are correcting your output format.",
            metadata={**metadata, "action": "cod_format_repair"}
        )
        raw_text = response["content"]
        steps = parse_steps(raw_text)

        if len(steps) != policy.required_steps:
            raise ValueError(f"CoD Format Error: Expected {policy.required_steps} steps, got {len(steps)} after repair.")

    # Check Word Counts
    repaired_steps = {}
    for i in range(1, policy.required_steps + 1):
        content = steps.get(i, "")
        word_count = len(content.split())

        if word_count > policy.max_words_per_step:
            log["violations"].append(f"step_{i}_length_{word_count}")

            # Repair
            repair_prompt = f"Step {i} exceeded {policy.max_words_per_step} words (count: {word_count}). Rewrite it in {policy.max_words_per_step} words or fewer, preserving meaning:\n'{content}'"

            response = _run_async_call_cod(
                wrapper=llm_wrapper,
                prompt=repair_prompt,
                system_prompt="You are condensing your reasoning.",
                metadata={**metadata, "action": f"cod_step_{i}_repair"}
            )
            repaired_content = response["content"]

            # Check again
            if len(repaired_content.split()) > policy.max_words_per_step:
                raise ValueError(f"CoD Length Error: Step {i} still exceeds limit after repair.")

            repaired_steps[i] = repaired_content
            log["repairs"].append(f"step_{i}")
        else:
            repaired_steps[i] = content

    # Reconstruct Text
    final_text_parts = []
    for i in range(1, policy.required_steps + 1):
        final_text_parts.append(f"Step {i}: {repaired_steps[i]}")

    # Append original rationale if present (extracted via regex)
    rationale_match = re.search(r"(Rationale:.*)", raw_text, re.IGNORECASE | re.DOTALL)
    if rationale_match:
        final_text_parts.append(rationale_match.group(1))

    return "\n".join(final_text_parts), log


# ------------------------------------------------------------------------------
# Task 17, Step 3: Ensure final Rationale coherence
# ------------------------------------------------------------------------------
def validate_coherence(
    draft_text: str,
    rationale_text: str,
    llm_wrapper: Any,
    metadata: Dict[str, Any]
) -> Tuple[str, float, Dict[str, Any]]:
    """
    Validates that the rationale is coherent with the draft steps using Jaccard similarity.

    Args:
        draft_text (str): Combined text of steps.
        rationale_text (str): Final rationale.

    Returns:
        Tuple: (Final Rationale, Score, Log)
    """
    def tokenize(text):
        return set(re.findall(r"\w+", text.lower()))

    draft_tokens = tokenize(draft_text)
    rat_tokens = tokenize(rationale_text)

    if not draft_tokens or not rat_tokens:
        return rationale_text, 0.0, {"error": "empty_text"}

    # Compute Jaccard Score
    intersection = len(draft_tokens & rat_tokens)
    union = len(draft_tokens | rat_tokens)
    jaccard = intersection / union if union > 0 else 0.0

    log = {"jaccard_score": jaccard}

    if jaccard < 0.3:
        log["action"] = "repair_coherence"
        repair_prompt = f"Your rationale is not sufficiently grounded in your draft steps (Jaccard score {jaccard:.2f}). Rewrite the Rationale to synthesize the following steps:\n{draft_text}"

        response = _run_async_call_cod(
            wrapper=llm_wrapper,
            prompt=repair_prompt,
            system_prompt="You are ensuring coherence.",
            metadata={**metadata, "action": "cod_coherence_repair"}
        )
        rationale_text = response["content"]

    return rationale_text, jaccard, log


# ------------------------------------------------------------------------------
# Task 17, Orchestrator Function
# ------------------------------------------------------------------------------
def enforce_cod_protocol(
    agent_output: str,
    config: Dict[str, Any],
    llm_wrapper: Any,
    metadata: Dict[str, Any]
) -> Tuple[str, Dict[str, Any]]:
    """
    Orchestrates Chain-of-Draft enforcement.

    Args:
        agent_output (str): Raw output.
        config (Dict): Config.

    Returns:
        Tuple[str, Dict]: Validated output and compliance log.
    """
    policy = get_cod_policy(config)

    # Step 1 & 2: Validate Steps
    validated_text, step_log = validate_and_repair_cod(agent_output, policy, llm_wrapper, metadata)

    # Extract Rationale for Step 3
    rationale_match = re.search(r"Rationale:\s*(.*)", validated_text, re.IGNORECASE | re.DOTALL)
    rationale_text = rationale_match.group(1).strip() if rationale_match else ""

    # Extract Draft Text
    draft_text = re.sub(r"Rationale:.*", "", validated_text, flags=re.IGNORECASE | re.DOTALL).strip()

    # Step 3: Coherence
    final_rationale, score, coherence_log = validate_coherence(draft_text, rationale_text, llm_wrapper, metadata)

    # Reassemble
    final_output = f"{draft_text}\n\nRationale: {final_rationale}"

    return final_output, {"step_log": step_log, "coherence_log": coherence_log}
````

````python
# Task 18 — Execute backtesting simulations for all variants

# ==============================================================================
# Task 18: Execute backtesting simulations for all variants
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 18, Step 1: Define the three experimental variants
# ------------------------------------------------------------------------------
@dataclass
class ExperimentVariant:
    """
    Configuration for a specific experimental condition in the FedSight study.

    This dataclass encapsulates the parameters that define a distinct experimental
    variant, such as the baseline model, the In-Context Learning (ICL) variant,
    or the Chain-of-Draft (CoD) variant. It controls the activation of specific
    reasoning enhancements during the simulation.

    Attributes:
        name (str): A unique identifier for the variant (e.g., "baseline", "icl", "cod").
        enable_icl (bool): A flag indicating whether to inject historical lessons via
                           In-Context Learning into the agent's prompt.
        enable_cod (bool): A flag indicating whether to enforce Chain-of-Draft constraints
                           on the agent's reasoning process.
        description (str): A human-readable description of the variant's methodology.
    """
    name: str
    enable_icl: bool
    enable_cod: bool
    description: str


def define_variants() -> Dict[str, ExperimentVariant]:
    """
    Defines the three experimental variants for the FedSight study.

    This function constructs a registry of the experimental conditions required
    to reproduce the study's results. It defines the 'baseline', 'icl', and 'cod'
    variants with their respective feature toggles.

    Returns:
        Dict[str, ExperimentVariant]: A dictionary mapping variant names to their
                                      configuration objects.
    """
    return {
        "baseline": ExperimentVariant(
            name="baseline",
            enable_icl=False,
            enable_cod=False,
            description="Standard multi-agent simulation with no memory or enforcement."
        ),
        "icl": ExperimentVariant(
            name="icl",
            enable_icl=True,
            enable_cod=False,
            description="In-Context Learning with historical lesson injection."
        ),
        "cod": ExperimentVariant(
            name="cod",
            enable_icl=False,
            enable_cod=True,
            description="Chain-of-Draft enforcement on member reasoning."
        )
    }


# ------------------------------------------------------------------------------
# Task 18, Step 2: Execute 16 meetings × 5 runs per variant
# ------------------------------------------------------------------------------
class CoDInterceptionWrapper:
    """
    Wraps the LLM client to intercept Member calls and apply Chain-of-Draft (CoD) enforcement.

    This wrapper acts as a middleware layer that selectively applies the CoD validation
    and repair protocol to calls made by Member agents. It delegates standard calls
    to the underlying `LLMClientWrapper` but intercepts responses from Members to
    ensure they adhere to the strict reasoning constraints defined in the study.
    This is used exclusively when `enable_cod=True`.

    Attributes:
        base_wrapper (Any): The underlying `LLMClientWrapper` instance that handles
                            API communication, retries, and logging.
        config (Dict[str, Any]): The global configuration dictionary containing CoD policy settings.
        client (Any): The raw OpenAI client instance exposed from the base wrapper,
                      required for embedding operations in other parts of the pipeline.
    """

    def __init__(self, base_wrapper: Any, config: Dict[str, Any]):
        """
        Initializes the CoDInterceptionWrapper.

        Args:
            base_wrapper (Any): The underlying `LLMClientWrapper` instance.
            config (Dict[str, Any]): The global configuration dictionary.
        """
        self.base_wrapper = base_wrapper
        self.config = config
        # Expose client attribute for embedding calls if needed by other components
        self.client = base_wrapper.client

    async def call_completion(
        self,
        prompt: str,
        system_prompt: str,
        metadata: Dict[str, Any]
    ) -> Dict[str, Any]:
        """
        Delegates to the base wrapper, but applies CoD logic if the role is 'Member'.

        If the call originates from a Member agent, this method executes the standard
        completion call and then passes the result through the `enforce_cod_protocol`
        function (Task 17). This ensures that the final output conforms to the
        Chain-of-Draft constraints (e.g., step limits, word counts) before being
        returned to the simulation workflow.

        Args:
            prompt (str): The user prompt string.
            system_prompt (str): The system instruction string.
            metadata (Dict[str, Any]): Contextual metadata for logging (e.g., role, meeting_id).

        Returns:
            Dict[str, Any]: The response dictionary containing the (possibly repaired) content
                            and usage statistics. Note that usage from repair loops is logged
                            separately by the base wrapper but not aggregated into this single return object.
        """
        # Check if this is a Member call
        if metadata.get("role") == "Member":
            # 1. Initial Call
            # Execute the standard LLM call via the base wrapper
            response = await self.base_wrapper.call_completion(prompt, system_prompt, metadata)
            raw_content = response["content"]

            # 2. Enforce CoD (Task 17)
            # We assume enforce_cod_protocol is available from Task 17 context.
            # To ensure this code is functional, we assume the import or availability.
            # In a script, this would be imported.
            final_text, log = enforce_cod_protocol(
                agent_output=raw_content,
                config=self.config,
                llm_wrapper=self.base_wrapper, # Pass base wrapper to avoid recursion in repairs
                metadata=metadata
            )

            # Update content with the validated/repaired text
            response["content"] = final_text

            # Note: Usage from repair calls is logged by base_wrapper but not aggregated here
            # into the single response object. This is a known limitation accepted for
            # the "Average Tokens" metric which reads logs directly.
            return response
        else:
            # Pass through for Analyst, Economist, Drafter without modification
            return await self.base_wrapper.call_completion(prompt, system_prompt, metadata)


def execute_variant_runs(
    variant: ExperimentVariant,
    agent_packets: Dict[pd.Timestamp, Dict[str, Any]],
    member_agents_registry: List[Dict[str, Any]],
    config: Dict[str, Any],
    llm_wrapper: Any,
    memory_store: List[Dict[str, Any]]
) -> Tuple[List[Any], List[Dict[str, Any]]]:
    """
    Executes all simulations for a specific variant.

    Args:
        variant (ExperimentVariant): The variant config.
        agent_packets (Dict): Data packets keyed by meeting date.
        member_agents_registry (List): Base member personas.
        config (Dict): Global config.
        llm_wrapper (Any): Base client.
        memory_store (List): ICL memory.

    Returns:
        Tuple[List[RunArtifact], List[Dict]]: Artifacts and Ledger.
    """
    artifacts = []
    ledger = []

    # Sort meetings
    sorted_dates = sorted(agent_packets.keys())

    # Configure Wrapper
    if variant.enable_cod:
        active_wrapper = CoDInterceptionWrapper(llm_wrapper, config)
    else:
        active_wrapper = llm_wrapper

    for meeting_date in sorted_dates:
        packet = agent_packets[meeting_date]
        meeting_id = meeting_date # Timestamp

        # Prepare ICL Context if enabled
        run_config = copy.deepcopy(config)

        if variant.enable_icl:
            run_registry = []
            for member in member_agents_registry:
                # Retrieve lessons (Task 16 Step 2)
                request = construct_retrieval_query(config, packet)
                lessons = retrieve_relevant_lessons(
                    member["agent_id"], request, memory_store, llm_wrapper,
                    config["embedding_config"]["embedding_model_name"]
                )

                # Inject into System Prompt (Workaround for Task 13 template rigidity)
                # We prepend the lessons to the system prompt so the agent sees them.
                # Task 16 `inject_lessons_into_prompt` formats the block.
                augmented_system = inject_lessons_into_prompt(member["system_prompt"], lessons)

                new_member = member.copy()
                new_member["system_prompt"] = augmented_system
                run_registry.append(new_member)
        else:
            run_registry = member_agents_registry

        # Execute Runs
        for run_id in range(1, config["simulation_config"]["runs_per_meeting"] + 1):
            try:
                # 1. Run Simulation (Task 13)
                artifact = run_single_simulation(
                    meeting_id=meeting_id,
                    run_id=run_id,
                    variant=variant.name,
                    meeting_packet=packet,
                    member_agents_registry=run_registry,
                    config=run_config,
                    llm_wrapper=active_wrapper
                )

                # 2. Apply Mapping (Task 14)
                final_artifact = apply_vote_mapping(artifact)

                artifacts.append(final_artifact)
                ledger.append({
                    "variant": variant.name,
                    "meeting_id": str(meeting_id.date()),
                    "run_id": run_id,
                    "status": "success",
                    "timestamp": datetime.utcnow().isoformat()
                })

            except Exception as e:
                # Log Failure
                ledger.append({
                    "variant": variant.name,
                    "meeting_id": str(meeting_id.date()),
                    "run_id": run_id,
                    "status": "failed",
                    "error": str(e),
                    "traceback": traceback.format_exc(),
                    "timestamp": datetime.utcnow().isoformat()
                })
                # Continue to next run

    return artifacts, ledger


# ------------------------------------------------------------------------------
# Task 18, Orchestrator Function
# ------------------------------------------------------------------------------
def run_backtesting_campaign(
    config: Dict[str, Any],
    agent_packets: Dict[pd.Timestamp, Dict[str, Any]],
    member_agents_registry: List[Dict[str, Any]],
    llm_wrapper: Any,
    memory_store: List[Dict[str, Any]]
) -> Tuple[List[Any], List[Dict[str, Any]]]:
    """
    Orchestrates the full backtesting campaign across all variants.

    Args:
        config (Dict): Study config.
        agent_packets (Dict): Input data.
        member_agents_registry (List): Agents.
        llm_wrapper (Any): Client.
        memory_store (List): ICL memory.

    Returns:
        Tuple: All run artifacts and the full run ledger.
    """
    variants = define_variants()
    all_artifacts = []
    full_ledger = []

    # Order: Baseline -> ICL -> CoD
    for v_name in ["baseline", "icl", "cod"]:
        variant = variants[v_name]
        # In production, use logging instead of print
        # print(f"Starting variant: {v_name}")

        # Execute run
        artifacts, ledger = execute_variant_runs(
            variant, agent_packets, member_agents_registry,
            config, llm_wrapper, memory_store
        )

        all_artifacts.extend(artifacts)
        full_ledger.extend(ledger)

    return all_artifacts, full_ledger
````

````python
# Task 19 — Store all simulation artifacts for evaluation

# ==============================================================================
# Task 19: Store all simulation artifacts for evaluation
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 19, Step 1: Define the artifact storage schema
# ------------------------------------------------------------------------------
def initialize_artifact_storage(db_path: str = "fedsight_artifacts.db") -> None:
    """
    Initializes the SQLite database schema for storing simulation artifacts.

    Schema 'runs':
    - variant (TEXT): Experiment variant.
    - meeting_id (TEXT): ISO date string.
    - run_id (INTEGER): Run index.
    - analyst_output (TEXT): Raw output.
    - economist_options (TEXT): Raw output.
    - member_votes (TEXT): JSON serialized list of votes.
    - aggregated_decision (TEXT): Final decision string.
    - predicted_change_bps (INTEGER): Quantitative prediction.
    - generated_statement (TEXT): Simulated statement.
    - token_usage (TEXT): JSON serialized usage stats.
    - provenance (TEXT): JSON serialized metadata.

    Primary Key: (variant, meeting_id, run_id)

    Args:
        db_path (str): Path to the SQLite database file.
    """
    schema_sql = """
    CREATE TABLE IF NOT EXISTS runs (
        variant TEXT NOT NULL,
        meeting_id TEXT NOT NULL,
        run_id INTEGER NOT NULL,
        analyst_output TEXT,
        economist_options TEXT,
        member_votes TEXT,
        aggregated_decision TEXT,
        predicted_change_bps INTEGER,
        generated_statement TEXT,
        token_usage TEXT,
        provenance TEXT,
        PRIMARY KEY (variant, meeting_id, run_id)
    );
    """

    with sqlite3.connect(db_path) as conn:
        conn.execute(schema_sql)


# ------------------------------------------------------------------------------
# Task 19, Step 2 & 3: Store artifacts atomically and durably with provenance
# ------------------------------------------------------------------------------
def persist_run_artifact(
    artifact: 'RunArtifact',
    db_path: str = "fedsight_artifacts.db"
) -> None:
    """
    Persists a single RunArtifact to the database using an upsert operation.
    Handles serialization of nested structures (votes, usage, provenance) to JSON.

    Args:
        artifact (RunArtifact): The simulation artifact to store.
        db_path (str): Database path.
    """
    # Serialize nested fields
    # Note: meeting_id is Timestamp, convert to string
    meeting_id_str = str(artifact.meeting_id.date())

    # Helper to serialize objects
    def to_json(obj: Any) -> str:
        # Handle Timestamp inside dicts if any
        return json.dumps(obj, default=str)

    member_votes_json = to_json(artifact.member_votes)
    token_usage_json = to_json(artifact.token_usage)
    provenance_json = to_json(artifact.provenance)

    sql = """
    INSERT OR REPLACE INTO runs (
        variant, meeting_id, run_id,
        analyst_output, economist_options, member_votes,
        aggregated_decision, predicted_change_bps, generated_statement,
        token_usage, provenance
    ) VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?)
    """

    values = (
        artifact.variant,
        meeting_id_str,
        artifact.run_id,
        artifact.analyst_output,
        artifact.economist_options,
        member_votes_json,
        artifact.aggregated_decision,
        artifact.predicted_change_bps,
        artifact.generated_statement,
        token_usage_json,
        provenance_json
    )

    with sqlite3.connect(db_path) as conn:
        conn.execute(sql, values)


# ------------------------------------------------------------------------------
# Task 19, Orchestrator Function
# ------------------------------------------------------------------------------
def store_simulation_results(
    artifacts: List['RunArtifact'],
    db_path: str = "fedsight_artifacts.db"
) -> int:
    """
    Orchestrates the storage of a batch of simulation artifacts.

    Args:
        artifacts (List[RunArtifact]): List of artifacts to store.
        db_path (str): Database path.

    Returns:
        int: Number of artifacts successfully stored.
    """
    # Step 1: Ensure Schema
    initialize_artifact_storage(db_path)

    count = 0
    for art in artifacts:
        try:
            # Step 2 & 3: Persist
            persist_run_artifact(art, db_path)
            count += 1
        except Exception as e:
            print(f"Failed to persist artifact {art.variant}/{art.meeting_id}/{art.run_id}: {e}")

    return count
````

````python
# Task 20 — Create the orchestrator function for the end-to-end pipeline

# ==============================================================================
# Task 20: Create the orchestrator function for the end-to-end pipeline
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 20, Step 1: Define the orchestrator interface and dependency graph
# ------------------------------------------------------------------------------

@dataclass
class PipelineContext:
    """
    Holds the state of the FedSight pipeline execution.

    This dataclass serves as the central state container for the end-to-end simulation pipeline.
    It accumulates configuration data, intermediate processing results, agent registries,
    and final simulation artifacts as the pipeline progresses through its sequential stages.
    This design ensures that state is passed explicitly between tasks, facilitating
    modularity and testability.

    Attributes:
        config (Dict[str, Any]): The full, validated study configuration dictionary.
        config_hash (str): A SHA-256 hash of the configuration for provenance tracking.
        canonical_dates (Optional[Set[Any]]): A set of validated meeting dates (timestamps)
                                              defining the scope of the simulation.
        agent_packets (Optional[Dict[Any, Any]]): A dictionary of preprocessed data packets
                                                  prepared for agent consumption, keyed by meeting date.
        evaluation_packets (Optional[Dict[Any, Any]]): A dictionary of ground truth data packets
                                                       reserved for evaluation, keyed by meeting date.
        member_registry (Optional[List[Any]]): A list of instantiated member agent personas.
        llm_wrapper (Optional[Any]): The configured LLM client wrapper instance.
        memory_store (Optional[List[Any]]): The In-Context Learning (ICL) memory store containing
                                            historical lessons.
        run_artifacts (Optional[List[Any]]): A list of `RunArtifact` objects generated from the
                                             simulation backtesting phase.
        run_ledger (Optional[List[Any]]): A list of execution status records for each simulation run.
    """
    config: Dict[str, Any]
    config_hash: str
    canonical_dates: Optional[Set[Any]] = None
    agent_packets: Optional[Dict[Any, Any]] = None
    evaluation_packets: Optional[Dict[Any, Any]] = None
    member_registry: Optional[List[Any]] = None
    llm_wrapper: Optional[Any] = None
    memory_store: Optional[List[Any]] = None
    run_artifacts: Optional[List[Any]] = None
    run_ledger: Optional[List[Any]] = None



def run_fedsight_pipeline(
    df_macro_raw: pd.DataFrame,
    df_unstructured_raw: pd.DataFrame,
    df_participants_raw: pd.DataFrame,
    df_meeting_outcomes: pd.DataFrame,
    df_fomc_statements_actual: pd.DataFrame,
    config: Dict[str, Any],
    historical_data: Optional[Dict[str, Any]] = None, # For ICL
    db_path: str = "fedsight_artifacts.db",
    log_path: str = "llm_logs.jsonl"
) -> Dict[str, Any]:
    """
    Orchestrates the end-to-end FedSight AI simulation pipeline.

    Execution Flow:
    1.  **Validation**: Checks config and data schemas (Tasks 1-2).
    2.  **Cleansing**: Prepares macro, narrative, and participant data (Tasks 3-5).
    3.  **Construction**: Builds canonical meeting packets and leakage boundaries (Task 6).
    4.  **Preprocessing**: Verbalizes Dot Plots and bundles context (Tasks 7-8).
    5.  **Agent Setup**: Clusters participants and initializes LLM client (Tasks 9-11).
    6.  **Architecture**: Defines CrewAI structure (Task 12).
    7.  **Training**: Executes In-Context Learning if enabled (Task 15).
    8.  **Simulation**: Runs backtesting for all variants with CoD enforcement (Tasks 13, 14, 16, 17, 18).
    9.  **Storage**: Persists artifacts to database (Task 19).

    Args:
        df_macro_raw (pd.DataFrame): Raw macroeconomic indicators.
        df_unstructured_raw (pd.DataFrame): Raw text data (Beige Book, etc.).
        df_participants_raw (pd.DataFrame): Participant metadata for clustering.
        df_meeting_outcomes (pd.DataFrame): Ground truth decisions.
        df_fomc_statements_actual (pd.DataFrame): Ground truth statements.
        config (Dict): Full study configuration.
        historical_data (Optional[Dict]): Pre-constructed packets for ICL training meetings.
        db_path (str): Path for SQLite artifact storage.
        log_path (str): Path for LLM call logs.

    Returns:
        Dict[str, Any]: A dictionary containing:
            - 'artifacts': List of RunArtifact objects.
            - 'ledger': Execution status ledger.
            - 'context': The final PipelineContext object.
    """
    print("Starting FedSight AI Pipeline...")

    # --------------------------------------------------------------------------
    # Phase 1: Validation & Setup
    # --------------------------------------------------------------------------
    print("Phase 1: Validation...")
    # Task 1: Config
    validated_config, config_hash = validate_study_config(config)

    # Task 2: Dataframes
    data_dict = validate_all_dataframes(
        df_macro_raw, df_unstructured_raw, df_meeting_outcomes,
        df_fomc_statements_actual, df_participants_raw
    )

    ctx = PipelineContext(config=validated_config, config_hash=config_hash)

    # --------------------------------------------------------------------------
    # Phase 2: Data Cleansing & Preparation
    # --------------------------------------------------------------------------
    print("Phase 2: Cleansing...")
    # Task 3: Macro
    macro_res = cleanse_macro_data(data_dict["df_macro_raw"])

    # Task 4: Unstructured
    unstruct_res = cleanse_unstructured_data(data_dict["df_unstructured_raw"])

    # Task 5: Evaluation Data
    eval_res = cleanse_evaluation_data(
        data_dict["df_participants_raw"],
        data_dict["df_meeting_outcomes"],
        data_dict["df_fomc_statements_actual"]
    )

    # --------------------------------------------------------------------------
    # Phase 3: Packet Construction
    # --------------------------------------------------------------------------
    print("Phase 3: Packet Construction...")
    # Task 6: Join & Boundary
    canonical_res = construct_canonical_datasets(
        macro_res["df_macro_final"],
        unstruct_res["df_unstructured_clean"],
        eval_res["df_outcomes_clean"],
        eval_res["df_statements_clean"]
    )

    # Task 7: Dot Plot
    # Note: Modifies packets in place or returns new ones
    dot_plot_res = preprocess_dot_plots(canonical_res["agent_packets"])

    # Task 8: Context Bundle
    context_res = preprocess_agent_contexts(dot_plot_res["agent_packets"])

    ctx.agent_packets = context_res["agent_packets"]
    ctx.evaluation_packets = canonical_res["evaluation_packets"]

    # --------------------------------------------------------------------------
    # Phase 4: Agent Engineering
    # --------------------------------------------------------------------------
    print("Phase 4: Agent Engineering...")
    # Task 9: Clustering
    cluster_res = cluster_participants(
        eval_res["df_participants_clean"],
        eval_res["participant_feature_spec"]
    )

    # Task 10: Personas
    # We need to inject the cluster mapping into the config or pass it?
    # Task 10 reads from config. We assume config matches the cluster output
    # or we update config here. The prompt implies config drives it.
    # We proceed with config-based generation.
    ctx.member_registry = materialize_agents(ctx.config)

    # Task 11: LLM Client
    ctx.llm_wrapper = setup_llm_infrastructure(ctx.config, log_path)

    # --------------------------------------------------------------------------
    # Phase 5: Training (ICL)
    # --------------------------------------------------------------------------
    print("Phase 5: ICL Training...")
    # Task 15
    if historical_data:
        # Ensure historical packets are preprocessed similarly if raw
        # For this implementation, we assume they are ready-to-use packets
        ctx.memory_store = run_icl_training(
            ctx.config,
            historical_data["packets"],
            historical_data["outcomes"],
            ctx.member_registry,
            ctx.llm_wrapper
        )
    else:
        ctx.memory_store = []
        print("Skipping ICL: No historical data provided.")

    # --------------------------------------------------------------------------
    # Phase 6: Simulation (Backtesting)
    # --------------------------------------------------------------------------
    print("Phase 6: Simulation...")
    # Task 18 (which calls 13, 14, 16, 17)
    artifacts, ledger = run_backtesting_campaign(
        ctx.config,
        ctx.agent_packets,
        ctx.member_registry,
        ctx.llm_wrapper,
        ctx.memory_store
    )
    ctx.run_artifacts = artifacts
    ctx.run_ledger = ledger

    # --------------------------------------------------------------------------
    # Phase 7: Storage
    # --------------------------------------------------------------------------
    print("Phase 7: Storage...")
    # Task 19
    store_simulation_results(ctx.run_artifacts, db_path)

    print("Pipeline Complete.")

    return {
        "artifacts": ctx.run_artifacts,
        "ledger": ctx.run_ledger,
        "context": ctx
    }
````

````python
# Task 21 — Define meeting-level predictions from run-level predictions

# ==============================================================================
# Task 21: Define meeting-level predictions from run-level predictions
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 21, Step 1: Retrieve all run-level predictions for a meeting
# ------------------------------------------------------------------------------
def extract_run_predictions(run_artifacts: List['RunArtifact']) -> pd.DataFrame:
    """
    Extracts quantitative predictions from run artifacts into a structured DataFrame.

    Args:
        run_artifacts (List[RunArtifact]): List of simulation artifacts.

    Returns:
        pd.DataFrame: DataFrame with columns ['variant', 'meeting_id', 'run_id', 'predicted_change_bps'].
                      Indexed by (variant, meeting_id, run_id).
    """
    data = []
    for art in run_artifacts:
        # Ensure predicted_change_bps is present (Task 14 should have populated it)
        if art.predicted_change_bps is None:
            # Log warning or skip? Strict protocol implies we shouldn't have None here if Task 14 succeeded.
            # We'll skip and let the completeness check flag it.
            continue

        data.append({
            "variant": art.variant,
            "meeting_id": art.meeting_id, # Timestamp
            "run_id": art.run_id,
            "predicted_change_bps": art.predicted_change_bps
        })

    df = pd.DataFrame(data)

    if df.empty:
        return pd.DataFrame(columns=['variant', 'meeting_id', 'run_id', 'predicted_change_bps'])

    # Enforce types
    df["predicted_change_bps"] = df["predicted_change_bps"].astype(int)

    # Set Index
    df.set_index(["variant", "meeting_id", "run_id"], inplace=True)
    df.sort_index(inplace=True)

    return df


# ------------------------------------------------------------------------------
# Task 21, Step 2: Apply the configured run-to-meeting collapse rule
# ------------------------------------------------------------------------------
def collapse_runs_to_meeting(
    df_runs: pd.DataFrame
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Collapses 5 stochastic runs into a single meeting prediction using the Mode.
    Applies a deterministic tie-breaking rule.

    Tie-Break Rule:
    1. Smallest absolute magnitude (conservative).
    2. If abs equal (e.g. -25 vs +25), prefer 0 (status quo).
    3. If still tied (e.g. -25 vs +25 and 0 not in tie), prefer negative (dovish bias? or just deterministic sort).
       We use standard sort: -25 < +25.

    Args:
        df_runs (pd.DataFrame): Run-level predictions.

    Returns:
        Tuple[pd.DataFrame, pd.DataFrame]:
            - df_collapsed: Meeting-level predictions ['predicted_change_bps'].
            - df_audit: Audit trail of the aggregation logic.
    """
    # Utility for group aggregation
    def aggregate_group(series):
        counts = Counter(series)
        max_freq = max(counts.values())
        candidates = [val for val, count in counts.items() if count == max_freq]

        if len(candidates) == 1:
            winner = candidates[0]
            tie_break = False
        else:
            tie_break = True
            # Sort candidates by tie-break rules:
            # 1. Abs magnitude (ascending)
            # 2. Value itself (ascending) -> prefers negative in +/- tie
            candidates.sort(key=lambda x: (abs(x), x))
            winner = candidates[0]

        return pd.Series({
            "predicted_change_bps": winner,
            "tie_break_invoked": tie_break,
            "raw_values": sorted(list(series)), # For audit
            "candidates": candidates # For audit
        })

    # Group by variant and meeting
    # Reset index to access columns, then group
    grouped = df_runs.reset_index().groupby(["variant", "meeting_id"])["predicted_change_bps"]

    results = grouped.apply(aggregate_group)

    # Unstack or reshape is not needed if apply returns DataFrame/Series correctly aligned
    # results is a DataFrame with MultiIndex (variant, meeting_id) and columns from the Series
    df_collapsed = results[["predicted_change_bps"]].copy()
    df_audit = results[["tie_break_invoked", "raw_values", "candidates"]].copy()

    return df_collapsed, df_audit


# ----------------------------------------------------------------------------------
# Task 21, Step 3: Log meeting-level aggregation details (Implicit in Step 2 return)
# ----------------------------------------------------------------------------------
# Step 2 returns df_audit which fulfills this requirement.


# ------------------------------------------------------------------------------
# Task 21, Orchestrator Function
# ------------------------------------------------------------------------------
def compute_meeting_predictions(
    run_artifacts: List['RunArtifact']
) -> Dict[str, pd.DataFrame]:
    """
    Orchestrates the aggregation of run-level predictions into meeting-level decisions.

    Args:
        run_artifacts (List[RunArtifact]): The full set of simulation outputs.

    Returns:
        Dict[str, pd.DataFrame]:
            - 'run_predictions': DataFrame of all runs.
            - 'meeting_predictions': DataFrame of collapsed predictions.
            - 'aggregation_audit': DataFrame detailing the collapse logic.
    """
    # Step 1: Extract
    df_runs = extract_run_predictions(run_artifacts)

    # Step 2: Collapse
    df_meeting, df_audit = collapse_runs_to_meeting(df_runs)

    return {
        "run_predictions": df_runs,
        "meeting_predictions": df_meeting,
        "aggregation_audit": df_audit
    }
````

````python
# Task 22 — Compute Total Accuracy metric

# ==============================================================================
# Task 22: Compute Total Accuracy metric
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 22, Step 1: Retrieve ground truth and predictions for all meetings
# ------------------------------------------------------------------------------
def align_predictions_with_truth(
    meeting_predictions: pd.DataFrame,
    df_outcomes: pd.DataFrame
) -> pd.DataFrame:
    """
    Joins meeting-level predictions with ground truth outcomes.

    Args:
        meeting_predictions (pd.DataFrame): From Task 21.
        df_outcomes (pd.DataFrame): From Task 5.

    Returns:
        pd.DataFrame: Aligned DataFrame with 'y_pred' and 'y_true'.
    """
    # Prepare Truth
    truth = df_outcomes[["fftr_change_bps"]].copy()
    truth.index.name = "meeting_id"

    # Prepare Preds
    # meeting_predictions has MultiIndex (variant, meeting_id)
    # We reset index to join
    preds = meeting_predictions.reset_index()

    # Join
    # We join on meeting_id. Truth is broadcast across variants.
    aligned = pd.merge(
        preds,
        truth,
        left_on="meeting_id",
        right_index=True,
        how="inner",
        validate="many_to_one"
    )

    # Rename for clarity
    aligned.rename(columns={
        "predicted_change_bps": "y_pred",
        "fftr_change_bps": "y_true"
    }, inplace=True)

    return aligned


# ------------------------------------------------------------------------------
# Task 22, Step 2: Apply the Total Accuracy formula
# ------------------------------------------------------------------------------
def calculate_total_accuracy(aligned_df: pd.DataFrame) -> pd.Series:
    """
    Computes Total Accuracy per variant.

    Formula: 1/n * Sum(I(y_hat == y))

    Args:
        aligned_df (pd.DataFrame): Aligned data.

    Returns:
        pd.Series: Accuracy score indexed by variant.
    """
    # Check equality
    aligned_df["is_correct"] = (aligned_df["y_pred"] == aligned_df["y_true"]).astype(int)

    # Group by variant and mean
    accuracy = aligned_df.groupby("variant")["is_correct"].mean()

    return accuracy


# ------------------------------------------------------------------------------
# Task 22, Orchestrator Function
# ------------------------------------------------------------------------------
def compute_total_accuracy(
    meeting_predictions: pd.DataFrame,
    df_outcomes: pd.DataFrame
) -> pd.DataFrame:
    """
    Orchestrates Total Accuracy computation.

    Args:
        meeting_predictions (pd.DataFrame): Predictions.
        df_outcomes (pd.DataFrame): Truth.

    Returns:
        pd.DataFrame: DataFrame with 'total_accuracy' per variant.
    """
    # Step 1: Align
    aligned = align_predictions_with_truth(meeting_predictions, df_outcomes)

    # Step 2: Calculate
    accuracy_series = calculate_total_accuracy(aligned)

    # Format as DataFrame
    results = accuracy_series.to_frame(name="total_accuracy")

    return results
````

````python
# Task 23 — Compute Agent Accuracy metric

# ==============================================================================
# Task 23: Compute Agent Accuracy metric
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 23, Step 1: Define the "correct" vote and agent meeting-level vote
# ------------------------------------------------------------------------------
def derive_correct_and_agent_votes(
    run_artifacts: List['RunArtifact'],
    df_outcomes: pd.DataFrame
) -> Tuple[pd.Series, pd.DataFrame]:
    """
    Derives the ground truth vote for each meeting and the modal vote for each agent.

    Correct Vote Logic:
    - Change > 0 bps -> Option C (Hawkish)
    - Change = 0 bps -> Option B (Neutral)
    - Change < 0 bps -> Option A (Dovish)

    Agent Vote Logic:
    - Mode of votes across 5 runs.
    - Tie-break: Option B (Neutral) if involved, else alphabetical.

    Args:
        run_artifacts (List[RunArtifact]): Simulation outputs.
        df_outcomes (pd.DataFrame): Ground truth outcomes.

    Returns:
        Tuple[pd.Series, pd.DataFrame]:
            - correct_votes: Series indexed by meeting_id with values 'Option A/B/C'.
            - agent_votes: DataFrame indexed by (variant, meeting_id, agent_id) with 'modal_vote'.
    """
    # 1. Derive Correct Votes
    def map_bps_to_option(bps: int) -> str:
        if bps > 0: return "Option C"
        if bps < 0: return "Option A"
        return "Option B"

    # Ensure index is meeting_id (Timestamp)
    if "meeting_date" in df_outcomes.columns:
        truth_df = df_outcomes.set_index("meeting_date")
    else:
        truth_df = df_outcomes

    correct_votes = truth_df["fftr_change_bps"].apply(map_bps_to_option)
    correct_votes.name = "correct_vote"

    # 2. Derive Agent Votes
    # Extract all individual votes
    vote_records = []
    for art in run_artifacts:
        for vote_rec in art.member_votes:
            vote_records.append({
                "variant": art.variant,
                "meeting_id": art.meeting_id,
                "agent_id": vote_rec["agent_id"],
                "vote": vote_rec["vote"]
            })

    df_votes = pd.DataFrame(vote_records)

    # Compute Mode per Agent
    def get_modal_vote(series):
        counts = Counter(series)
        max_freq = max(counts.values())
        candidates = [v for v, c in counts.items() if c == max_freq]

        if len(candidates) == 1:
            return candidates[0]

        # Tie-break
        if "Option B" in candidates:
            return "Option B"
        return sorted(candidates)[0]

    agent_votes = df_votes.groupby(["variant", "meeting_id", "agent_id"])["vote"].apply(get_modal_vote)
    agent_votes = agent_votes.to_frame(name="modal_vote")

    return correct_votes, agent_votes


# ------------------------------------------------------------------------------
# Task 23, Step 2: Apply the Agent Accuracy formula
# ------------------------------------------------------------------------------
def calculate_agent_accuracy(
    correct_votes: pd.Series,
    agent_votes: pd.DataFrame
) -> pd.Series:
    """
    Computes Agent Accuracy per variant.

    Formula: 1/N * Sum_i ( 1/K * Sum_k ( I(Vote_ik == Vote*_i) ) )

    Args:
        correct_votes (pd.Series): Truth.
        agent_votes (pd.DataFrame): Agent modal votes.

    Returns:
        pd.Series: Accuracy score indexed by variant.
    """
    # Join Truth to Agent Votes
    # agent_votes has MultiIndex (variant, meeting_id, agent_id)
    # correct_votes has Index (meeting_id)
    # Reset index to join
    df = agent_votes.reset_index()

    # Merge
    merged = pd.merge(
        df,
        correct_votes,
        left_on="meeting_id",
        right_index=True,
        how="inner"
    )

    # Check match
    merged["is_correct"] = (merged["modal_vote"] == merged["correct_vote"]).astype(int)

    # Average over agents (K) then meetings (N)
    # Or simply average over all rows since N*K is the total count and weights are equal
    # Formula: 1/N * Sum_i ( 1/K * Sum_k ... ) is equivalent to Total_Correct / (N*K)
    # provided every meeting has K agents.
    accuracy = merged.groupby("variant")["is_correct"].mean()

    return accuracy


# ------------------------------------------------------------------------------
# Task 23, Orchestrator Function
# ------------------------------------------------------------------------------
def compute_agent_accuracy(
    run_artifacts: List['RunArtifact'],
    df_outcomes: pd.DataFrame
) -> pd.DataFrame:
    """
    Orchestrates Agent Accuracy computation.

    Args:
        run_artifacts (List[RunArtifact]): Simulation outputs.
        df_outcomes (pd.DataFrame): Truth.

    Returns:
        pd.DataFrame: DataFrame with 'agent_accuracy' per variant.
    """
    # Step 1: Derive Votes
    correct_votes, agent_votes = derive_correct_and_agent_votes(run_artifacts, df_outcomes)

    # Step 2: Calculate
    accuracy_series = calculate_agent_accuracy(correct_votes, agent_votes)

    # Format
    results = accuracy_series.to_frame(name="agent_accuracy")

    return results
````

````python
# Task 24 — Compute Voting Stability metric

# ==============================================================================
# Task 24: Compute Voting Stability metric
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 24, Step 1: Compute the modal vote across runs
# ------------------------------------------------------------------------------
def compute_modal_votes(
    run_artifacts: List['RunArtifact']
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Computes the modal vote for each agent-meeting pair across runs.

    Args:
        run_artifacts (List[RunArtifact]): Simulation outputs.

    Returns:
        Tuple[pd.DataFrame, pd.DataFrame]:
            - df_run_votes: All individual votes indexed by (variant, meeting_id, run_id, agent_id).
            - df_modal_votes: Modal votes indexed by (variant, meeting_id, agent_id).
    """
    # Extract all votes
    records = []
    for art in run_artifacts:
        for v in art.member_votes:
            records.append({
                "variant": art.variant,
                "meeting_id": art.meeting_id,
                "run_id": art.run_id,
                "agent_id": v["agent_id"],
                "vote": v["vote"]
            })

    df_run_votes = pd.DataFrame(records)

    # Compute Mode
    def get_mode(series):
        counts = Counter(series)
        max_freq = max(counts.values())
        candidates = [v for v, c in counts.items() if c == max_freq]

        if len(candidates) == 1:
            return candidates[0]
        if "Option B" in candidates:
            return "Option B"
        return sorted(candidates)[0]

    df_modal_votes = df_run_votes.groupby(["variant", "meeting_id", "agent_id"])["vote"].apply(get_mode)
    df_modal_votes = df_modal_votes.to_frame(name="modal_vote")

    return df_run_votes, df_modal_votes


# ------------------------------------------------------------------------------
# Task 24, Step 2: Apply the Voting Stability formula
# ------------------------------------------------------------------------------
def calculate_voting_stability(
    df_run_votes: pd.DataFrame,
    df_modal_votes: pd.DataFrame
) -> pd.Series:
    """
    Computes Voting Stability per variant.

    Formula: Avg( I(Vote_ijk == Mode_ik) )

    Args:
        df_run_votes (pd.DataFrame): Individual votes.
        df_modal_votes (pd.DataFrame): Modal votes.

    Returns:
        pd.Series: Stability score indexed by variant.
    """
    # Join
    # df_run_votes has (variant, meeting_id, run_id, agent_id) columns (if reset index)
    # df_modal_votes has index (variant, meeting_id, agent_id)

    merged = pd.merge(
        df_run_votes,
        df_modal_votes,
        on=["variant", "meeting_id", "agent_id"],
        how="left"
    )

    # Check match
    merged["is_stable"] = (merged["vote"] == merged["modal_vote"]).astype(int)

    # Average
    stability = merged.groupby("variant")["is_stable"].mean()

    return stability


# ------------------------------------------------------------------------------
# Task 24, Orchestrator Function
# ------------------------------------------------------------------------------
def compute_voting_stability(
    run_artifacts: List['RunArtifact']
) -> pd.DataFrame:
    """
    Orchestrates Voting Stability computation.

    Args:
        run_artifacts (List[RunArtifact]): Simulation outputs.

    Returns:
        pd.DataFrame: DataFrame with 'voting_stability' per variant.
    """
    # Step 1: Compute Modes
    df_run_votes, df_modal_votes = compute_modal_votes(run_artifacts)

    # Step 2: Calculate
    stability_series = calculate_voting_stability(df_run_votes, df_modal_votes)

    # Format
    results = stability_series.to_frame(name="voting_stability")

    return results
````

````python
# Task 25 — Compute Semantic Similarity metric

# ==============================================================================
# Task 25: Compute Semantic Similarity metric
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 25, Step 1: Embed generated and actual statements
# ------------------------------------------------------------------------------
def generate_statement_embeddings(
    run_artifacts: List['RunArtifact'],
    df_actual_statements: pd.DataFrame,
    config: Dict[str, Any],
    llm_wrapper: Any
) -> Dict[str, np.ndarray]:
    """
    Generates embeddings for all generated and actual FOMC statements.
    Uses caching by text hash to minimize API calls.

    Args:
        run_artifacts (List[RunArtifact]): Simulation outputs.
        df_actual_statements (pd.DataFrame): Ground truth statements.
        config (Dict): Study configuration.
        llm_wrapper (Any): Client wrapper.

    Returns:
        Dict[str, np.ndarray]: Map of text_hash -> embedding vector.
    """
    embedding_model = config["embedding_config"]["embedding_model_name"]
    norm_rules = config["embedding_config"]["text_normalization_rules"]

    # Helper: Normalize Text
    def normalize(text: str) -> str:
        if not text: return ""
        if norm_rules.get("strip_whitespace", True):
            text = text.strip()
        if norm_rules.get("collapse_multiple_spaces", True):
            text = re.sub(r'\s+', ' ', text)
        if norm_rules.get("lowercase", False):
            text = text.lower()
        return text

    # Collect all unique texts
    unique_texts = set()

    # 1. Generated Statements
    for art in run_artifacts:
        unique_texts.add(normalize(art.generated_statement))

    # 2. Actual Statements
    for text in df_actual_statements["fomc_statement_actual_text"]:
        unique_texts.add(normalize(text))

    # Remove empty strings
    unique_texts.discard("")

    # Batch Embed
    embeddings_map = {}

    async def _batch_embed():
        # Simple sequential embedding for robustness (or batch if API allows)
        for text in unique_texts:
            text_hash = hashlib.sha256(text.encode("utf-8")).hexdigest()
            try:
                resp = await llm_wrapper.client.embeddings.create(
                    input=text,
                    model=embedding_model
                )
                embeddings_map[text_hash] = np.array(resp.data[0].embedding)
            except Exception as e:
                print(f"Embedding failed for text hash {text_hash}: {e}")
                # Store zero vector or handle downstream
                embeddings_map[text_hash] = np.zeros(1) # Placeholder size, will fail norm check

    # Execute Async
    try:
        loop = asyncio.get_event_loop()
    except RuntimeError:
        loop = asyncio.new_event_loop()
        asyncio.set_event_loop(loop)

    if loop.is_running():
        import nest_asyncio
        nest_asyncio.apply()
        loop.run_until_complete(_batch_embed())
    else:
        loop.run_until_complete(_batch_embed())

    return embeddings_map


# ------------------------------------------------------------------------------
# Task 25, Step 2: Compute cosine similarity per meeting-run
# ------------------------------------------------------------------------------
def calculate_run_similarities(
    run_artifacts: List['RunArtifact'],
    df_actual_statements: pd.DataFrame,
    embeddings_map: Dict[str, np.ndarray],
    config: Dict[str, Any]
) -> pd.DataFrame:
    """
    Computes cosine similarity for each run against the actual statement.

    Args:
        run_artifacts (List[RunArtifact]): Simulation outputs.
        df_actual_statements (pd.DataFrame): Truth.
        embeddings_map (Dict): Pre-computed embeddings.
        config (Dict): Config for normalization rules.

    Returns:
        pd.DataFrame: DataFrame with 'similarity_score', indexed by (variant, meeting_id, run_id).
    """
    norm_rules = config["embedding_config"]["text_normalization_rules"]

    def normalize(text: str) -> str:
        if not text: return ""
        if norm_rules.get("strip_whitespace", True):
            text = text.strip()
        if norm_rules.get("collapse_multiple_spaces", True):
            text = re.sub(r'\s+', ' ', text)
        if norm_rules.get("lowercase", False):
            text = text.lower()
        return text

    # Index actual statements by meeting_id (Timestamp)
    # Ensure index is Timestamp
    if "meeting_date" in df_actual_statements.columns:
        actual_map = df_actual_statements.set_index("meeting_date")["fomc_statement_actual_text"]
    else:
        actual_map = df_actual_statements["fomc_statement_actual_text"]

    results = []

    for art in run_artifacts:
        # Get Generated Vector
        gen_text = normalize(art.generated_statement)
        gen_hash = hashlib.sha256(gen_text.encode("utf-8")).hexdigest()
        gen_vec = embeddings_map.get(gen_hash)

        # Get Actual Vector
        # art.meeting_id is Timestamp
        if art.meeting_id not in actual_map.index:
            # Missing ground truth statement
            continue

        act_text = normalize(actual_map.loc[art.meeting_id])
        act_hash = hashlib.sha256(act_text.encode("utf-8")).hexdigest()
        act_vec = embeddings_map.get(act_hash)

        if gen_vec is None or act_vec is None:
            sim = 0.0
        else:
            norm_g = np.linalg.norm(gen_vec)
            norm_a = np.linalg.norm(act_vec)

            if norm_g == 0 or norm_a == 0:
                sim = 0.0
            else:
                sim = np.dot(gen_vec, act_vec) / (norm_g * norm_a)

        results.append({
            "variant": art.variant,
            "meeting_id": art.meeting_id,
            "run_id": art.run_id,
            "similarity_score": float(sim)
        })

    df = pd.DataFrame(results)
    if not df.empty:
        df.set_index(["variant", "meeting_id", "run_id"], inplace=True)

    return df


# ------------------------------------------------------------------------------
# Task 25, Step 3: Aggregate to meeting level and overall
# ------------------------------------------------------------------------------
def aggregate_similarity_metrics(
    df_run_similarities: pd.DataFrame
) -> pd.Series:
    """
    Aggregates similarity scores.

    1. Average over runs per meeting.
    2. Average over meetings per variant.

    Args:
        df_run_similarities (pd.DataFrame): Run-level scores.

    Returns:
        pd.Series: Semantic Similarity per variant.
    """
    # Avg over runs (level 2)
    meeting_sim = df_run_similarities.groupby(["variant", "meeting_id"])["similarity_score"].mean()

    # Avg over meetings (level 1)
    variant_sim = meeting_sim.groupby("variant").mean()

    return variant_sim


# ------------------------------------------------------------------------------
# Task 25, Orchestrator Function
# ------------------------------------------------------------------------------
def compute_semantic_similarity(
    run_artifacts: List['RunArtifact'],
    df_actual_statements: pd.DataFrame,
    config: Dict[str, Any],
    llm_wrapper: Any
) -> pd.DataFrame:
    """
    Orchestrates Semantic Similarity computation.

    Args:
        run_artifacts (List[RunArtifact]): Outputs.
        df_actual_statements (pd.DataFrame): Truth.
        config (Dict): Config.
        llm_wrapper (Any): Client.

    Returns:
        pd.DataFrame: DataFrame with 'semantic_similarity' per variant.
    """
    # Step 1: Embed
    embeddings_map = generate_statement_embeddings(
        run_artifacts, df_actual_statements, config, llm_wrapper
    )

    # Step 2: Compute Run Similarities
    df_runs = calculate_run_similarities(
        run_artifacts, df_actual_statements, embeddings_map, config
    )

    # Step 3: Aggregate
    sim_series = aggregate_similarity_metrics(df_runs)

    # Format
    results = sim_series.to_frame(name="semantic_similarity")

    return results
````

````python
# Task 26 — Compute Average Tokens metric

# ==============================================================================
# Task 26: Compute Average Tokens metric
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 26, Step 1: Retrieve token usage for all meeting-runs
# ------------------------------------------------------------------------------
def extract_token_usage(run_artifacts: List['RunArtifact']) -> pd.DataFrame:
    """
    Extracts total token usage from run artifacts.

    Args:
        run_artifacts (List[RunArtifact]): Simulation outputs.

    Returns:
        pd.DataFrame: DataFrame with 'total_tokens', indexed by (variant, meeting_id, run_id).
    """
    data = []
    for art in run_artifacts:
        usage = art.token_usage.get("total_tokens", 0)
        data.append({
            "variant": art.variant,
            "meeting_id": art.meeting_id,
            "run_id": art.run_id,
            "total_tokens": usage
        })

    df = pd.DataFrame(data)
    if not df.empty:
        df.set_index(["variant", "meeting_id", "run_id"], inplace=True)

    return df


# ------------------------------------------------------------------------------
# Task 26, Step 2: Compute Average Tokens per meeting-run
# ------------------------------------------------------------------------------
def calculate_average_tokens(token_df: pd.DataFrame) -> pd.Series:
    """
    Computes Average Tokens per variant.

    Formula: 1/(n*J) * Sum(tokens)

    Args:
        token_df (pd.DataFrame): Token usage data.

    Returns:
        pd.Series: Average tokens indexed by variant.
    """
    # Mean over all runs per variant
    avg_tokens = token_df.groupby("variant")["total_tokens"].mean()
    return avg_tokens


# ------------------------------------------------------------------------------
# Task 26, Orchestrator Function
# ------------------------------------------------------------------------------
def compute_average_tokens(
    run_artifacts: List['RunArtifact']
) -> pd.DataFrame:
    """
    Orchestrates Average Tokens computation.

    Args:
        run_artifacts (List[RunArtifact]): Simulation outputs.

    Returns:
        pd.DataFrame: DataFrame with 'average_tokens' per variant.
    """
    # Step 1: Extract
    df_tokens = extract_token_usage(run_artifacts)

    # Step 2: Calculate
    avg_series = calculate_average_tokens(df_tokens)

    # Format
    results = avg_series.to_frame(name="average_tokens")

    return results
````

````python
# Task 27 — Compute Mean Absolute Error (MAE) metric

# ==============================================================================
# Task 27: Compute Mean Absolute Error (MAE) metric
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 27, Step 1: Retrieve ground truth and meeting-level predictions
# ------------------------------------------------------------------------------
def align_predictions_for_mae(
    meeting_predictions: pd.DataFrame,
    df_outcomes: pd.DataFrame
) -> pd.DataFrame:
    """
    Joins meeting-level predictions with ground truth outcomes for MAE calculation.
    Ensures both are in basis points.

    Args:
        meeting_predictions (pd.DataFrame): From Task 21.
        df_outcomes (pd.DataFrame): From Task 5.

    Returns:
        pd.DataFrame: Aligned DataFrame with 'y_pred' and 'y_true' (bps).
    """
    # Prepare Truth
    truth = df_outcomes[["fftr_change_bps"]].copy()
    truth.index.name = "meeting_id"

    # Prepare Preds
    preds = meeting_predictions.reset_index()

    # Join
    aligned = pd.merge(
        preds,
        truth,
        left_on="meeting_id",
        right_index=True,
        how="inner",
        validate="many_to_one"
    )

    # Rename
    aligned.rename(columns={
        "predicted_change_bps": "y_pred",
        "fftr_change_bps": "y_true"
    }, inplace=True)

    return aligned


# ------------------------------------------------------------------------------
# Task 27, Step 2: Apply the MAE formula
# ------------------------------------------------------------------------------
def calculate_mae(aligned_df: pd.DataFrame) -> pd.Series:
    """
    Computes Mean Absolute Error per variant.

    Formula: 1/n * Sum(|y_hat - y|)

    Args:
        aligned_df (pd.DataFrame): Aligned data.

    Returns:
        pd.Series: MAE in basis points indexed by variant.
    """
    # Absolute Error
    aligned_df["abs_error"] = (aligned_df["y_pred"] - aligned_df["y_true"]).abs()

    # Mean
    mae = aligned_df.groupby("variant")["abs_error"].mean()

    return mae


# ------------------------------------------------------------------------------
# Task 27, Orchestrator Function
# ------------------------------------------------------------------------------
def compute_mae(
    meeting_predictions: pd.DataFrame,
    df_outcomes: pd.DataFrame
) -> pd.DataFrame:
    """
    Orchestrates MAE computation.

    Args:
        meeting_predictions (pd.DataFrame): Predictions.
        df_outcomes (pd.DataFrame): Truth.

    Returns:
        pd.DataFrame: DataFrame with 'mae_bps' per variant.
    """
    # Step 1: Align
    aligned = align_predictions_for_mae(meeting_predictions, df_outcomes)

    # Step 2: Calculate
    mae_series = calculate_mae(aligned)

    # Format
    results = mae_series.to_frame(name="mae_bps")

    return results
````

````python
# Task 28 — Produce the final results table and comparison across variants

# ==============================================================================
# Task 28: Produce the final results table and comparison across variants
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 28, Step 1: Aggregate all six metrics into a canonical results table
# ------------------------------------------------------------------------------
def aggregate_metrics(
    total_accuracy: pd.DataFrame,
    agent_accuracy: pd.DataFrame,
    voting_stability: pd.DataFrame,
    semantic_similarity: pd.DataFrame,
    average_tokens: pd.DataFrame,
    mae: pd.DataFrame
) -> pd.DataFrame:
    """
    Aggregates individual metric DataFrames into a single results table.

    Args:
        total_accuracy (pd.DataFrame): 'total_accuracy' per variant.
        agent_accuracy (pd.DataFrame): 'agent_accuracy' per variant.
        voting_stability (pd.DataFrame): 'voting_stability' per variant.
        semantic_similarity (pd.DataFrame): 'semantic_similarity' per variant.
        average_tokens (pd.DataFrame): 'average_tokens' per variant.
        mae (pd.DataFrame): 'mae_bps' per variant.

    Returns:
        pd.DataFrame: Combined DataFrame indexed by variant.
    """
    dfs = [
        total_accuracy, agent_accuracy, voting_stability,
        semantic_similarity, average_tokens, mae
    ]

    # Merge all on index (variant)
    results = pd.concat(dfs, axis=1, join="outer")

    # Ensure column order
    cols = [
        "total_accuracy", "agent_accuracy", "voting_stability",
        "semantic_similarity", "average_tokens", "mae_bps"
    ]
    results = results[cols]

    return results


# ------------------------------------------------------------------------------
# Task 28, Step 2: Format results for reporting
# ------------------------------------------------------------------------------
def format_results_table(results_df: pd.DataFrame) -> pd.DataFrame:
    """
    Formats the results DataFrame for display.

    Formatting Rules:
    - Accuracy/Stability/Similarity: Percentage with 2 decimals (e.g., 93.75%).
    - Average Tokens: Comma-separated integer (e.g., 60,464).
    - MAE: 4 decimal places (e.g., 0.0156).

    Args:
        results_df (pd.DataFrame): Raw metrics.

    Returns:
        pd.DataFrame: Formatted string DataFrame.
    """
    formatted = results_df.copy()

    # Percentages
    pct_cols = ["total_accuracy", "agent_accuracy", "voting_stability", "semantic_similarity"]
    for col in pct_cols:
        if col in formatted.columns:
            formatted[col] = formatted[col].apply(lambda x: f"{x*100:.2f}%" if pd.notnull(x) else "N/A")

    # Tokens
    if "average_tokens" in formatted.columns:
        formatted["average_tokens"] = formatted["average_tokens"].apply(lambda x: f"{int(x):,}" if pd.notnull(x) else "N/A")

    # MAE
    if "mae_bps" in formatted.columns:
        formatted["mae_bps"] = formatted["mae_bps"].apply(lambda x: f"{x:.4f}" if pd.notnull(x) else "N/A")

    return formatted


# ------------------------------------------------------------------------------
# Task 28, Step 3: Compare reproduced metrics against paper benchmarks
# ------------------------------------------------------------------------------
def compare_against_benchmarks(results_df: pd.DataFrame) -> pd.DataFrame:
    """
    Compares reproduced FedSight CoD results against paper benchmarks.

    Paper Benchmarks (FedSight CoD):
    - Total Accuracy: 93.75% (0.9375)
    - Agent Accuracy: 90.22% (0.9022)
    - Voting Stability: 93.33% (0.9333)
    - Similarity: 73.82% (0.7382)
    - Average Tokens: 60,464
    - MAE: 0.0156

    Args:
        results_df (pd.DataFrame): Raw metrics.

    Returns:
        pd.DataFrame: Comparison table for 'cod' variant.
    """
    benchmarks = {
        "total_accuracy": 0.9375,
        "agent_accuracy": 0.9022,
        "voting_stability": 0.9333,
        "semantic_similarity": 0.7382,
        "average_tokens": 60464,
        "mae_bps": 0.0156
    }

    if "cod" not in results_df.index:
        return pd.DataFrame()

    cod_results = results_df.loc["cod"]

    comparison_data = []
    for metric, bench_val in benchmarks.items():
        repro_val = cod_results.get(metric)
        if pd.notnull(repro_val):
            delta = repro_val - bench_val
            comparison_data.append({
                "metric": metric,
                "paper_value": bench_val,
                "reproduced_value": repro_val,
                "delta": delta
            })

    return pd.DataFrame(comparison_data).set_index("metric")


# ------------------------------------------------------------------------------
# Task 28, Orchestrator Function
# ------------------------------------------------------------------------------
def generate_final_report(
    metrics_dict: Dict[str, pd.DataFrame]
) -> Dict[str, pd.DataFrame]:
    """
    Orchestrates the generation of the final results report.

    Args:
        metrics_dict (Dict): Dictionary of metric DataFrames.

    Returns:
        Dict[str, pd.DataFrame]:
            - 'raw_results': Aggregated raw metrics.
            - 'formatted_results': Display-ready table.
            - 'benchmark_comparison': Comparison vs paper.
    """
    # Step 1: Aggregate
    raw_results = aggregate_metrics(
        metrics_dict["total_accuracy"],
        metrics_dict["agent_accuracy"],
        metrics_dict["voting_stability"],
        metrics_dict["semantic_similarity"],
        metrics_dict["average_tokens"],
        metrics_dict["mae"]
    )

    # Step 2: Format
    formatted_results = format_results_table(raw_results)

    # Step 3: Compare
    comparison = compare_against_benchmarks(raw_results)

    return {
        "raw_results": raw_results,
        "formatted_results": formatted_results,
        "benchmark_comparison": comparison
    }
````

````python
# Task 29 — Produce the reproducibility package and deviation report

# ==============================================================================
# Task 29: Produce the reproducibility package and deviation report
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 29, Step 1: Package all artifacts for reproducibility
# ------------------------------------------------------------------------------
def create_reproducibility_package(
    ctx: 'PipelineContext',
    raw_data_frames: Dict[str, pd.DataFrame],
    output_dir: str = "fedsight_reproduction"
) -> Dict[str, str]:
    """
    Creates a structured archive of all study inputs, outputs, and configuration.

    Directory Structure:
    - config/ (JSON config)
    - data/ (Parquet dumps of raw inputs)
    - artifacts/ (JSONL of run artifacts)
    - results/ (CSV of metrics)
    - logs/ (Placeholder for log files)

    Args:
        ctx (PipelineContext): The final state of the pipeline.
        raw_data_frames (Dict[str, pd.DataFrame]): Map of name -> DataFrame for inputs.
        output_dir (str): Target directory.

    Returns:
        Dict[str, str]: Map of component names to their file paths in the package.
    """
    base_path = Path(output_dir)
    if base_path.exists():
        shutil.rmtree(base_path)
    base_path.mkdir(parents=True)

    paths = {}

    # 1. Config
    config_dir = base_path / "config"
    config_dir.mkdir()
    config_path = config_dir / "fedsight_study_config.json"
    with open(config_path, "w") as f:
        json.dump(ctx.config, f, indent=2, default=str)
    paths["config"] = str(config_path)

    # 2. Data
    data_dir = base_path / "data"
    data_dir.mkdir()
    for name, df in raw_data_frames.items():
        file_path = data_dir / f"{name}.parquet"
        # Ensure index is preserved
        df.to_parquet(file_path, index=True)
        paths[f"data_{name}"] = str(file_path)

    # 3. Artifacts
    artifacts_dir = base_path / "artifacts"
    artifacts_dir.mkdir()
    if ctx.run_artifacts:
        artifacts_path = artifacts_dir / "run_artifacts.jsonl"
        with open(artifacts_path, "w") as f:
            for art in ctx.run_artifacts:
                # Convert dataclass to dict
                d = asdict(art)
                # Handle Timestamp serialization
                d["meeting_id"] = str(d["meeting_id"])
                f.write(json.dumps(d, default=str) + "\n")
        paths["artifacts"] = str(artifacts_path)

    # 4. Results
    results_dir = base_path / "results"
    results_dir.mkdir()
    if ctx.metrics is not None:
        metrics_path = results_dir / "metrics.csv"
        ctx.metrics.to_csv(metrics_path)
        paths["metrics"] = str(metrics_path)

    # 5. Logs
    logs_dir = base_path / "logs"
    logs_dir.mkdir()
    # Placeholder: In a real run, we'd copy the log file from ctx.config or known path

    return paths


# ------------------------------------------------------------------------------
# Task 29, Step 2: Document all implementation choices in a deviation report
# ------------------------------------------------------------------------------
def generate_deviation_report(
    ctx: 'PipelineContext',
    output_dir: str = "fedsight_reproduction"
) -> str:
    """
    Generates a Markdown report documenting implementation choices and parameters.

    Args:
        ctx (PipelineContext): Pipeline state.
        output_dir (str): Directory to save report.

    Returns:
        str: Path to the report file.
    """
    report_path = Path(output_dir) / "results" / "deviation_report.md"

    # Extract configs safely
    sim_config = ctx.config.get("simulation_config", {})
    llm_config = ctx.config.get("llm_engine_config", {})
    prep_config = ctx.config.get("preprocessing_config", {})

    content = f"""# FedSight AI Reproduction: Deviation & Configuration Report

    ## 1. Configuration Identity
    - **Config Hash (SHA-256):** `{ctx.config_hash}`
    - **Timestamp:** {pd.Timestamp.now().isoformat()}

    ## 2. Implementation Choices (Degrees of Freedom)
    The following parameters were chosen by the implementation team:

    | Parameter | Value | Section |
    | :--- | :--- | :--- |
    | LLM Model | `{llm_config.get('model_name', 'N/A')}` | LLM Engine |
    | Temperature | `{llm_config.get('temperature', 'N/A')}` | LLM Engine |
    | Top P | `{llm_config.get('top_p', 'N/A')}` | LLM Engine |
    | Vote Aggregation | `{sim_config.get('vote_aggregation_rule', 'N/A')}` | Simulation |
    | Tie-Break Rule | `{sim_config.get('tie_break_rule', 'N/A')}` | Simulation |
    | Run Collapse Rule | `{sim_config.get('run_to_meeting_prediction_rule', 'N/A')}` | Simulation |
    | Dot Plot Missing Policy | `{prep_config.get('dot_plot_missing_policy', 'N/A')}` | Preprocessing |

    ## 3. Prompt Template Hashes
    To ensure prompt reproducibility, the following SHA-256 hashes correspond to the templates used:

    """
    templates = ctx.config.get("prompt_templates", {})
    for name, template in templates.items():
        h = hashlib.sha256(template.encode("utf-8")).hexdigest()
        content += f"- **{name}:** `{h}`\n"

    with open(report_path, "w") as f:
        f.write(content)

    return str(report_path)


# ------------------------------------------------------------------------------
# Task 29, Step 3: Validate reproducibility with a second independent run
# ------------------------------------------------------------------------------
def verify_reproducibility(
    original_ctx: 'PipelineContext',
    re_run_func: Callable[..., Dict[str, Any]],
    **kwargs
) -> pd.DataFrame:
    """
    Executes a second run of the pipeline and compares metrics to validate reproducibility.

    Args:
        original_ctx (PipelineContext): State from the first run.
        re_run_func (Callable): Function to execute the pipeline (run_fedsight_pipeline).
        **kwargs: Arguments for re_run_func.

    Returns:
        pd.DataFrame: Comparison of metrics between Run 1 and Run 2.
    """
    print("Starting Reproducibility Check (Run 2)...")

    # Execute Run 2
    res_2 = re_run_func(**kwargs)
    ctx_2 = res_2["context"]

    # Compare Metrics
    metrics_1 = original_ctx.metrics
    metrics_2 = ctx_2.metrics

    if metrics_1 is None or metrics_2 is None:
        return pd.DataFrame()

    # Align and Diff
    # We compare all columns present in both
    common_cols = list(set(metrics_1.columns) & set(metrics_2.columns))
    common_indices = list(set(metrics_1.index) & set(metrics_2.index))

    comparison = []
    for idx in common_indices:
        for col in common_cols:
            val_1 = metrics_1.loc[idx, col]
            val_2 = metrics_2.loc[idx, col]

            # Handle numeric vs string
            try:
                # Remove % if present for float comparison
                v1_clean = float(str(val_1).replace('%', '').replace(',', ''))
                v2_clean = float(str(val_2).replace('%', '').replace(',', ''))
                delta = v1_clean - v2_clean
                match = (delta == 0)
            except:
                delta = "N/A"
                match = (str(val_1) == str(val_2))

            comparison.append({
                "variant": idx,
                "metric": col,
                "run_1": val_1,
                "run_2": val_2,
                "delta": delta,
                "match": match
            })

    return pd.DataFrame(comparison).set_index(["variant", "metric"])


# ------------------------------------------------------------------------------
# Task 29, Orchestrator Function
# ------------------------------------------------------------------------------
def finalize_reproduction_package(
    ctx: 'PipelineContext',
    raw_data_frames: Dict[str, pd.DataFrame],
    output_dir: str = "fedsight_reproduction"
) -> Dict[str, Any]:
    """
    Orchestrates the packaging and documentation of the study.

    Args:
        ctx (PipelineContext): Final pipeline state.
        raw_data_frames (Dict): Input dataframes for archiving.
        output_dir (str): Output path.

    Returns:
        Dict: Paths to generated assets.
    """
    # Step 1: Package
    paths = create_reproducibility_package(ctx, raw_data_frames, output_dir)

    # Step 2: Report
    report_path = generate_deviation_report(ctx, output_dir)
    paths["deviation_report"] = report_path

    return paths
````

````python
# Task 30 — Archive and finalize the study reproduction

# ==============================================================================
# Task 30: Archive and finalize the study reproduction
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 30, Step 1: Create a structured archive with a fixed directory schema
# ------------------------------------------------------------------------------
def validate_archive_structure(
    reproducibility_paths: Dict[str, str],
    output_dir: str = "fedsight_reproduction"
) -> Path:
    """
    Validates that the reproduction directory contains the required schema.

    Schema:
    - config/
    - data/
    - artifacts/
    - results/
    - logs/

    Args:
        reproducibility_paths (Dict): Paths returned by Task 29.
        output_dir (str): Root directory.

    Returns:
        Path: The validated root path.

    Raises:
        FileNotFoundError: If required components are missing.
    """
    root = Path(output_dir)
    if not root.exists():
        raise FileNotFoundError(f"Archive root {root} does not exist.")

    required_subdirs = ["config", "data", "artifacts", "results", "logs"]
    for subdir in required_subdirs:
        path = root / subdir
        if not path.exists() or not path.is_dir():
            raise FileNotFoundError(f"Missing required subdirectory: {subdir}")

    # Verify specific files from the bundle map exist
    for key, path_str in reproducibility_paths.items():
        if not Path(path_str).exists():
            raise FileNotFoundError(f"Missing bundle file for '{key}': {path_str}")

    return root


# ------------------------------------------------------------------------------
# Task 30, Step 2: Compute and store cryptographic hashes
# ------------------------------------------------------------------------------
def generate_archive_checksums(archive_root: Path) -> Path:
    """
    Computes SHA-256 hashes for all files in the archive and stores them in checksums.txt.

    Args:
        archive_root (Path): Root directory.

    Returns:
        Path: Path to the generated checksums.txt file.
    """
    checksums_path = archive_root / "checksums.txt"

    hashes = {}

    # Walk the tree
    for root, _, files in os.walk(archive_root):
        for file in files:
            if file == "checksums.txt":
                continue

            file_path = Path(root) / file
            relative_path = file_path.relative_to(archive_root)

            sha256 = hashlib.sha256()
            with open(file_path, "rb") as f:
                for chunk in iter(lambda: f.read(4096), b""):
                    sha256.update(chunk)

            hashes[str(relative_path)] = sha256.hexdigest()

    # Write sorted by path
    with open(checksums_path, "w") as f:
        f.write("# SHA-256 Checksums for FedSight Reproduction Archive\n")
        for path in sorted(hashes.keys()):
            f.write(f"{hashes[path]}  {path}\n")

    return checksums_path


# ------------------------------------------------------------------------------
# Task 30, Step 3: Finalize and sign off on reproduction completeness
# ------------------------------------------------------------------------------
def sign_off_reproduction(
    ctx: 'PipelineContext',
    checksums_path: Path,
    archive_root: Path
) -> Dict[str, Any]:
    """
    Validates completion criteria and generates a sign-off record.

    Criteria:
    - 3 Variants attempted.
    - Metrics computed.
    - Deviation report exists.
    - Checksums exist.

    Args:
        ctx (PipelineContext): Pipeline state.
        checksums_path (Path): Path to checksums.
        archive_root (Path): Archive root.

    Returns:
        Dict: Completion status record.
    """
    status = {
        "timestamp": pd.Timestamp.now().isoformat(),
        "config_hash": ctx.config_hash,
        "criteria": {},
        "final_verdict": "PENDING"
    }

    # 1. Variants
    # Check metrics index for baseline, icl, cod
    if ctx.metrics is not None:
        found_variants = set(ctx.metrics.index)
        required_variants = {"baseline", "icl", "cod"}
        missing = required_variants - found_variants
        status["criteria"]["variants_complete"] = (len(missing) == 0)
        status["criteria"]["missing_variants"] = list(missing)
    else:
        status["criteria"]["variants_complete"] = False

    # 2. Metrics
    # Check columns
    required_metrics = {
        "total_accuracy", "agent_accuracy", "voting_stability",
        "semantic_similarity", "average_tokens", "mae_bps"
    }
    if ctx.metrics is not None:
        found_metrics = set(ctx.metrics.columns)
        missing_metrics = required_metrics - found_metrics
        status["criteria"]["metrics_complete"] = (len(missing_metrics) == 0)
    else:
        status["criteria"]["metrics_complete"] = False

    # 3. Artifacts
    status["criteria"]["deviation_report_exists"] = (archive_root / "results" / "deviation_report.md").exists()
    status["criteria"]["checksums_exist"] = checksums_path.exists()

    # Verdict
    if all(status["criteria"].values()):
        status["final_verdict"] = "SUCCESS"
    else:
        status["final_verdict"] = "INCOMPLETE"

    # Write Status
    with open(archive_root / "COMPLETION_STATUS.json", "w") as f:
        json.dump(status, f, indent=2)

    return status


# ------------------------------------------------------------------------------
# Task 30, Orchestrator Function
# ------------------------------------------------------------------------------
def finalize_archive(
    ctx: 'PipelineContext',
    reproducibility_paths: Dict[str, str],
    output_dir: str = "fedsight_reproduction"
) -> Dict[str, Any]:
    """
    Orchestrates the final archiving and sign-off.

    Args:
        ctx (PipelineContext): Pipeline state.
        reproducibility_paths (Dict): Paths from Task 29.
        output_dir (str): Root dir.

    Returns:
        Dict: Final status record.
    """
    # Step 1: Validate
    root = validate_archive_structure(reproducibility_paths, output_dir)

    # Step 2: Checksums
    checksums_path = generate_archive_checksums(root)

    # Step 3: Sign-off
    status = sign_off_reproduction(ctx, checksums_path, root)

    return status
````

````python
# Top-Level Orchestrator Callable

# ==============================================================================
# Final Orchestration: Extended Pipeline and Top-Level Lifecycle
# ==============================================================================

# ------------------------------------------------------------------------------
# 1. Extended Pipeline Orchestrator (Tasks 1-28)
# ------------------------------------------------------------------------------
def run_fedsight_pipeline_extended(
    df_macro_raw: pd.DataFrame,
    df_unstructured_raw: pd.DataFrame,
    df_participants_raw: pd.DataFrame,
    df_meeting_outcomes: pd.DataFrame,
    df_fomc_statements_actual: pd.DataFrame,
    config: Dict[str, Any],
    historical_data: Optional[Dict[str, Any]] = None,
    db_path: str = "fedsight_artifacts.db",
    log_path: str = "llm_logs.jsonl"
) -> Dict[str, Any]:
    """
    Orchestrates the complete FedSight AI simulation and evaluation pipeline,
    including all metric computations (Tasks 1-28).

    Args:
        df_macro_raw (pd.DataFrame): Raw macroeconomic indicators.
        df_unstructured_raw (pd.DataFrame): Raw text data.
        df_participants_raw (pd.DataFrame): Participant metadata.
        df_meeting_outcomes (pd.DataFrame): Ground truth decisions.
        df_fomc_statements_actual (pd.DataFrame): Ground truth statements.
        config (Dict): Full study configuration.
        historical_data (Optional[Dict]): Pre-constructed packets for ICL.
        db_path (str): Path for SQLite artifact storage.
        log_path (str): Path for LLM call logs.

    Returns:
        Dict[str, Any]: A dictionary containing:
            - 'artifacts': List of RunArtifact objects.
            - 'metrics': DataFrame of final evaluation metrics.
            - 'ledger': Execution status ledger.
            - 'context': The final PipelineContext object.
            - 'reports': Dictionary of detailed metric reports.
    """
    print("Starting FedSight AI Extended Pipeline...")

    # --- Phase 1: Validation & Setup ---
    print("Phase 1: Validation...")
    validated_config, config_hash = validate_study_config(config)
    data_dict = validate_all_dataframes(
        df_macro_raw, df_unstructured_raw, df_meeting_outcomes,
        df_fomc_statements_actual, df_participants_raw
    )
    ctx = PipelineContext(config=validated_config, config_hash=config_hash)

    # --- Phase 2: Data Cleansing ---
    print("Phase 2: Cleansing...")
    macro_res = cleanse_macro_data(data_dict["df_macro_raw"])
    unstruct_res = cleanse_unstructured_data(data_dict["df_unstructured_raw"])
    eval_res = cleanse_evaluation_data(
        data_dict["df_participants_raw"],
        data_dict["df_meeting_outcomes"],
        data_dict["df_fomc_statements_actual"]
    )

    # --- Phase 3: Packet Construction ---
    print("Phase 3: Packet Construction...")
    canonical_res = construct_canonical_datasets(
        macro_res["df_macro_final"],
        unstruct_res["df_unstructured_clean"],
        eval_res["df_outcomes_clean"],
        eval_res["df_statements_clean"]
    )
    dot_plot_res = preprocess_dot_plots(canonical_res["agent_packets"])
    context_res = preprocess_agent_contexts(dot_plot_res["agent_packets"])

    ctx.agent_packets = context_res["agent_packets"]
    ctx.evaluation_packets = canonical_res["evaluation_packets"]

    # --- Phase 4: Agent Engineering ---
    print("Phase 4: Agent Engineering...")
    cluster_res = cluster_participants(
        eval_res["df_participants_clean"],
        eval_res["participant_feature_spec"]
    )
    ctx.member_registry = materialize_agents(ctx.config)
    ctx.llm_wrapper = setup_llm_infrastructure(ctx.config, log_path)

    # --- Phase 5: Training (ICL) ---
    print("Phase 5: ICL Training...")
    if historical_data:
        ctx.memory_store = run_icl_training(
            ctx.config,
            historical_data["packets"],
            historical_data["outcomes"],
            ctx.member_registry,
            ctx.llm_wrapper
        )
    else:
        ctx.memory_store = []
        print("Skipping ICL: No historical data provided.")

    # --- Phase 6: Simulation ---
    print("Phase 6: Simulation...")
    artifacts, ledger = run_backtesting_campaign(
        ctx.config,
        ctx.agent_packets,
        ctx.member_registry,
        ctx.llm_wrapper,
        ctx.memory_store
    )
    ctx.run_artifacts = artifacts
    ctx.run_ledger = ledger

    # --- Phase 7: Storage ---
    print("Phase 7: Storage...")
    store_simulation_results(ctx.run_artifacts, db_path)

    # --- Phase 8: Evaluation (Tasks 21-28) ---
    print("Phase 8: Evaluation...")

    # Task 21: Meeting Predictions
    pred_res = compute_meeting_predictions(ctx.run_artifacts)

    # Task 22: Total Accuracy
    acc_res = compute_total_accuracy(
        pred_res["meeting_predictions"],
        eval_res["df_outcomes_clean"]
    )

    # Task 23: Agent Accuracy
    agent_acc_res = compute_agent_accuracy(
        ctx.run_artifacts,
        eval_res["df_outcomes_clean"]
    )

    # Task 24: Voting Stability
    stab_res = compute_voting_stability(ctx.run_artifacts)

    # Task 25: Semantic Similarity
    sim_res = compute_semantic_similarity(
        ctx.run_artifacts,
        eval_res["df_statements_clean"],
        ctx.config,
        ctx.llm_wrapper
    )

    # Task 26: Average Tokens
    tok_res = compute_average_tokens(ctx.run_artifacts)

    # Task 27: MAE
    mae_res = compute_mae(
        pred_res["meeting_predictions"],
        eval_res["df_outcomes_clean"]
    )

    # Task 28: Final Report
    metrics_dict = {
        "total_accuracy": acc_res,
        "agent_accuracy": agent_acc_res,
        "voting_stability": stab_res,
        "semantic_similarity": sim_res,
        "average_tokens": tok_res,
        "mae": mae_res
    }

    final_report = generate_final_report(metrics_dict)
    ctx.metrics = final_report["raw_results"]

    print("Pipeline Complete.")

    return {
        "artifacts": ctx.run_artifacts,
        "metrics": ctx.metrics,
        "ledger": ctx.run_ledger,
        "context": ctx,
        "reports": final_report
    }


# ------------------------------------------------------------------------------
# 2. Top-Level Lifecycle Orchestrator (Tasks 1-30)
# ------------------------------------------------------------------------------
def execute_full_study_lifecycle(
    df_macro_raw: pd.DataFrame,
    df_unstructured_raw: pd.DataFrame,
    df_participants_raw: pd.DataFrame,
    df_meeting_outcomes: pd.DataFrame,
    df_fomc_statements_actual: pd.DataFrame,
    config: Dict[str, Any],
    historical_data: Optional[Dict[str, Any]] = None,
    output_dir: str = "fedsight_reproduction",
    db_path: str = "fedsight_artifacts.db",
    log_path: str = "llm_logs.jsonl"
) -> Dict[str, Any]:
    """
    Executes the entire FedSight AI study lifecycle, from data ingestion to
    final reproducibility sign-off.

    Lifecycle:
    1.  **Execution**: Runs the extended pipeline (Tasks 1-28).
    2.  **Packaging**: Bundles artifacts and generates deviation report (Task 29).
    3.  **Archiving**: Validates structure, hashes files, and signs off (Task 30).

    Args:
        df_macro_raw (pd.DataFrame): Raw macro data.
        df_unstructured_raw (pd.DataFrame): Raw narrative data.
        df_participants_raw (pd.DataFrame): Participant data.
        df_meeting_outcomes (pd.DataFrame): Outcomes data.
        df_fomc_statements_actual (pd.DataFrame): Statements data.
        config (Dict): Study configuration.
        historical_data (Optional[Dict]): ICL training data.
        output_dir (str): Directory for reproduction package.
        db_path (str): SQLite DB path.
        log_path (str): Log file path.

    Returns:
        Dict[str, Any]: A dictionary containing:
            - 'pipeline_results': Output from run_fedsight_pipeline_extended.
            - 'reproducibility_paths': Map of packaged file paths.
            - 'final_status': The completion sign-off record.
    """
    # 1. Run Extended Pipeline
    pipeline_results = run_fedsight_pipeline_extended(
        df_macro_raw, df_unstructured_raw, df_participants_raw,
        df_meeting_outcomes, df_fomc_statements_actual,
        config, historical_data, db_path, log_path
    )

    ctx = pipeline_results["context"]

    # Prepare raw dataframes dict for packaging
    raw_data_frames = {
        "macro_raw": df_macro_raw,
        "unstructured_raw": df_unstructured_raw,
        "participants_raw": df_participants_raw,
        "outcomes": df_meeting_outcomes,
        "statements": df_fomc_statements_actual
    }

    # 2. Package Artifacts (Task 29)
    print("Packaging artifacts...")
    repro_paths = finalize_reproduction_package(ctx, raw_data_frames, output_dir)

    # 3. Archive & Sign-off (Task 30)
    print("Finalizing archive...")
    final_status = finalize_archive(ctx, repro_paths, output_dir)

    print(f"Study Lifecycle Complete. Verdict: {final_status['final_verdict']}")

    return {
        "pipeline_results": pipeline_results,
        "reproducibility_paths": repro_paths,
        "final_status": final_status
    }
````
